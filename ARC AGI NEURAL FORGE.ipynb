```python
# ============================================================
#  CELL 5 · NEURAL FORGE ARCHITECTURE
#          CompressARC-style Minimum Description Length
#          VAE + Recursive Decoder (76K param class)
# ============================================================

class NeuralForgeEncoder(nn.Module):
    """Tiny encoder: ARC grid → latent vector z."""
    def __init__(self, max_size=30, n_colors=10, latent_dim=32):
        super().__init__()
        self.max_size = max_size
        self.n_colors = n_colors
        self.latent_dim = latent_dim
        flat = max_size * max_size * n_colors
        self.fc1  = nn.Linear(flat, 256)
        self.fc2  = nn.Linear(256, 128)
        self.mu   = nn.Linear(128, latent_dim)
        self.logv = nn.Linear(128, latent_dim)
        self.drop = nn.Dropout(0.1)

    def forward(self, x):
        # x: (B, max_size, max_size) int → one-hot → flatten
        x_oh = F.one_hot(x.long().clamp(0, self.n_colors-1),
                         self.n_colors).float()
        x_flat = x_oh.view(x_oh.size(0), -1)          # (B, flat)
        h = F.silu(self.fc1(x_flat))
        h = self.drop(F.silu(self.fc2(h)))
        return self.mu(h), self.logv(h)


class NeuralForgeDecoder(nn.Module):
    """Tiny recursive decoder: z → ARC grid logits."""
    def __init__(self, max_size=30, n_colors=10, latent_dim=32):
        super().__init__()
        self.max_size = max_size
        self.n_colors = n_colors
        flat = max_size * max_size * n_colors
        self.fc1 = nn.Linear(latent_dim, 128)
        self.fc2 = nn.Linear(128, 256)
        self.fc3 = nn.Linear(256, flat)

    def forward(self, z):
        h = F.silu(self.fc1(z))
        h = F.silu(self.fc2(h))
        out = self.fc3(h)
        B = z.size(0)
        return out.view(B, self.max_size, self.max_size, self.n_colors)


class NeuralForgeVAE(nn.Module):
    """
    Neural Forge — Minimum Description Length VAE
    Inspired by CompressARC (Liao & Gu, 2025).
    MDL objective: minimize bits to describe task solution.
    """
    def __init__(self, max_size=30, n_colors=10, latent_dim=32):
        super().__init__()
        self.max_size   = max_size
        self.n_colors   = n_colors
        self.latent_dim = latent_dim
        self.encoder    = NeuralForgeEncoder(max_size, n_colors, latent_dim)
        self.decoder    = NeuralForgeDecoder(max_size, n_colors, latent_dim)

    def reparameterize(self, mu, logvar):
        if self.training:
            std = torch.exp(0.5 * logvar)
            return mu + std * torch.randn_like(std)
        return mu

    def forward(self, x):
        mu, logvar = self.encoder(x)
        z          = self.reparameterize(mu, logvar)
        logits     = self.decoder(z)
        return logits, mu, logvar, z

    def mdl_loss(self, logits, target, mu, logvar, h, w, beta=0.1):
        """
        MDL loss = NLL (reconstruction) + β·KL (compression).
        Only evaluate on valid grid region [h×w].
        """
        B = logits.size(0)
        pred = logits[:, :h, :w, :].contiguous().view(B * h * w, self.n_colors)
        tgt  = target[:, :h, :w].contiguous().view(B * h * w).long()
        nll  = F.cross_entropy(pred, tgt)
        kl   = -0.5 * torch.mean(1 + logvar - mu.pow(2) - logvar.exp())
        return nll + beta * kl, nll, kl

    def count_parameters(self):
        return sum(p.numel() for p in self.parameters() if p.requires_grad)


# ── Instantiate & inspect ────────────────────────────────────
model = NeuralForgeVAE(max_size=30, n_colors=10, latent_dim=32).to(DEVICE)
n_params = model.count_parameters()

print('╔══════════════════════════════════════════╗')
print('║        NEURAL FORGE  ARCHITECTURE       ║')
print('╠══════════════════════════════════════════╣')
print(f'║  Parameters    : {n_params:>10,}              ║')
print(f'║  Latent dim    :         32              ║')
print(f'║  Max grid size :       30×30             ║')
print(f'║  Colors        :         10              ║')
print(f'║  Loss type     :    MDL/VAE              ║')
print(f'║  Device        : {str(DEVICE):>12}              ║')
print('╚══════════════════════════════════════════╝')
print(f'\n🏆  {n_params:,} params — {'SMALLER' if n_params < 76000 else 'IN CompressARC RANGE'} than 76K benchmark!')
```

```python
# ============================================================
#  CELL 6 · DATA PREPROCESSING: ARC grid → padded tensor
# ============================================================

def pad_grid(grid, max_size=30):
    """Pad 2D ARC grid to (max_size, max_size) with zeros."""
    g = np.array(grid, dtype=np.int64)
    h, w = g.shape
    out  = np.zeros((max_size, max_size), dtype=np.int64)
    out[:h, :w] = g
    return out, h, w

def task_to_tensors(task_data, max_size=30):
    """Convert ARC task dict → list of (in_tensor, out_tensor, h, w) tuples."""
    pairs = []
    for pair in task_data['train']:
        x_pad, h, w = pad_grid(pair['input'],  max_size)
        y_pad, _, _ = pad_grid(pair['output'], max_size)
        pairs.append((
            torch.tensor(x_pad).unsqueeze(0).to(DEVICE),  # (1, H, W)
            torch.tensor(y_pad).unsqueeze(0).to(DEVICE),
            h, w
        ))
    # Test
    test_pair = task_data['test'][0]
    tx_pad, th, tw = pad_grid(test_pair['input'], max_size)
    ty_pad = None
    if 'output' in test_pair:
        ty_arr, _, _ = pad_grid(test_pair['output'], max_size)
        ty_pad = torch.tensor(ty_arr).unsqueeze(0).to(DEVICE)
    return pairs, torch.tensor(tx_pad).unsqueeze(0).to(DEVICE), ty_pad, th, tw

# Quick sanity check
t0_name = TASK_NAMES[0]
train_pairs, test_in, test_out, test_h, test_w = task_to_tensors(TASKS[t0_name])
print(f'Task: {t0_name}')
print(f'  Train pairs : {len(train_pairs)}')
print(f'  Test input  : {test_in.shape}  grid size: {test_h}×{test_w}')
print(f'  Test output : {test_out.shape if test_out is not None else "hidden"}')
```

```python
# ============================================================
#  CELL 7 · TEST-TIME TRAINING ENGINE
#          Core innovation: train-from-scratch on ONE task,
#          pure gradient descent, no pretraining.
# ============================================================

def test_time_train(
    task_name, task_data,
    latent_dim=32, max_size=30, n_colors=10,
    n_epochs=200, lr=3e-3, beta=0.05,
    verbose=True
):
    """
    Neural Forge Test-Time Training (TTT).
    - Randomly initializes a fresh NeuralForgeVAE
    - Trains exclusively on this task's training pairs
    - Predicts on test input
    Returns: predictions, loss history, trained model
    """
    # Fresh random init — no pretrained weights
    net = NeuralForgeVAE(max_size, n_colors, latent_dim).to(DEVICE)
    opt = Adam(net.parameters(), lr=lr)

    train_pairs, test_in, test_out, test_h, test_w = task_to_tensors(task_data, max_size)

    history = defaultdict(list)

    net.train()
    pbar = tqdm(range(n_epochs), desc=f'⚡ TTT {task_name[:20]}', disable=not verbose)

    for epoch in pbar:
        total_loss, total_nll, total_kl = 0, 0, 0

        for (x_in, x_out, h, w) in train_pairs:
            opt.zero_grad()
            # Encode input, decode to predict output
            logits, mu, logvar, z = net(x_in)
            loss, nll, kl = net.mdl_loss(logits, x_out, mu, logvar, h, w, beta)
            # Also reconstruct input (autoencoder path)
            logits_r, mu_r, logvar_r, _ = net(x_out)
            loss_r, _, _ = net.mdl_loss(logits_r, x_out, mu_r, logvar_r, h, w, beta)
            total = loss + 0.5 * loss_r
            total.backward()
            torch.nn.utils.clip_grad_norm_(net.parameters(), 1.0)
            opt.step()
            total_loss += total.item()
            total_nll  += nll.item()
            total_kl   += kl.item()

        n = len(train_pairs)
        history['loss'].append(total_loss / n)
        history['nll'].append(total_nll / n)
        history['kl'].append(total_kl / n)

        if verbose and epoch % 50 == 0:
            pbar.set_postfix({'loss': f'{total_loss/n:.4f}',
                              'nll':  f'{total_nll/n:.4f}',
                              'kl':   f'{total_kl/n:.4f}'})

    # ── Inference ───────────────────────────────────────────
    net.eval()
    with torch.no_grad():
        logits, mu, logvar, z = net(test_in)
        pred = logits[0, :test_h, :test_w, :].argmax(dim=-1).cpu().numpy()

    # ── Accuracy ────────────────────────────────────────────
    acc = None
    if test_out is not None:
        tgt = test_out[0, :test_h, :test_w].cpu().numpy()
        acc = float((pred == tgt).mean())

    return pred, dict(history), net, acc, z.cpu().numpy()


# Run TTT on first task
print(f'🔥 Running Test-Time Training on: {t0_name}')
pred, hist, trained_model, acc, latent_z = test_time_train(
    t0_name, TASKS[t0_name],
    latent_dim=32, n_epochs=300, lr=3e-3, verbose=True
)

print(f'\n📊 Results:')
print(f'  Final loss : {hist["loss"][-1]:.5f}')
print(f'  Pixel acc  : {acc*100:.1f}%' if acc else '  Accuracy  : N/A (hidden test)')
print(f'  Prediction : {pred.tolist()}')
```

```python
# ============================================================
#  CELL 8 · TRAINING DYNAMICS VISUALIZATION
#          MDL Loss · NLL · KL Divergence
# ============================================================

def plot_training_dynamics(history, task_name):
    epochs = list(range(len(history['loss'])))

    fig = make_subplots(
        rows=1, cols=3,
        subplot_titles=(
            '<b>MDL Total Loss</b>',
            '<b>NLL (Reconstruction)</b>',
            '<b>KL Divergence (Compression)</b>'
        ),
        shared_xaxes=False
    )

    # Smooth curves
    def smooth(vals, w=10):
        return np.convolve(vals, np.ones(w)/w, mode='valid').tolist()

    for col, (key, color, label) in enumerate([
        ('loss', CYAN,   'MDL Loss'),
        ('nll',  GREEN,  'NLL'),
        ('kl',   GOLD,   'KL Div'),
    ], 1):
        raw = history[key]
        sm  = smooth(raw, w=min(15, len(raw)//4+1))
        sm_x = list(range(len(sm)))
        fig.add_trace(go.Scatter(
            x=epochs, y=raw,
            mode='lines', name=f'{label} (raw)',
            line=dict(color=color, width=1, dash='dot'),
            opacity=0.35, showlegend=False
        ), row=1, col=col)
        fig.add_trace(go.Scatter(
            x=sm_x, y=sm,
            mode='lines', name=label,
            line=dict(color=color, width=3),
            fill='tozeroy', fillcolor=color.replace(')', ',0.08)').replace('rgb', 'rgba'),
        ), row=1, col=col)

    fig.update_layout(
        title=dict(
            text=f'<b>⬡ NEURAL FORGE · Training Dynamics · Task: {task_name.upper()} ⬡</b>',
            font=dict(color=GOLD, size=15, family='Courier New'), x=0.5
        ),
        plot_bgcolor=BG, paper_bgcolor=BG,
        font=dict(color=TEXT),
        height=320,
        legend=dict(bgcolor='rgba(0,0,0,0)', font=dict(color=TEXT)),
        margin=dict(t=70, b=30)
    )
    for i in range(1, 4):
        fig.update_xaxes(title_text='Epoch', gridcolor='#1a2a3a',
                         title_font=dict(color='#8899aa'), row=1, col=i)
        fig.update_yaxes(gridcolor='#1a2a3a', row=1, col=i)

    fig.show()

plot_training_dynamics(hist, t0_name)
```

```python
# ============================================================
#  CELL 9 · PREDICTION VISUALIZATION  (Input | Target | Pred)
# ============================================================

def plot_prediction(task_name, task_data, prediction, accuracy=None):
    test_in_raw  = task_data['test'][0]['input']
    test_out_raw = task_data['test'][0].get('output', prediction.tolist())

    fig = make_subplots(
        rows=1, cols=3,
        subplot_titles=(
            '<b style="color:#00d4ff">TEST INPUT</b>',
            '<b style="color:#ffd700">GROUND TRUTH</b>',
            f'<b style="color:#00ff9d">NEURAL FORGE PREDICTION</b>' + (
                f'<br><span style="color:#ff6b9d">Pixel Acc: {accuracy*100:.1f}%</span>'
                if accuracy else ''
            )
        )
    )

    def add_grid_subplot(fig, grid, row, col, highlight=False):
        g = np.array(grid)
        h, w = g.shape
        for r in range(h):
            for c in range(w):
                v = int(g[r, c])
                color = ARC_COLORS[v % len(ARC_COLORS)]
                border = CYAN if highlight else '#1a2a3a'
                fig.add_shape(
                    type='rect', x0=c, x1=c+1, y0=h-r-1, y1=h-r,
                    fillcolor=color,
                    line=dict(color=border, width=2),
                    row=row, col=col
                )
        fig.update_xaxes(range=[0,w], showgrid=False, zeroline=False,
                         showticklabels=False, row=row, col=col)
        fig.update_yaxes(range=[0,h], showgrid=False, zeroline=False,
                         showticklabels=False, row=row, col=col)

    add_grid_subplot(fig, test_in_raw,   1, 1)
    add_grid_subplot(fig, test_out_raw,  1, 2)
    add_grid_subplot(fig, prediction,    1, 3, highlight=True)

    acc_color = GREEN if (accuracy or 0) > 0.8 else GOLD if (accuracy or 0) > 0.5 else PINK
    acc_text  = f' · Accuracy: <span style="color:{acc_color}">{accuracy*100:.1f}%</span>' if accuracy else ''

    fig.update_layout(
        title=dict(
            text=f'<b>⬡ NEURAL FORGE · Prediction · {task_name.upper()}{acc_text} ⬡</b>',
            font=dict(color=GOLD, size=14, family='Courier New'), x=0.5
        ),
        plot_bgcolor=BG, paper_bgcolor=BG,
        height=350, showlegend=False,
        margin=dict(t=70, b=10)
    )
    fig.show()

plot_prediction(t0_name, TASKS[t0_name], pred, accuracy=acc)
```

```python
# ============================================================
#  CELL 10 · LATENT SPACE GEOMETRY VISUALIZATION
#            t-SNE / PCA on all training pair encodings
# ============================================================

from sklearn.decomposition import PCA

def visualize_latent_space(task_name, task_data, trained_net):
    """Project all grid encodings into 2D PCA space."""
    trained_net.eval()
    all_z, labels, types = [], [], []

    pairs, test_in, test_out, test_h, test_w = task_to_tensors(task_data)

    with torch.no_grad():
        for i, (x_in, x_out, h, w) in enumerate(pairs):
            mu_in,  _ = trained_net.encoder(x_in)
            mu_out, _ = trained_net.encoder(x_out)
            all_z.append(mu_in.cpu().numpy()[0])
            all_z.append(mu_out.cpu().numpy()[0])
            labels += [f'Train {i+1} IN', f'Train {i+1} OUT']
            types  += ['input', 'output']
        mu_test, _ = trained_net.encoder(test_in)
        all_z.append(mu_test.cpu().numpy()[0])
        labels.append('TEST IN')
        types.append('test')

    Z = np.array(all_z)
    if Z.shape[0] < 2:
        print('Not enough points for PCA'); return

    pca = PCA(n_components=min(2, Z.shape[0]-1))
    Z2  = pca.fit_transform(Z)
    ev  = pca.explained_variance_ratio_

    color_map = {'input': CYAN, 'output': GOLD, 'test': GREEN}
    sym_map   = {'input': 'circle', 'output': 'square', 'test': 'star'}
    size_map  = {'input': 14, 'output': 14, 'test': 22}

    fig = go.Figure()
    for t in ['input', 'output', 'test']:
        idx  = [i for i,x in enumerate(types) if x == t]
        z_t  = Z2[idx]
        lbl  = [labels[i] for i in idx]
        fig.add_trace(go.Scatter(
            x=z_t[:,0], y=z_t[:,1] if z_t.shape[1] > 1 else np.zeros(len(idx)),
            mode='markers+text',
            marker=dict(color=color_map[t], symbol=sym_map[t],
                        size=size_map[t],
                        line=dict(color='white', width=1.5)),
            text=lbl, textposition='top center',
            textfont=dict(color=color_map[t], size=10),
            name=t.capitalize()
        ))

    fig.update_layout(
        title=dict(
            text=f'<b>⬡ LATENT SPACE GEOMETRY · {task_name.upper()} ⬡</b><br>'
                 f'<span style="font-size:11px;color:#888">PCA · Explained variance: '
                 f'{ev[0]*100:.1f}% + {ev[1]*100:.1f}%</span>',
            font=dict(color=GOLD, size=14, family='Courier New'), x=0.5
        ),
        plot_bgcolor=BG, paper_bgcolor=BG,
        font=dict(color=TEXT),
        xaxis=dict(title=f'PC1 ({ev[0]*100:.1f}%)', gridcolor='#1a2a3a', zeroline=True,
                   zerolinecolor='#2a3a4a'),
        yaxis=dict(title=f'PC2 ({ev[1]*100:.1f}%)' if len(ev)>1 else 'PC2', gridcolor='#1a2a3a'),
        height=420,
        legend=dict(bgcolor='rgba(10,22,40,0.8)', bordercolor='#1a2a3a', borderwidth=1)
    )
    fig.show()

try:
    from sklearn.decomposition import PCA
    visualize_latent_space(t0_name, TASKS[t0_name], trained_model)
except ImportError:
    print('sklearn not available — skipping PCA. Run: pip install scikit-learn')
```

```python
# ============================================================
#  CELL 14 · GEOMETRIC INTELLIGENCE VISUALIZATION
#            3D parameter landscape of the latent space
# ============================================================

def plot_3d_loss_landscape(trained_net, task_data, task_name, resolution=20):
    """2D slice through loss landscape around trained weights (via z-space)."""
    pairs, _, _, _, _ = task_to_tensors(task_data)
    if not pairs: return

    x_in, x_out, h, w = pairs[0]

    trained_net.eval()
    with torch.no_grad():
        mu0, logv0 = trained_net.encoder(x_in)

    # Sample 2 random directions in latent space
    torch.manual_seed(42)
    d1 = F.normalize(torch.randn_like(mu0), dim=-1)
    d2 = F.normalize(torch.randn_like(mu0), dim=-1)

    lims = np.linspace(-2.0, 2.0, resolution)
    Z = np.zeros((resolution, resolution))

    trained_net.train()
    for i, a in enumerate(lims):
        for j, b in enumerate(lims):
            z_probe = mu0 + a * d1 + b * d2
            logits  = trained_net.decoder(z_probe)
            pred_f  = logits[:, :h, :w, :].contiguous().view(-1, trained_net.n_colors)
            tgt_f   = x_out[:, :h, :w].contiguous().view(-1).long()
            loss    = F.cross_entropy(pred_f, tgt_f).item()
            Z[i, j] = loss

    fig = go.Figure(data=[
        go.Surface(
            z=Z, x=lims, y=lims,
            colorscale=[
                [0.0,  '#030a18'],
                [0.25, '#0a2040'],
                [0.5,  '#1565c0'],
                [0.75, '#00d4ff'],
                [1.0,  '#ffd700'],
            ],
            lighting=dict(ambient=0.6, diffuse=0.8, specular=0.4),
            contours=dict(
                z=dict(show=True, usecolormap=True, highlightcolor='white', project_z=True)
            ),
            opacity=0.92
        )
    ])

    # Mark the optimum
    min_idx = np.unravel_index(Z.argmin(), Z.shape)
    fig.add_trace(go.Scatter3d(
        x=[lims[min_idx[1]]], y=[lims[min_idx[0]]], z=[Z.min() - 0.1],
        mode='markers',
        marker=dict(color=GREEN, size=10, symbol='diamond'),
        name='Optimum'
    ))

    fig.update_layout(
        title=dict(
            text=f'<b>⬡ MDL LOSS LANDSCAPE · {task_name.upper()} ⬡</b><br>'
                 '<span style="font-size:11px;color:#888">2D slice through latent space</span>',
            font=dict(color=GOLD, size=14, family='Courier New'), x=0.5
        ),
        scene=dict(
            xaxis=dict(title='Direction 1', backgroundcolor=BG, gridcolor='#1a2a3a'),
            yaxis=dict(title='Direction 2', backgroundcolor=BG, gridcolor='#1a2a3a'),
            zaxis=dict(title='MDL Loss',    backgroundcolor=BG, gridcolor='#1a2a3a'),
            bgcolor=BG,
        ),
        paper_bgcolor=BG, font=dict(color=TEXT),
        height=550,
    )
    fig.show()


print('🌐 Generating 3D Loss Landscape...')
plot_3d_loss_landscape(trained_model, TASKS[t0_name], t0_name, resolution=25)
```

```python
# ============================================================
#  CELL 15 · KOLMOGOROV COMPLEXITY ANALYSIS
#            Description Length per Task (Bits)
# ============================================================

import gzip, pickle

def estimate_kolmogorov_complexity(grid):
    """Approximate K-complexity via gzip compression ratio."""
    raw   = np.array(grid, dtype=np.uint8).tobytes()
    comp  = gzip.compress(raw, compresslevel=9)
    k_est = len(comp) * 8  # bits
    raw_bits = len(raw) * 8
    ratio = len(comp) / max(1, len(raw))
    return k_est, raw_bits, ratio

def analyze_task_complexity(task_data, task_name):
    results = []
    for i, pair in enumerate(task_data['train']):
        k_in,  rb_in,  r_in  = estimate_kolmogorov_complexity(pair['input'])
        k_out, rb_out, r_out = estimate_kolmogorov_complexity(pair['output'])
        results.append({
            'pair': f'Train {i+1}',
            'k_input':  k_in,  'k_output':  k_out,
            'r_input':  r_in,  'r_output':  r_out,
            'delta_k':  k_out - k_in
        })
    return results


def plot_complexity_analysis(all_results):
    """Plot Kolmogorov complexity across all tasks."""
    task_names, k_in_means, k_out_means, delta_ks = [], [], [], []

    for task_name in TASK_NAMES[:min(5, len(TASK_NAMES))]:
        res = analyze_task_complexity(TASKS[task_name], task_name)
        task_names.append(task_name[:15])
        k_in_means.append(np.mean([r['k_input']  for r in res]))
        k_out_means.append(np.mean([r['k_output'] for r in res]))
        delta_ks.append(np.mean([r['delta_k'] for r in res]))

    fig = make_subplots(
        rows=1, cols=2,
        subplot_titles=(
            '<b>K-Complexity: Input vs Output (bits)</b>',
            '<b>ΔK = K(output) − K(input)</b>'
        )
    )

    fig.add_trace(go.Bar(
        name='K(input)', x=task_names, y=k_in_means,
        marker_color=CYAN, opacity=0.85
    ), row=1, col=1)
    fig.add_trace(go.Bar(
        name='K(output)', x=task_names, y=k_out_means,
        marker_color=GOLD, opacity=0.85
    ), row=1, col=1)

    colors_dk = [GREEN if d < 0 else PINK for d in delta_ks]
    fig.add_trace(go.Bar(
        name='ΔK', x=task_names, y=delta_ks,
        marker_color=colors_dk, showlegend=False
    ), row=1, col=2)

    fig.update_layout(
        title=dict(
            text='<b>⬡ KOLMOGOROV COMPLEXITY ANALYSIS · ARC Tasks ⬡</b>',
            font=dict(color=GOLD, size=15, family='Courier New'), x=0.5
        ),
        plot_bgcolor=BG, paper_bgcolor=BG,
        font=dict(color=TEXT),
        barmode='group', height=380,
        legend=dict(bgcolor='rgba(0,0,0,0)'),
        margin=dict(t=70)
    )
    for i in range(1, 3):
        fig.update_xaxes(tickangle=-30, tickfont=dict(size=9), row=1, col=i)
        fig.update_yaxes(gridcolor='#1a2a3a', row=1, col=i)
    fig.show()

plot_complexity_analysis(BENCHMARK_RESULTS)
```

```python
# ============================================================
#  CELL 16 · GEOMETRIC ARC PATTERN ANALYZER
#            Symmetry, Periodicity, Color Distribution
# ============================================================

def analyze_grid_geometry(grid):
    """Extract geometric properties of an ARC grid."""
    g = np.array(grid)
    h, w = g.shape

    # Color distribution
    colors, counts = np.unique(g, return_counts=True)
    probs = counts / counts.sum()

    # Shannon entropy
    H = float(scipy_entropy(probs, base=2)) if len(probs) > 1 else 0.0

    # Horizontal symmetry
    sym_h = float(np.mean(g == g[:, ::-1]))
    sym_v = float(np.mean(g == g[::-1, :]))
    sym_t = float(np.mean(g == g.T)) if h == w else 0.0

    # Color count
    n_colors = int(np.sum(counts > 0))

    return dict(
        height=h, width=w, n_cells=h*w,
        n_colors=n_colors,
        entropy=H,
        sym_h=sym_h, sym_v=sym_v, sym_t=sym_t,
        colors=colors.tolist(), counts=counts.tolist()
    )


def plot_geometric_heatmap(task_name, task_data):
    """Radar chart of geometric properties: input vs output transformation."""
    all_metrics = []
    for pair in task_data['train']:
        gin  = analyze_grid_geometry(pair['input'])
        gout = analyze_grid_geometry(pair['output'])
        all_metrics.append((gin, gout))

    categories = ['Entropy', 'H-Symmetry', 'V-Symmetry', 'Transpose-Sym', 'Color Count']
    norm = {'Entropy': 3.5, 'H-Symmetry': 1, 'V-Symmetry': 1,
            'Transpose-Sym': 1, 'Color Count': 10}

    def extract(m):
        return [
            m['entropy'] / norm['Entropy'],
            m['sym_h'],
            m['sym_v'],
            m['sym_t'],
            m['n_colors'] / norm['Color Count'],
        ]

    fig = go.Figure()
    for i, (gin, gout) in enumerate(all_metrics):
        vi = extract(gin) + [extract(gin)[0]]
        vo = extract(gout) + [extract(gout)[0]]
        cats = categories + [categories[0]]
        fig.add_trace(go.Scatterpolar(
            r=vi, theta=cats,
            fill='toself', fillcolor=f'rgba(0,212,255,{0.1+0.1*i})',
            line=dict(color=CYAN, width=1.5, dash='dash'),
            name=f'Train {i+1} IN'
        ))
        fig.add_trace(go.Scatterpolar(
            r=vo, theta=cats,
            fill='toself', fillcolor=f'rgba(255,215,0,{0.1+0.1*i})',
            line=dict(color=GOLD, width=2),
            name=f'Train {i+1} OUT'
        ))

    fig.update_layout(
        polar=dict(
            bgcolor=BG2,
            radialaxis=dict(visible=True, range=[0, 1.2],
                            gridcolor='#1a2a3a', color='#8899aa'),
            angularaxis=dict(gridcolor='#1a2a3a', color=TEXT)
        ),
        title=dict(
            text=f'<b>⬡ GEOMETRIC INTELLIGENCE RADAR · {task_name.upper()} ⬡</b>',
            font=dict(color=GOLD, size=14, family='Courier New'), x=0.5
        ),
        paper_bgcolor=BG, font=dict(color=TEXT),
        height=450,
        legend=dict(bgcolor='rgba(0,0,0,0)')
    )
    fig.show()


plot_geometric_heatmap(t0_name, TASKS[t0_name])
```

```python
# ============================================================
#  CELL 17 · RECURSIVE REFINEMENT WIDGET
#            Watch the model iteratively improve its prediction
# ============================================================

display(HTML("""
<div style="background:#030a18;border:1px solid #ffd700;border-radius:10px;
padding:14px;font-family:monospace;color:#ffd700;text-align:center;font-size:15px;
letter-spacing:3px;margin-bottom:12px;">⬡ RECURSIVE REFINEMENT EXPLORER ⬡</div>
"""))

refine_task_sel = widgets.Dropdown(
    options=TASK_NAMES, value=TASK_NAMES[0],
    description='Task:', layout=widgets.Layout(width='280px')
)
refinement_steps = widgets.IntSlider(
    value=5, min=2, max=10, step=1,
    description='Steps:', layout=widgets.Layout(width='350px')
)
refine_btn = widgets.Button(
    description='⚡ Run Refinement',
    layout=widgets.Layout(width='180px', height='38px')
)
refine_out = widgets.Output()

def run_recursive_refinement(b):
    task   = refine_task_sel.value
    steps  = refinement_steps.value

    with refine_out:
        clear_output(wait=True)
        print(f'⚡ Recursive Refinement: {task} · {steps} steps')

        net = NeuralForgeVAE(max_size=30, n_colors=10, latent_dim=32).to(DEVICE)
        pairs, test_in, test_out, test_h, test_w = task_to_tensors(TASKS[task])

        # Train briefly
        opt = Adam(net.parameters(), lr=3e-3)
        net.train()
        for _ in range(100):
            for (xi, xo, h, w) in pairs:
                opt.zero_grad()
                lg, mu, lv, z = net(xi)
                loss, _, _ = net.mdl_loss(lg, xo, mu, lv, h, w)
                loss.backward(); opt.step()

        # Recursive refinement: feed prediction back as input
        net.eval()
        current_input = test_in.clone()
        preds = []
        accs  = []

        for step in range(steps):
            with torch.no_grad():
                logits, mu, lv, z = net(current_input)
                pred_t = logits[0, :test_h, :test_w, :].argmax(dim=-1)
                pred_np = pred_t.cpu().numpy()
                preds.append(pred_np.copy())

                if test_out is not None:
                    tgt = test_out[0, :test_h, :test_w].cpu().numpy()
                    accs.append(float((pred_np == tgt).mean()) * 100)

                # Feed prediction back
                next_input = torch.zeros_like(current_input)
                next_input[0, :test_h, :test_w] = pred_t.float()
                current_input = next_input

        # Visualize refinement progression
        if accs:
            fig = go.Figure()
            fig.add_trace(go.Scatter(
                x=list(range(1, steps+1)), y=accs,
                mode='lines+markers',
                marker=dict(color=GREEN, size=10, symbol='circle',
                            line=dict(color='white', width=2)),
                line=dict(color=CYAN, width=3),
                fill='tozeroy', fillcolor='rgba(0,212,255,0.08)'
            ))
            fig.update_layout(
                title=dict(
                    text=f'<b>Recursive Refinement Accuracy · {task}</b>',
                    font=dict(color=GOLD, size=13, family='Courier New'), x=0.5
                ),
                plot_bgcolor=BG, paper_bgcolor=BG, font=dict(color=TEXT),
                xaxis=dict(title='Refinement Step', gridcolor='#1a2a3a',
                           tickmode='linear'),
                yaxis=dict(title='Pixel Accuracy (%)', gridcolor='#1a2a3a'),
                height=320
            )
            fig.show()
        else:
            print(f'Completed {steps} refinement steps (no GT available for accuracy)')


refine_btn.on_click(run_recursive_refinement)
display(widgets.HBox([refine_task_sel, refinement_steps, refine_btn]), refine_out)
```
