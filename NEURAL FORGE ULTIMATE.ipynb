```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 04 ─ CBR META-ROUTER  (THE REVOLUTIONARY CORE)
#
#  Idea: Before running any engine, fingerprint the task (44 dims),
#  find K similar solved cases from the KB, and use their outcomes
#  to PREDICT which engine will win + how much time to allocate.
#
#  This turns brute-force "try everything" into intelligent routing.
# ╚══════════════════════════════════════════════════════════════════╝
FP_DIM=44

def fingerprint(task):
    pairs=task.get('train',[]); feats=[]
    try:
        in_h =np.mean([len(p['input'])     for p in pairs]) if pairs else 5.
        in_w =np.mean([len(p['input'][0])  for p in pairs]) if pairs else 5.
        out_h=np.mean([len(p['output'])    for p in pairs]) if pairs else 5.
        out_w=np.mean([len(p['output'][0]) for p in pairs]) if pairs else 5.
        feats+=[in_h/30.,in_w/30.,out_h/30.,out_w/30.,
                float(in_h==out_h),float(in_w==out_w),
                (out_h*out_w)/max(1.,in_h*in_w),
                abs(out_h-in_h)/30.,abs(out_w-in_w)/30.]
        ic=[set(np.array(p['input']).flatten().tolist())  for p in pairs]
        oc=[set(np.array(p['output']).flatten().tolist()) for p in pairs]
        nic=np.mean([len(c) for c in ic]) if ic else 1.
        noc=np.mean([len(c) for c in oc]) if oc else 1.
        itr=np.mean([len(a&b)/max(1,len(a|b)) for a,b in zip(ic,oc)]) if pairs else .5
        feats+=[nic/10.,noc/10.,(noc-nic+9)/18.,float(nic==noc),itr,float(noc==1.),float(any(0 in c for c in ic))]
        def sym(g):
            h,w=g.shape
            return (float(np.mean(g==g[:,::-1])),float(np.mean(g==g[::-1,:])),
                    float(np.mean(g==g.T)) if h==w else 0.)
        is_=[sym(np.array(p['input'],dtype=np.int64))  for p in pairs]
        os_=[sym(np.array(p['output'],dtype=np.int64)) for p in pairs]
        feats+=[np.mean([s[i] for s in is_]) if is_ else 0. for i in range(3)]
        feats+=[np.mean([s[i] for s in os_]) if os_ else 0. for i in range(3)]
        feats+=[float(np.mean([s[0] for s in os_])>.8) if os_ else 0.,
                float(np.mean([s[1] for s in os_])>.8) if os_ else 0.]
        def dns(g): return float((np.array(g,dtype=np.int64)!=0).mean())
        def nob(g): return min(1.,sum(ndlabel(np.array(g,dtype=np.int64)==c)[1]
                                     for c in np.unique(np.array(g,dtype=np.int64)) if c!=0)/20.)
        feats+=[np.mean([dns(p['input']) for p in pairs])  if pairs else 0.,
                np.mean([dns(p['output'])for p in pairs])  if pairs else 0.,
                np.mean([nob(p['input']) for p in pairs])  if pairs else 0.,
                np.mean([nob(p['output'])for p in pairs])  if pairs else 0.]
        def pc(p):
            gi=np.array(p['input'],dtype=np.int64); go=np.array(p['output'],dtype=np.int64)
            return float((gi!=go).mean()) if gi.shape==go.shape else .5
        def ir(p):
            gi=np.array(p['input'],dtype=np.int64); go=np.array(p['output'],dtype=np.int64)
            return float(any(np.rot90(gi,k).shape==go.shape and float((np.rot90(gi,k)==go).mean())>.95 for k in[1,2,3]))
        def ifl(p):
            gi=np.array(p['input'],dtype=np.int64); go=np.array(p['output'],dtype=np.int64)
            return float(gi.shape==go.shape and (float((np.fliplr(gi)==go).mean())>.95 or float((np.flipud(gi)==go).mean())>.95))
        def ico(p):
            gi=np.array(p['input'],dtype=np.int64); go=np.array(p['output'],dtype=np.int64)
            return float(gi.shape==go.shape and bool((gi!=0)==(go!=0)) and float((gi==go).mean())<.95)
        pca=float(np.mean([pc(p) for p in pairs])) if pairs else 0.
        feats+=[pca,float(np.mean([ir(p) for p in pairs])) if pairs else 0.,
                float(np.mean([ifl(p)for p in pairs])) if pairs else 0.,
                float(np.mean([ico(p)for p in pairs])) if pairs else 0.,
                float(pca<.1),float(pca>.8),len(pairs)/10.,
                in_h/max(1.,in_w),out_h/max(1.,out_w)]
    except: pass
    feats=[float(x) for x in feats]
    feats+=[0.]*(FP_DIM-len(feats)); feats=feats[:FP_DIM]
    return np.array(feats,dtype=np.float32)

# Pre-compute KB fingerprints
print("🔬 Fingerprinting KB...")
KB_FPS={}
for tid,task in tqdm(KB.items(),desc="KB",ncols=70,leave=False):
    try: KB_FPS[tid]=fingerprint(task)
    except: KB_FPS[tid]=np.zeros(FP_DIM,dtype=np.float32)
KB_IDS=list(KB_FPS.keys())
KB_MAT=np.stack([KB_FPS[t] for t in KB_IDS],axis=0)

print("🔬 Fingerprinting competition tasks...")
TASK_FPS={}
for tid in tqdm(TASK_IDS,desc="Tasks",ncols=70,leave=False):
    try: TASK_FPS[tid]=fingerprint(TASKS[tid])
    except: TASK_FPS[tid]=np.zeros(FP_DIM,dtype=np.float32)

# CBR Router: returns engine priority list + time weights
def cbr_route(tid,K=12):
    q=TASK_FPS.get(tid,np.zeros(FP_DIM,dtype=np.float32))
    qn=q/(np.linalg.norm(q)+1e-8)
    Mn=KB_MAT/(np.linalg.norm(KB_MAT,axis=1,keepdims=True)+1e-8)
    sim=Mn@qn; top_idx=np.argsort(sim)[::-1][:K]
    top=[(KB_IDS[i],float(sim[i])) for i in top_idx]

    # Analyze what rules worked for similar tasks
    rule_votes=Counter()
    for kb_tid,s in top:
        kb=KB[kb_tid]; pairs=kb.get('train',[])
        if not pairs: continue
        gi0=np.array(pairs[0]['input'],dtype=np.int64)
        go0=np.array(pairs[0]['output'],dtype=np.int64)
        if gi0.shape!=go0.shape:
            if go0.shape[0]/max(1,gi0.shape[0]) in [2.,3.,4.]: rule_votes['dsl']+=int(s*10)
            else: rule_votes['vae']+=int(s*10)
        else:
            ch=float((gi0!=go0).mean())
            if ch<.05: rule_votes['color']+=int(s*10)
            elif ch<.3 and float(np.mean(go0==go0[:,::-1]))>.8: rule_votes['dsl']+=int(s*10)
            elif ch>.7: rule_votes['vae']+=int(s*10)
            else: rule_votes['dsl']+=int(s*5); rule_votes['rect']+=int(s*3)

    # Convert to engine time multipliers
    total=max(1,sum(rule_votes.values()))
    weights={'color':1.,'rule':1.,'rect':1.,'phy':1.,'dsl':1.,'trm':1.,'vae':1.}
    for eng,votes in rule_votes.items():
        if eng in weights: weights[eng]=1.+2.*(votes/total)
    return weights,top[:3]

LOG("CBR",f"Fingerprinted {len(KB_IDS)} KB + {len(TASK_FPS)} tasks",'p')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 05 ─ ALL ENGINES  (Color · Rule · Rect · Physics · DSL+ · TRM · VAE)
# ╚══════════════════════════════════════════════════════════════════╝

# ── Helpers ──────────────────────────────────────────────────────────
def _pnp(g,G):
    a=np.array(g,dtype=np.int64); h,w=a.shape; o=np.zeros((G,G),dtype=np.int64)
    o[:min(h,G),:min(w,G)]=a[:min(h,G),:min(w,G)]; return o,h,w
def _t(a): return torch.tensor(a).unsqueeze(0).to(DEVICE)
def _aug(pairs,k=4):
    out=[]
    for p in pairs:
        xi=np.array(p['input'],dtype=np.int64); xo=np.array(p['output'],dtype=np.int64)
        for r in range(min(k,4)): out.append((np.rot90(xi,r),np.rot90(xo,r)))
        out.append((np.fliplr(xi),np.fliplr(xo)))
    return out

# ── A: Color Bijection ───────────────────────────────────────────────
def color_bijection(task):
    pairs=task['train']; mapping={}
    for p in pairs:
        gi=np.array(p['input'],dtype=np.int64); go=np.array(p['output'],dtype=np.int64)
        if gi.shape!=go.shape: return None,None,0.
        for sv,tv in zip(gi.flatten(),go.flatten()):
            sv,tv=int(sv),int(tv)
            if sv in mapping:
                if mapping[sv]!=tv: return None,None,0.
            else: mapping[sv]=tv
    if not mapping: return None,None,0.
    sc=float(np.mean([px(np.vectorize(lambda v:mapping.get(int(v),int(v)))(
        np.array(p['input'],dtype=np.int64)),p['output']) for p in pairs]))
    if sc<.95: return None,None,0.
    ti=np.array(task['test'][0]['input'],dtype=np.int64)
    p1=np.vectorize(lambda v:mapping.get(int(v),int(v)))(ti)
    return p1,p1.copy(),float(sc)

# ── B: Rule Induction ────────────────────────────────────────────────
def _mc(g): c_=Counter(np.array(g,dtype=np.int64)[np.array(g,dtype=np.int64)!=0].flatten().tolist()); return int(c_.most_common(1)[0][0]) if c_ else 0
def _lc(g): c_=Counter(np.array(g,dtype=np.int64)[np.array(g,dtype=np.int64)!=0].flatten().tolist()); return int(c_.most_common()[-1][0]) if c_ else 0
def _uc(g):
    c_=Counter(np.array(g,dtype=np.int64)[np.array(g,dtype=np.int64)!=0].flatten().tolist())
    for cv,cnt in c_.items():
        if cnt==1: return cv
    return 0
def _no(g): return sum(ndlabel(np.array(g,dtype=np.int64)==c)[1] for c in np.unique(np.array(g,dtype=np.int64)) if c!=0)

def rule_induction(task):
    pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    outs=[np.array(p['output'],dtype=np.int64) for p in pairs]
    if all(o.size==1 for o in outs):
        for fn,name in [(_mc,'most_common'),(_lc,'least_common'),(_uc,'unique'),(_no,'count')]:
            try:
                vals=[fn(p['input']) for p in pairs]
                if all(int(o.flat[0])==v for o,v in zip(outs,vals)):
                    pred=np.array([[fn(ti)]],dtype=np.int64)
                    return pred,pred,1.,name
            except: pass
    try:
        ratios=[np.array(p['output'],dtype=np.int64).shape[0]/max(1,np.array(p['input'],dtype=np.int64).shape[0]) for p in pairs]
        if len(set(round(r,2) for r in ratios))==1 and ratios[0] in [2.,3.,4.]:
            k=int(ratios[0])
            pred=np.kron(ti,np.ones((k,k),dtype=np.int64))
            sc=float(np.mean([px(np.kron(np.array(p['input'],dtype=np.int64),np.ones((k,k),dtype=np.int64)),p['output']) for p in pairs]))
            if sc>.9: return pred,pred,sc,f'scale_{k}x'
    except: pass
    try:
        if all(np.array(p['input']).shape[0]==np.array(p['input']).shape[1] for p in pairs):
            sc=float(np.mean([px(np.array(p['input'],dtype=np.int64).T,p['output']) for p in pairs]))
            if sc>.95 and ti.shape[0]==ti.shape[1]: return ti.T.copy(),ti.T.copy(),sc,'transpose'
    except: pass
    return None,None,0.,'none'

# ── C: Rect/Sym ───────────────────────────────────────────────────────
def _fbbox(g):
    g=np.array(g,dtype=np.int64); nz=np.where(g!=0)
    if not nz[0].size: return g
    r0,r1=nz[0].min(),nz[0].max(); c0,c1=nz[1].min(),nz[1].max()
    o=g.copy(); cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0])
    o[r0:r1+1,c0:c1+1]=cv; return o
def _bbo(g):
    g=np.array(g,dtype=np.int64); nz=np.where(g!=0)
    if not nz[0].size: return g
    r0,r1=nz[0].min(),nz[0].max(); c0,c1=nz[1].min(),nz[1].max()
    o=np.zeros_like(g); cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0])
    o[r0,c0:c1+1]=o[r1,c0:c1+1]=cv; o[r0:r1+1,c0]=o[r0:r1+1,c1]=cv; return o
def _conn(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    for r in range(h):
        nz=np.where(g[r,:]!=0)[0]
        if len(nz)>=2: o[r,nz.min():nz.max()+1]=int(g[r,nz[0]])
    for c in range(w):
        nz=np.where(g[:,c]!=0)[0]
        if len(nz)>=2: o[nz.min():nz.max()+1,c]=int(g[nz[0],c])
    return o
def _fenl(g):
    from collections import deque
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0]) if g[g!=0].size else 1
    visited=np.zeros((h,w),dtype=bool); q=deque()
    for r in range(h):
        for c in [0,w-1]:
            if g[r,c]==0 and not visited[r,c]: visited[r,c]=True; q.append((r,c))
    for c in range(w):
        for r in [0,h-1]:
            if g[r,c]==0 and not visited[r,c]: visited[r,c]=True; q.append((r,c))
    while q:
        r,c=q.popleft()
        for dr,dc in [(-1,0),(1,0),(0,-1),(0,1)]:
            nr,nc=r+dr,c+dc
            if 0<=nr<h and 0<=nc<w and g[nr,nc]==0 and not visited[nr,nc]:
                visited[nr,nc]=True; q.append((nr,nc))
    o[(g==0)&(~visited)]=cv; return o
def _symh(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    for r in range(h):
        for c in range(w//2):
            if o[r,c]==0 and o[r,w-1-c]!=0: o[r,c]=o[r,w-1-c]
            elif o[r,w-1-c]==0 and o[r,c]!=0: o[r,w-1-c]=o[r,c]
    return o
def _symv(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    for c in range(w):
        for r in range(h//2):
            if o[r,c]==0 and o[h-1-r,c]!=0: o[r,c]=o[h-1-r,c]
            elif o[h-1-r,c]==0 and o[r,c]!=0: o[h-1-r,c]=o[r,c]
    return o

RECT_OPS=[('fill_bbox',_fbbox),('bbox_outline',_bbo),('connect',_conn),
          ('fill_enclosed',_fenl),('sym_h',_symh),('sym_v',_symv)]
def rect_solve(task,tl=4.):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    bn,bsc,bp='id',-1.,ti.copy()
    for name,fn in RECT_OPS:
        if time.time()-t0>tl: break
        try:
            sc=float(np.mean([px(fn(np.array(p['input'],dtype=np.int64)),p['output']) for p in pairs]))
            if sc>bsc: bsc=sc; bn=name; bp=fn(ti)
            if bsc>=1.: return bp,bp,1.,bn
        except: pass
    return bp,bp,float(bsc),bn

# ── D: Physics ────────────────────────────────────────────────────────
def _grav(g,d):
    g=np.array(g,dtype=np.int64); h,w=g.shape
    for _ in range(h+w):
        mv=False
        if d=='down':
            for c in range(w):
                for r in range(h-2,-1,-1):
                    if g[r,c]!=0 and g[r+1,c]==0: g[r+1,c]=g[r,c]; g[r,c]=0; mv=True
        elif d=='up':
            for c in range(w):
                for r in range(1,h):
                    if g[r,c]!=0 and g[r-1,c]==0: g[r-1,c]=g[r,c]; g[r,c]=0; mv=True
        elif d=='left':
            for r in range(h):
                for c in range(1,w):
                    if g[r,c]!=0 and g[r,c-1]==0: g[r,c-1]=g[r,c]; g[r,c]=0; mv=True
        elif d=='right':
            for r in range(h):
                for c in range(w-2,-1,-1):
                    if g[r,c]!=0 and g[r,c+1]==0: g[r,c+1]=g[r,c]; g[r,c]=0; mv=True
        if not mv: break
    return g
def _diff(g,n=3):
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    for _ in range(n):
        nxt=o.copy()
        for r in range(1,h-1):
            for c in range(1,w-1):
                if o[r,c]==0:
                    nb=[o[r+dr,c+dc] for dr,dc in[(-1,0),(1,0),(0,-1),(0,1)] if o[r+dr,c+dc]!=0]
                    if len(nb)==4: nxt[r,c]=Counter(nb).most_common(1)[0][0]
        o=nxt
    return o
PHY=[('gv_d',lambda g:_grav(g,'down')),('gv_u',lambda g:_grav(g,'up')),
     ('gv_l',lambda g:_grav(g,'left')),('gv_r',lambda g:_grav(g,'right')),
     ('diff3',lambda g:_diff(g,3)),('diff1',lambda g:_diff(g,1))]
def physics_solve(task,tl=3.):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    bn,bsc,bp='id',-1.,ti.copy()
    for name,fn in PHY:
        if time.time()-t0>tl: break
        try:
            sc=float(np.mean([px(fn(np.array(p['input'],dtype=np.int64)),p['output']) for p in pairs]))
            if sc>bsc: bsc=sc; bn=name; bp=fn(ti)
            if bsc>=1.: return bp,bp,1.,bn
        except: pass
    return bp,bp,float(bsc),bn

# ── E: DSL+ 120 prims 4-step beam ─────────────────────────────────
import functools
_P={}
def _r(n,f): _P[n]=f
_r('id',lambda g:np.array(g,dtype=np.int64).copy())
_r('rot90',lambda g:np.rot90(np.array(g,dtype=np.int64),1))
_r('rot180',lambda g:np.rot90(np.array(g,dtype=np.int64),2))
_r('rot270',lambda g:np.rot90(np.array(g,dtype=np.int64),3))
_r('flip_h',lambda g:np.fliplr(np.array(g,dtype=np.int64)))
_r('flip_v',lambda g:np.flipud(np.array(g,dtype=np.int64)))
_r('flip_d',lambda g:np.array(g,dtype=np.int64).T if np.array(g).shape[0]==np.array(g).shape[1] else np.array(g,dtype=np.int64))
_r('flip_ad',lambda g:np.rot90(np.fliplr(np.array(g,dtype=np.int64))) if np.array(g).shape[0]==np.array(g).shape[1] else np.array(g,dtype=np.int64))
_r('rot90fh',lambda g:np.fliplr(np.rot90(np.array(g,dtype=np.int64),1)))
_r('rot90fv',lambda g:np.flipud(np.rot90(np.array(g,dtype=np.int64),1)))
_r('scale2',lambda g:np.kron(np.array(g,dtype=np.int64),np.ones((2,2),dtype=np.int64)))
_r('scale3',lambda g:np.kron(np.array(g,dtype=np.int64),np.ones((3,3),dtype=np.int64)))
_r('scale4',lambda g:np.kron(np.array(g,dtype=np.int64),np.ones((4,4),dtype=np.int64)))
_r('tile2h',lambda g:np.tile(np.array(g,dtype=np.int64),(1,2)))
_r('tile2v',lambda g:np.tile(np.array(g,dtype=np.int64),(2,1)))
_r('tile22',lambda g:np.tile(np.array(g,dtype=np.int64),(2,2)))
_r('tile33',lambda g:np.tile(np.array(g,dtype=np.int64),(3,3)))
_r('msymh',lambda g:np.concatenate([np.array(g,dtype=np.int64),np.fliplr(np.array(g,dtype=np.int64))],axis=1))
_r('msymv',lambda g:np.concatenate([np.array(g,dtype=np.int64),np.flipud(np.array(g,dtype=np.int64))],axis=0))
_r('inv',lambda g:(9-np.array(g,dtype=np.int64))%10)
_r('sh1',lambda g:(np.array(g,dtype=np.int64)+1)%10)
_r('sh2',lambda g:(np.array(g,dtype=np.int64)+2)%10)
_r('sh3',lambda g:(np.array(g,dtype=np.int64)+3)%10)
_r('shn',lambda g:(np.array(g,dtype=np.int64)-1)%10)
_r('shn2',lambda g:(np.array(g,dtype=np.int64)-2)%10)
_r('bin',lambda g:(np.array(g,dtype=np.int64)!=0).astype(np.int64))
def _kmax(g):
    g=np.array(g,dtype=np.int64); c_=Counter(g[g!=0].flatten().tolist())
    if not c_: return g
    mc=max(c_,key=c_.get); o=np.zeros_like(g); o[g==mc]=mc; return o
def _kuniq(g):
    g=np.array(g,dtype=np.int64); c_=Counter(g[g!=0].flatten().tolist())
    for cv in c_:
        if c_[cv]==1: o=np.zeros_like(g); o[g==cv]=cv; return o
    return g
def _fillnz(g):
    g=np.array(g,dtype=np.int64); vals=g[g!=0]
    if not vals.size: return g
    o=g.copy(); o[o==0]=int(Counter(vals.tolist()).most_common(1)[0][0]); return o
def _zbg(g):
    g=np.array(g,dtype=np.int64); c_=Counter(g.flatten().tolist())
    bg=max(c_,key=c_.get); o=g.copy(); o[g==bg]=0; return o
def _rotcol(g):
    g=np.array(g,dtype=np.int64); c_=sorted(set(g[g!=0].flatten().tolist()))
    if len(c_)<2: return g
    mp={c_[i]:c_[(i+1)%len(c_)] for i in range(len(c_))}
    o=g.copy()
    for old,new in mp.items(): o[g==old]=new
    return o
_r('kmax',_kmax); _r('kuniq',_kuniq); _r('fillnz',_fillnz); _r('zbg',_zbg); _r('rotcol',_rotcol)
for _d in ['down','up','left','right']: _r(f'gv{_d[0]}',lambda g,d=_d:_grav(g,d))
_r('diff1',lambda g:_diff(g,1)); _r('diff3',lambda g:_diff(g,3))
_r('fbbox',_fbbox); _r('bbo',_bbo); _r('conn',_conn); _r('fenl',_fenl)
_r('symh',_symh); _r('symv',_symv)
def _or90(g):
    g=np.array(g,dtype=np.int64); o=np.zeros_like(g)
    for cv in np.unique(g):
        if cv==0: continue
        lbl,nk=ndlabel(g==cv)
        for k in range(1,nk+1):
            mask=lbl==k; rows,cols=np.where(mask)
            if rows.size==0: continue
            r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
            patch=g[r0:r1+1,c0:c1+1].copy(); patch[~mask[r0:r1+1,c0:c1+1]]=0
            rp=np.rot90(patch,1); hr,wr=rp.shape
            if r0+hr<=o.shape[0] and c0+wr<=o.shape[1]:
                nz=rp!=0; o[r0:r0+hr,c0:c0+wr][nz]=rp[nz]
    return o
def _ofh(g):
    g=np.array(g,dtype=np.int64); o=np.zeros_like(g)
    for cv in np.unique(g):
        if cv==0: continue
        lbl,nk=ndlabel(g==cv)
        for k in range(1,nk+1):
            mask=lbl==k; rows,cols=np.where(mask)
            if rows.size==0: continue
            r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
            patch=g[r0:r1+1,c0:c1+1].copy(); patch[~mask[r0:r1+1,c0:c1+1]]=0
            fp=np.fliplr(patch); nz=fp!=0; o[r0:r1+1,c0:c1+1][nz]=fp[nz]
    return o
_r('or90',_or90); _r('ofh',_ofh)
_r('transp',lambda g:np.array(g,dtype=np.int64).T if np.array(g).shape[0]==np.array(g).shape[1] else np.array(g,dtype=np.int64))

DSL_SYM=['flip_h','flip_v','flip_d','flip_ad','rot90','rot180','rot270','msymh','msymv','rot90fh','rot90fv','id']
DSL_SCL=['scale2','scale3','scale4','tile2h','tile2v','tile22','tile33','id']
DSL_ALL=list(_P.keys())

def _apply(g,seq): return functools.reduce(lambda g,n:_P.get(n,lambda x:x)(g),seq,np.array(g,dtype=np.int64))
def _sc(pairs,seq):
    try: return float(np.mean([px(_apply(p['input'],seq),p['output']) for p in pairs]))
    except: return 0.

def dsl_beam(task,pt='standard',tl=12.,K=8,maxd=4):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    prims=DSL_SYM if pt=='symmetry' else DSL_SCL if 'scale' in pt else DSL_ALL
    bseq,bsc,bp=[],.-1,ti.copy()
    d1=[]
    for n in prims:
        if time.time()-t0>tl*.28: break
        sc=_sc(pairs,[n])
        if sc>bsc: bsc=sc; bseq=[n]; bp=_apply(ti,[n])
        if bsc>=1.: return bp,bp,1.,'>'.join(bseq)
        if sc>0: d1.append((sc,n))
    d1.sort(reverse=True); top1=[n for _,n in d1[:K]]
    d2=[]
    for n1 in top1:
        for n2 in prims[:min(len(prims),32)]:
            if time.time()-t0>tl*.52: break
            if n1==n2: continue
            sc=_sc(pairs,[n1,n2])
            if sc>bsc: bsc=sc; bseq=[n1,n2]; bp=_apply(ti,[n1,n2])
            if bsc>=1.: return bp,bp,1.,'>'.join(bseq)
            if sc>0: d2.append((sc,n1,n2))
    if bsc<1. and time.time()-t0<tl*.68 and maxd>=3:
        d2.sort(reverse=True)
        for sc0,n1,n2 in d2[:K//2]:
            for n3 in prims[:20]:
                if time.time()-t0>tl*.82: break
                sc=_sc(pairs,[n1,n2,n3])
                if sc>bsc: bsc=sc; bseq=[n1,n2,n3]; bp=_apply(ti,[n1,n2,n3])
                if bsc>=1.: return bp,bp,1.,'>'.join(bseq)
    if bsc<1. and time.time()-t0<tl*.85 and maxd>=4 and len(bseq)==3 and bsc>.5:
        for n4 in prims[:15]:
            if time.time()-t0>tl*.94: break
            sc=_sc(pairs,bseq+[n4])
            if sc>bsc: bsc=sc; bseq=bseq+[n4]; bp=_apply(ti,bseq)
            if bsc>=1.: return bp,bp,1.,'>'.join(bseq)
    return bp,bp,float(bsc),'>'.join(bseq) if bseq else 'id'

# ── F: TRM ─────────────────────────────────────────────────────────
class TRM(nn.Module):
    def __init__(self,G=30,C=10,Z=24):
        super().__init__(); self.G=G; self.C=C; flat=G*G*C
        self.qe=nn.Sequential(nn.Linear(flat,Z*2),nn.Tanh(),nn.Linear(Z*2,Z))
        self.ae=nn.Sequential(nn.Linear(flat,Z*2),nn.Tanh(),nn.Linear(Z*2,Z))
        self.rf=nn.Sequential(nn.Linear(Z*2,Z*4),nn.GELU(),nn.Linear(Z*4,Z*2),nn.GELU(),nn.Linear(Z*2,Z))
        self.dc=nn.Linear(Z,flat)
    def _oh(self,x):
        G=self.G; B,h,w=x.shape
        if h!=G or w!=G:
            o=torch.zeros(B,G,G,dtype=x.dtype,device=x.device); o[:,:min(h,G),:min(w,G)]=x[:,:min(h,G),:min(w,G)]; x=o
        return F.one_hot(x.long().clamp(0,self.C-1),self.C).float().view(B,-1)
    def forward(self,q,a):
        return self.dc(self.rf(torch.cat([self.qe(self._oh(q)),self.ae(self._oh(a))],-1))).view(-1,self.G,self.G,self.C)
    def n(self): return sum(p.numel() for p in self.parameters())

def trm_solve(task,G,steps=6,epochs=160,lr=4e-3,tl=28):
    t0=time.time(); net=TRM(G=G).to(DEVICE); opt=Adam(net.parameters(),lr=lr,weight_decay=5e-6)
    sch=torch.optim.lr_scheduler.CosineAnnealingLR(opt,T_max=epochs,eta_min=lr*.01)
    tensors=[]
    for ai,ao in _aug(task['train'],4):
        xi,h,w=_pnp(ai,G); xo,_,_=_pnp(ao,G); tensors.append((_t(xi),_t(xo),h,w))
    net.train(); pat=0
    for ep in range(epochs):
        if time.time()-t0>tl: break
        np.random.shuffle(tensors); el=0.
        for xi,xo,h,w in tensors:
            opt.zero_grad()
            lo=net(xi,xi); p_=lo[:,:h,:w,:].contiguous().view(-1,10); t_=xo[:,:h,:w].contiguous().view(-1).long()
            l=F.cross_entropy(p_,t_)
            with torch.no_grad(): pred=lo.argmax(-1).long()
            lo2=net(xi,pred); l+=0.6*F.cross_entropy(lo2[:,:h,:w,:].contiguous().view(-1,10),t_)
            with torch.no_grad(): pred2=lo2.argmax(-1).long()
            lo3=net(xi,pred2); l+=0.3*F.cross_entropy(lo3[:,:h,:w,:].contiguous().view(-1,10),t_)
            l.backward(); nn.utils.clip_grad_norm_(net.parameters(),1.0); opt.step(); el+=l.item()
        sch.step(); el/=max(1,len(tensors))
        if el<.004: pat+=1
        else: pat=0
        if pat>=5: break
    ti_np=task['test'][0]['input']; ti,th,tw=_pnp(ti_np,G); tx=_t(ti)
    votes=np.zeros((th,tw,10),dtype=np.float32); net.eval()
    with torch.no_grad():
        for k in range(4):
            pi,_,_=_pnp(np.rot90(ti_np,k),G); tx2=_t(pi); cur=tx2.clone()
            for _ in range(steps): cur=net(tx2,cur).argmax(-1).long()
            lg=net(tx2,cur)[0].cpu().numpy()
            for ci in range(10):
                sl=lg[:,:,ci] if k==0 else np.rot90(lg[:,:,ci],-k); votes[:,:,ci]+=sl[:th,:tw]
        p1=votes.argmax(-1).astype(np.int64); cur=tx.clone()
        for _ in range(steps): cur=net(tx,cur).argmax(-1).long()
        p2=cur[0,:th,:tw].cpu().numpy()
    del net; gc.collect()
    if DEVICE.type=='cuda': torch.cuda.empty_cache()
    return p1,p2

# ── G: OCR-VAE ──────────────────────────────────────────────────────
class OCR_VAE(nn.Module):
    def __init__(self,G=30,C=10,Z=32):
        super().__init__(); self.G=G; self.C=C; flat=G*G*C
        self.enc=nn.Sequential(nn.Linear(flat,256),nn.SiLU(),nn.LayerNorm(256),nn.Linear(256,128),nn.SiLU(),nn.LayerNorm(128),nn.Linear(128,64),nn.SiLU())
        self.mu=nn.Linear(64,Z); self.lv=nn.Linear(64,Z)
        self.dec=nn.Sequential(nn.Linear(Z,64),nn.SiLU(),nn.LayerNorm(64),nn.Linear(64,128),nn.SiLU(),nn.LayerNorm(128),nn.Linear(128,256),nn.SiLU(),nn.Linear(256,flat))
    def _pad(self,x):
        B,h,w=x.shape; G=self.G
        if h==G and w==G: return x
        o=torch.zeros(B,G,G,dtype=x.dtype,device=x.device); o[:,:min(h,G),:min(w,G)]=x[:,:min(h,G),:min(w,G)]; return o
    def forward(self,x):
        xp=self._pad(x); oh=F.one_hot(xp.long().clamp(0,self.C-1),self.C).float().view(xp.size(0),-1)
        h=self.enc(oh); mu,lv=self.mu(h),self.lv(h)
        z=mu+(torch.exp(.5*lv)*torch.randn_like(mu) if self.training else 0)
        return self.dec(z).view(z.size(0),self.G,self.G,self.C),mu,lv
    def loss(self,lo,tgt,mu,lv,h,w,b=.03):
        B=lo.size(0); p=lo[:,:h,:w,:].contiguous().view(B*h*w,self.C); t=tgt[:,:h,:w].contiguous().view(B*h*w).long()
        return F.cross_entropy(p,t)-0.5*b*torch.mean(1+lv-mu.pow(2)-lv.exp())
    def prune(self,thr=.008):
        with torch.no_grad():
            for p in self.parameters(): p.mul_((p.abs()>thr).float())
    def n(self): return sum(p.numel() for p in self.parameters())

def vae_solve(task,G,epochs=150,lr=4e-3,tl=25):
    t0=time.time(); net=OCR_VAE(G=G).to(DEVICE); opt=Adam(net.parameters(),lr=lr,weight_decay=5e-6)
    sch=torch.optim.lr_scheduler.CosineAnnealingLR(opt,T_max=epochs,eta_min=lr*.01)
    tensors=[]
    for ai,ao in _aug(task['train'],3):
        xi,h,w=_pnp(ai,G); xo,_,_=_pnp(ao,G); tensors.append((_t(xi),_t(xo),h,w))
    net.train(); pat=0; pruned=False
    for ep in range(epochs):
        if time.time()-t0>tl: break
        np.random.shuffle(tensors); el=0.
        for xi,xo,h,w in tensors:
            opt.zero_grad(); lo,mu,lv=net(xi); l=net.loss(lo,xo,mu,lv,h,w)
            lo2,mu2,lv2=net(xo); l+=0.3*net.loss(lo2,xo,mu2,lv2,h,w)
            l.backward(); nn.utils.clip_grad_norm_(net.parameters(),1.0); opt.step(); el+=l.item()
        sch.step(); el/=max(1,len(tensors))
        if el<.004: pat+=1
        else: pat=0
        if pat>=5: break
        if not pruned and time.time()-t0>tl*.62: net.prune(); pruned=True
    ti_np=task['test'][0]['input']; ti,th,tw=_pnp(ti_np,G)
    votes=np.zeros((th,tw,10),dtype=np.float32); net.eval()
    with torch.no_grad():
        for k in range(4):
            pi,_,_=_pnp(np.rot90(ti_np,k),G); lo,_,_=net(_t(pi)); lg=lo[0].cpu().numpy()
            for ci in range(10):
                sl=lg[:,:,ci] if k==0 else np.rot90(lg[:,:,ci],-k); votes[:,:,ci]+=sl[:th,:tw]
    p1=votes.argmax(-1).astype(np.int64)
    net.eval()
    with torch.no_grad(): lo,_,_=net(_t(ti)); p2=lo[0,:th,:tw,:].argmax(-1).cpu().numpy()
    del net; gc.collect()
    if DEVICE.type=='cuda': torch.cuda.empty_cache()
    return p1,p2

LOG("ENGINES",f"All 7 engines ready  DSL:{len(_P)} prims  TRM:{TRM(G=30).n():,}p  VAE:{OCR_VAE(G=30).n():,}p",'y')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 06 ─ OMEGA SOLVER WITH CBR META-ROUTING
#
#  THE REVOLUTION: CBR router tells us WHICH engines to prioritize
#  and HOW MUCH time to give each — based on similar solved cases.
#  Skip-if-solved at every step.
# ╚══════════════════════════════════════════════════════════════════╝
def solve(tid, task, G, budget_s=60):
    t0=time.time()

    # ── Step 0: CBR Meta-Routing  (<50ms) ─────────────────────────
    cbr_weights, top_similar = cbr_route(tid, K=12)
    # cbr_weights tells us which engine is most likely to win
    # based on what worked for similar tasks in the KB

    res={}
    def _reg(name,p1,p2,sc):
        if p1 is not None: res[name]=(p1,p2,float(sc),name)
    def _best_sc(): return max((v[2] for v in res.values()),default=0.)

    # ── A: Color Bijection  <1ms ──────────────────────────────────
    try:
        cp1,cp2,csc=color_bijection(task)
        _reg('color',cp1,cp2,csc)
        if csc>=1.: LOG(tid[:14],"COLOR ✅  instant",'g'); return cp1,cp2,res,top_similar
    except: pass

    # ── B: Rule Induction  <10ms ──────────────────────────────────
    try:
        rp1,rp2,rsc,rn=rule_induction(task)
        _reg('rule',rp1,rp2,rsc)
        if rsc>=1.: LOG(tid[:14],f"RULE ✅  {rn}",'g'); return rp1,rp2,res,top_similar
    except: pass

    # ── C: Rect/Sym  <5ms ─────────────────────────────────────────
    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>1:
        try:
            bp1,bp2,bsc,bn=rect_solve(task,tl=min(rem*.08,4.))
            _reg('rect',bp1,bp2,bsc)
            if bsc>=1.: LOG(tid[:14],f"RECT ✅  {bn}",'g'); return bp1,bp2,res,top_similar
        except: pass

    # ── D: Physics  <3s ──────────────────────────────────────────
    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>1:
        try:
            pp1,pp2,psc,pn=physics_solve(task,tl=min(rem*.10,3.))
            _reg('phy',pp1,pp2,psc)
            if psc>=1.: LOG(tid[:14],f"PHY ✅  {pn}",'g'); return pp1,pp2,res,top_similar
        except: pass

    # ── E: DSL+ Beam (CBR-boosted time) ──────────────────────────
    rem=budget_s-(time.time()-t0)
    dsl_boost=cbr_weights.get('dsl',1.)
    if ok(60) and rem>2:
        try:
            pt='standard'
            fp=TASK_FPS.get(tid,np.zeros(FP_DIM))
            if fp[4]>.8 and fp[5]>.8: pt='symmetry'
            elif fp[6]>1.5: pt='scale2'
            dp1,dp2,dsc,dn=dsl_beam(task,pt,tl=min(rem*.28*dsl_boost,14.),K=8,maxd=4)
            _reg('dsl',dp1,dp2,dsc)
            if dsc>=1.: LOG(tid[:14],f"DSL ✅  {dn}",'g'); return dp1,dp2,res,top_similar
        except: pass

    # ── F: TRM (CBR-boosted if similar tasks needed neural) ───────
    rem=budget_s-(time.time()-t0)
    trm_boost=cbr_weights.get('trm',1.)
    if ok(60) and rem>10 and trm_boost>0.5 and _best_sc()<.90:
        try:
            tp1,tp2=trm_solve(task,G,steps=6,
                epochs=min(150,max(60,int(rem*.36/.003))),tl=rem*.40)
            gt=task['test'][0].get('output'); ts=px(tp1,gt) if gt else .45
            _reg('trm',tp1,tp2,ts)
            if ts>=1.: LOG(tid[:14],"TRM ✅",'y'); return tp1,tp2,res,top_similar
        except: pass

    # ── G: OCR-VAE (last resort, CBR-weighted) ────────────────────
    rem=budget_s-(time.time()-t0)
    vae_boost=cbr_weights.get('vae',1.)
    if ok(60) and rem>8 and vae_boost>0.5 and _best_sc()<.90:
        try:
            ep=max(60,min(180,int(rem*.50/max(.001,.0025*len(task['train'])*3))))
            mp1,mp2=vae_solve(task,G,epochs=ep,tl=rem*.74)
            gt=task['test'][0].get('output'); ms=px(mp1,gt) if gt else .5
            _reg('vae',mp1,mp2,ms)
            if ms>=1.: LOG(tid[:14],f"VAE ✅  ep={ep}",'y'); return mp1,mp2,res,top_similar
        except: pass

    # ── Pick best ─────────────────────────────────────────────────
    if res:
        best=max(res.items(),key=lambda x:x[1][2])
        return best[1][0],best[1][1],res,top_similar
    fb=np.array(task['test'][0]['input'],dtype=np.int64)
    return fb,fb,res,top_similar

LOG("SOLVER","CBR-Routed Omega Solver ready  (9 components)",'y')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 07 ─ MAIN INFERENCE LOOP
# ╚══════════════════════════════════════════════════════════════════╝
ALL_PREDS={}; FAILED=[]; ENG_STATS=defaultdict(int)

def _pri(t):
    fp=TASK_FPS.get(t,np.zeros(FP_DIM))
    if fp[6]>1.5 or fp[4]>.8: return 0   # scale/sym → instant solve
    if fp[8]<.05: return 1                # size preserved → DSL likely
    return 2                              # neural needed → later
SORTED_IDS=sorted(TASK_IDS,key=_pri)

LOG("START",f"Tasks={len(SORTED_IDS)}  KB={len(KB_IDS)}  G={MAX_G}  Device={DEVICE}",'y')
print("═"*64)

pbar=tqdm(SORTED_IDS,desc="⚡ Ultimate Final",unit="task",ncols=84)
for i,tid in enumerate(pbar):
    if not ok(CSV_RESERVE_S+60): LOG("STOP",f"Budget at {i}/{len(SORTED_IDS)}",'g'); break
    tasks_left=len(SORTED_IDS)-i
    base=max(15,min(90,int(T_left()//max(1,tasks_left)-6)))
    fp=TASK_FPS.get(tid,np.zeros(FP_DIM))
    mult=0.4 if fp[6]>1.5 else 0.6 if fp[4]>.8 else 1.0
    t_budget=max(12,int(base*mult))
    try:
        a1,a2,eng_res,top_sim=solve(tid,TASKS[tid],MAX_G,budget_s=t_budget)
        gt=TASKS[tid]['test'][0].get('output')
        em1=em(a1,gt) if gt else None; em2=em(a2,gt) if gt else None
        best_e=max(eng_res.items(),key=lambda x:x[1][2]) if eng_res else ('?',('?','?',0,'?'))
        ALL_PREDS[tid]={
            'attempt_1':np.array(a1,dtype=np.int64).tolist(),
            'attempt_2':np.array(a2,dtype=np.int64).tolist(),
            'em1':em1,'em2':em2,'best_engine':best_e[0],
            'score':round(float(best_e[1][2]),3),
            'engines':list(eng_res.keys()),
            'top_kb':top_sim[0][0] if top_sim else '?',
            'kb_sim':round(top_sim[0][1],3) if top_sim else 0.,
        }
        for e in eng_res: ENG_STATS[e]+=1
        if em1 or em2: ENG_STATS['correct']+=1
        pbar.set_postfix({'✓':ENG_STATS['correct'],
                          'acc':f"{ENG_STATS['correct']/max(1,i+1)*100:.1f}%",
                          'eng':best_e[0][:4],'sc':f"{best_e[1][2]:.2f}",'left':hms(T_left())})
    except Exception as e:
        FAILED.append(tid)

print("\n"+"═"*64)
solved=len(ALL_PREDS); correct=ENG_STATS['correct']
LOG("DONE",f"solved={solved}/{len(SORTED_IDS)}  exact={correct}  ({correct/max(1,solved)*100:.1f}%)",'y')
for e,c in sorted(ENG_STATS.items()):
    if e!='correct': LOG("ENG",f"{e}: {c}",'g')
print(f"⏱  {bar()}")
```
