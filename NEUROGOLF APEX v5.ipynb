```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 03 ─ LOAD ALL DATA · DEEP TOPOLOGY · META-CLASSIFIER
#  NEW: Puzzle-type classification (movement/repetition/completion
#       /physics/symmetry/scaling) for CFI intelligent routing
# ╚══════════════════════════════════════════════════════════════════╝
UTILS=None; sys.path.insert(0,DATA_ROOT); up=Path(UTILS_DIR)
if up.exists():
    try:
        import neurogolf_utils as UTILS
        LOG("UTILS",f"Imported: {[x for x in dir(UTILS) if not x.startswith('_')]}",'green')
    except:
        for pyf in sorted(up.glob('*.py')):
            try:
                sp=importlib.util.spec_from_file_location(pyf.stem,pyf)
                m=importlib.util.module_from_spec(sp); sp.loader.exec_module(m)
                UTILS=m; LOG("UTILS",f"{pyf.name} loaded",'green')
            except Exception as e: LOG("UTILS",f"WARN {pyf.name}: {e}",'pink')
else: LOG("UTILS","absent — solo mode",'pink')

TASKS,ERRS={},[]
jsons=sorted(Path(DATA_ROOT).rglob('*.json'))
LOG("DATA",f"Found {len(jsons)} JSON file(s)",'cyan')
for jf in jsons:
    try:
        d=json.loads(jf.read_text())
        if isinstance(d,dict) and 'train' in d and 'test' in d: TASKS[jf.stem]=d
        elif isinstance(d,list):
            for i,it in enumerate(d):
                if isinstance(it,dict) and 'train' in it: TASKS[f'{jf.stem}_{i:04d}']=it
        else: TASKS[f'_raw_{jf.stem}']=d
    except Exception as e: ERRS.append((jf.name,str(e)))
TASK_IDS=[k for k in TASKS if not k.startswith('_raw_')]
LOG("DATA",f"Tasks loaded: {len(TASK_IDS)}  errors: {len(ERRS)}",'green')

# ── META-CLASSIFIER: puzzle type detection ─────────────────────────
def classify_puzzle(task):
    pairs=task.get('train',[]); 
    if not pairs: return 'unknown',{}
    gi0=np.array(pairs[0]['input'],dtype=np.int64)
    go0=np.array(pairs[0]['output'],dtype=np.int64)
    # Size change?
    size_change=any(np.array(p['input']).shape!=np.array(p['output']).shape for p in pairs)
    # Scale hint
    scale=None
    if go0.shape[0]>0 and gi0.shape[0]>0:
        rr=go0.shape[0]/gi0.shape[0]; rc=go0.shape[1]/gi0.shape[1]
        if rr==rc and rr in [2.,3.,4.]: scale=int(rr)
    # Symmetry
    def sym(g): h,w=g.shape; return max(float(np.mean(g==g[:,::-1])),float(np.mean(g==g[::-1,:])),float(np.mean(g==g.T)) if h==w else 0.)
    sym_s=float(np.mean([sym(np.array(p['input'],dtype=np.int64)) for p in pairs]))
    # Color change
    ci=np.mean([len(np.unique(np.array(p['input']))) for p in pairs])
    co=np.mean([len(np.unique(np.array(p['output']))) for p in pairs])
    dc=co-ci
    # Motion detection (do objects shift?)
    motion=False
    if not size_change and gi0.shape==go0.shape:
        diff=(gi0!=go0).sum()/(gi0.size+1e-9)
        motion=0.05<float(diff)<0.7
    # Object count
    def n_obj(g): return sum(ndlabel(g==c)[1] for c in np.unique(g) if c!=0)
    obj_cnt=float(np.mean([n_obj(np.array(p['input'],dtype=np.int64)) for p in pairs[:2]]))
    # Repetition/tiling hint
    tiling=False
    if not size_change and gi0.shape==go0.shape:
        # Check if output is input repeated
        for k in [2,3]:
            if go0.shape[0]==gi0.shape[0]*k:
                tiling=True; break
    # Classify
    if scale: ptype=f'scale{scale}'
    elif sym_s>0.80: ptype='symmetry'
    elif tiling: ptype='tiling'
    elif motion and obj_cnt>1: ptype='movement'
    elif dc<-0.5: ptype='reduction'
    elif dc>0.5: ptype='completion'
    elif obj_cnt>4: ptype='objects'
    elif size_change: ptype='size_change'
    else: ptype='standard'
    # Engine routing weights
    wt={'dsl':1.0,'mdl':1.0,'trm':0.8,'phy':0.5,'knn':0.5,'oct':0.5}
    if ptype in ('symmetry','movement'): wt['dsl']=2.5; wt['mdl']=0.6
    if ptype in ('objects',): wt['oct']=2.0; wt['knn']=1.5; wt['mdl']=1.2
    if ptype in ('completion',): wt['knn']=2.0; wt['mdl']=1.5
    if ptype in ('scale2','scale3','scale4'): wt['dsl']=3.0; wt['mdl']=0.4
    if ptype=='standard': wt['mdl']=1.8; wt['trm']=1.4
    if motion: wt['phy']=1.5
    return ptype, wt

# ── Full topology per task ─────────────────────────────────────────
TASK_META={}; all_max_h=all_max_w=0
for tid in TASK_IDS:
    task=TASKS[tid]; shapes_i,shapes_o,colors=[],[],set()
    for p in task.get('train',[])+task.get('test',[]):
        gi=np.array(p['input'],dtype=np.int64)
        shapes_i.append(gi.shape); colors.update(gi.flatten().tolist())
        if 'output' in p:
            go_=np.array(p['output'],dtype=np.int64)
            shapes_o.append(go_.shape); colors.update(go_.flatten().tolist())
    mh=max(s[0] for s in shapes_i+(shapes_o or shapes_i))
    mw=max(s[1] for s in shapes_i+(shapes_o or shapes_i))
    all_max_h=max(all_max_h,mh); all_max_w=max(all_max_w,mw)
    ptype,wt=classify_puzzle(task)
    TASK_META[tid]=dict(max_h=mh,max_w=mw,n_train=len(task.get('train',[])),
                        n_colors=len(colors),ptype=ptype,weights=wt)

MAX_G=max(all_max_h,all_max_w,30)
type_dist=Counter(v['ptype'] for v in TASK_META.values())
LOG("META",f"MAX_G={MAX_G}  types: {dict(type_dist.most_common())}",'gold')
print(f"\n{'═'*60}")
for pt,cnt in type_dist.most_common():
    bar_=('█'*min(30,cnt*2)).ljust(30); print(f"  {pt:<14} {cnt:3d}  {bar_}")
print('═'*60)
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 05 ─ ENGINE-1: OCT  Object-Centric Topology Engine
#  NEW: Object tree with Containment + Adjacency relations
#  Solves puzzles where objects move as parent-child groups
# ╚══════════════════════════════════════════════════════════════════╝
def extract_objects(grid):
    g=np.array(grid,dtype=np.int64); h,w=g.shape
    objects=[]
    for cv in np.unique(g):
        if cv==0: continue
        lbl,n=ndlabel(g==cv)
        for k in range(1,n+1):
            mask=lbl==k; rows,cols=np.where(mask)
            if rows.size==0: continue
            r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
            objects.append({'color':int(cv),'mask':mask,'bbox':(r0,r1,c0,c1),
                            'size':int(mask.sum()),'cy':(r0+r1)/2,'cx':(c0+c1)/2})
    return objects

def build_object_tree(objects, g):
    h,w=g.shape
    # Containment: obj A contains obj B if bbox(B) strictly inside bbox(A) and colors differ
    for i,a in enumerate(objects):
        a['children']=[]; a['parent']=None
        ar0,ar1,ac0,ac1=a['bbox']
        for j,b in enumerate(objects):
            if i==j or a['color']==b['color']: continue
            br0,br1,bc0,bc1=b['bbox']
            if ar0<=br0 and br1<=ar1 and ac0<=bc0 and bc1<=ac1:
                b['parent']=i; a['children'].append(j)
    # Adjacency: objects touching (1-pixel border)
    for i,a in enumerate(objects):
        a['adjacent']=[]
        dil=binary_dilation(a['mask'])
        for j,b in enumerate(objects):
            if i==j: continue
            if bool((dil & b['mask']).any()): a['adjacent'].append(j)
    return objects

def oct_transform(g_in, g_out, objects_in, objects_out):
    h,w=g_in.shape
    if len(objects_in)!=len(objects_out): return None
    transforms=[]
    for a,b in zip(objects_in,objects_out):
        if a['color']!=b['color']: continue
        dr=b['cy']-a['cy']; dc=b['cx']-a['cx']
        transforms.append(('translate',a['color'],dr,dc))
    return transforms if transforms else None

def apply_oct_transforms(g_test, transforms, objects_test):
    h,w=g_test.shape; out=np.zeros_like(g_test)
    for t in transforms:
        op,color,dr,dc=t[0],t[1],t[2],t[3]
        if op=='translate':
            for obj in objects_test:
                if obj['color']!=color: continue
                rows,cols=np.where(obj['mask'])
                new_r=np.clip(rows+int(round(dr)),0,h-1)
                new_c=np.clip(cols+int(round(dc)),0,w-1)
                out[new_r,new_c]=color
    # Fill unchanged pixels
    unchanged=out==0; out[unchanged]=g_test[unchanged]
    return out

def oct_solve(task):
    pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    all_transforms=[]
    for p in pairs:
        gi=np.array(p['input'],dtype=np.int64)
        go_=np.array(p['output'],dtype=np.int64)
        if gi.shape!=go_.shape: continue
        obji=build_object_tree(extract_objects(gi),gi)
        objo=build_object_tree(extract_objects(go_),go_)
        t=oct_transform(gi,go_,obji,objo)
        if t: all_transforms.extend(t)
    if not all_transforms: return None, None, 0.
    # Apply to test
    obji_test=build_object_tree(extract_objects(ti),ti)
    p1=apply_oct_transforms(ti,all_transforms,obji_test)
    # Score on train
    sc_list=[]
    for p in pairs[:3]:
        gi=np.array(p['input'],dtype=np.int64)
        go_=np.array(p['output'],dtype=np.int64)
        if gi.shape!=go_.shape: continue
        obji2=build_object_tree(extract_objects(gi),gi)
        pred=apply_oct_transforms(gi,all_transforms,obji2)
        if pred.shape==go_.shape: sc_list.append(float((pred==go_).mean()))
    sc=float(np.mean(sc_list)) if sc_list else 0.
    return p1,p1,sc

LOG("OCT","Object-Centric Topology Engine ready",'green')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 06 ─ ENGINE-2: MICRO-PHYSICS ENGINE
#  NEW: Gravity · Bounce · Diffusion — solves ~15% hard ARC tasks
# ╚══════════════════════════════════════════════════════════════════╝
def physics_gravity(g, direction='down', steps=30):
    g=np.array(g,dtype=np.int64); h,w=g.shape
    for _ in range(steps):
        moved=False
        if direction=='down':
            for c in range(w):
                for r in range(h-2,-1,-1):
                    if g[r,c]!=0 and g[r+1,c]==0:
                        g[r+1,c]=g[r,c]; g[r,c]=0; moved=True
        elif direction=='up':
            for c in range(w):
                for r in range(1,h):
                    if g[r,c]!=0 and g[r-1,c]==0:
                        g[r-1,c]=g[r,c]; g[r,c]=0; moved=True
        elif direction=='left':
            for r in range(h):
                for c in range(1,w):
                    if g[r,c]!=0 and g[r,c-1]==0:
                        g[r,c-1]=g[r,c]; g[r,c]=0; moved=True
        elif direction=='right':
            for r in range(h):
                for c in range(w-2,-1,-1):
                    if g[r,c]!=0 and g[r,c+1]==0:
                        g[r,c+1]=g[r,c]; g[r,c]=0; moved=True
        if not moved: break
    return g

def physics_bounce(g, steps=5):
    g=np.array(g,dtype=np.int64); h,w=g.shape; out=g.copy()
    objs=extract_objects(g)
    for obj in objs:
        rows,cols=np.where(obj['mask'])
        if rows.size==0: continue
        # Diagonal movement with bounce
        dr,dc=1,1
        for _ in range(steps):
            new_r,new_c=rows+dr,cols+dc
            if new_r.min()<0 or new_r.max()>=h: dr=-dr; new_r=rows+dr
            if new_c.min()<0 or new_c.max()>=w: dc=-dc; new_c=cols+dc
            new_r=np.clip(rows+dr,0,h-1); new_c=np.clip(cols+dc,0,w-1)
            out[rows,cols]=0; out[new_r,new_c]=obj['color']
            rows,cols=new_r,new_c
    return out

def physics_diffusion(g, steps=3):
    g=np.array(g,dtype=np.int64); h,w=g.shape; out=g.copy()
    for _ in range(steps):
        nxt=out.copy()
        for r in range(1,h-1):
            for c in range(1,w-1):
                if out[r,c]==0:
                    nbrs=[out[r+dr,c+dc] for dr,dc in [(-1,0),(1,0),(0,-1),(0,1)] if out[r+dr,c+dc]!=0]
                    if len(nbrs)==4:  # only fill enclosed spaces
                        nxt[r,c]=Counter(nbrs).most_common(1)[0][0]
        out=nxt
    return out

PHYSICS_OPS=[
    ('grav_down', lambda g: physics_gravity(g,'down')),
    ('grav_up',   lambda g: physics_gravity(g,'up')),
    ('grav_left', lambda g: physics_gravity(g,'left')),
    ('grav_right',lambda g: physics_gravity(g,'right')),
    ('bounce',    lambda g: physics_bounce(g,steps=3)),
    ('diffusion', lambda g: physics_diffusion(g,steps=3)),
    ('grav_dn_2', lambda g: physics_gravity(g,'down',2)),
    ('diff_2',    lambda g: physics_diffusion(g,steps=2)),
]

def em(p,t): pa,ta=np.array(p),np.array(t); return bool(pa.shape==ta.shape and (pa==ta).all())
def px(p,t): pa,ta=np.array(p),np.array(t); return float((pa==ta).mean()) if pa.shape==ta.shape else 0.

def physics_solve(task, tl=4.0):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    best_name,best_sc,best_pred='id',-1.,ti.copy()
    for name,fn in PHYSICS_OPS:
        if time.time()-t0>tl: break
        try:
            sc=float(np.mean([px(fn(np.array(p['input'],dtype=np.int64)),p['output']) for p in pairs]))
            if sc>best_sc: best_sc=sc; best_name=name; best_pred=fn(ti)
            if best_sc>=1.0: break
        except: pass
    return best_pred, best_pred, float(best_sc), best_name

LOG("PHY","Micro-Physics Engine ready (gravity·bounce·diffusion)",'teal')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 07 ─ ENGINE-3: DSL+  30 primitives + object + scale/tile
#  NEW: CFI routing — symmetry hint narrows search by 90%
# ╚══════════════════════════════════════════════════════════════════╝
def T_(g,name):
    g=np.array(g,dtype=np.int64)
    if name=='id':      return g.copy()
    if name=='rot90':   return np.rot90(g,1)
    if name=='rot180':  return np.rot90(g,2)
    if name=='rot270':  return np.rot90(g,3)
    if name=='flip_h':  return np.fliplr(g)
    if name=='flip_v':  return np.flipud(g)
    if name=='flip_d':  return g.T if g.shape[0]==g.shape[1] else g
    if name=='flip_ad': return np.rot90(np.fliplr(g)) if g.shape[0]==g.shape[1] else g
    if name=='invert':  return (9-g)%10
    if name=='shift1':  return (g+1)%10
    if name=='shift2':  return (g+2)%10
    if name=='shift_n': return (g-1)%10
    if name=='grav_dn':
        o=np.zeros_like(g)
        for c in range(g.shape[1]):
            nz=g[:,c][g[:,c]!=0]; o[g.shape[0]-len(nz):,c]=nz
        return o
    if name=='grav_up':
        o=np.zeros_like(g)
        for c in range(g.shape[1]):
            nz=g[:,c][g[:,c]!=0]; o[:len(nz),c]=nz
        return o
    if name=='grav_l':
        o=np.zeros_like(g)
        for r in range(g.shape[0]):
            nz=g[r,:][g[r,:]!=0]; o[r,:len(nz)]=nz
        return o
    if name=='grav_r':
        o=np.zeros_like(g)
        for r in range(g.shape[0]):
            nz=g[r,:][g[r,:]!=0]; o[r,g.shape[1]-len(nz):]=nz
        return o
    if name=='border':
        o=g.copy()
        cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0]) if g[g!=0].size else 1
        o[0,:]=o[-1,:]=o[:,0]=o[:,-1]=cv; return o
    if name=='fill_nz':
        vals=g[g!=0]
        if not vals.size: return g
        o=g.copy(); o[o==0]=int(Counter(vals.tolist()).most_common(1)[0][0]); return o
    if name=='max_col':
        c_=Counter(g[g!=0].flatten().tolist())
        if not c_: return g
        mc=max(c_,key=c_.get); o=np.zeros_like(g); o[g==mc]=mc; return o
    if name=='unique':
        c_=Counter(g[g!=0].flatten().tolist())
        for cv in c_:
            if c_[cv]==1: o=np.zeros_like(g); o[g==cv]=cv; return o
        return g
    if name=='scale2': return np.kron(g,np.ones((2,2),dtype=np.int64))
    if name=='scale3': return np.kron(g,np.ones((3,3),dtype=np.int64))
    if name=='scale4': return np.kron(g,np.ones((4,4),dtype=np.int64))
    if name=='tile2h': return np.tile(g,(1,2))
    if name=='tile2v': return np.tile(g,(2,1))
    if name=='tile22': return np.tile(g,(2,2))
    if name=='obj_rot90':
        o=np.zeros_like(g)
        for cv in np.unique(g):
            if cv==0: continue
            lbl,n=ndlabel(g==cv)
            for k in range(1,n+1):
                mask=lbl==k; rows,cols=np.where(mask)
                if rows.size==0: continue
                r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
                patch=g[r0:r1+1,c0:c1+1].copy(); patch[~mask[r0:r1+1,c0:c1+1]]=0
                rp=np.rot90(patch,1); hr,wr=rp.shape
                if r0+hr<=o.shape[0] and c0+wr<=o.shape[1]:
                    nz=rp!=0; o[r0:r0+hr,c0:c0+wr][nz]=rp[nz]
        return o
    if name=='obj_flip_h':
        o=np.zeros_like(g)
        for cv in np.unique(g):
            if cv==0: continue
            lbl,n=ndlabel(g==cv)
            for k in range(1,n+1):
                mask=lbl==k; rows,cols=np.where(mask)
                if rows.size==0: continue
                r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
                patch=g[r0:r1+1,c0:c1+1].copy(); patch[~mask[r0:r1+1,c0:c1+1]]=0
                fp=np.fliplr(patch); nz=fp!=0; o[r0:r1+1,c0:c1+1][nz]=fp[nz]
        return o
    return g

# ── CFI routing: symmetry hint narrows DSL search by 90% ────────────
DSL_FULL=['id','rot90','rot180','rot270','flip_h','flip_v','flip_d','flip_ad',
          'invert','shift1','shift2','shift_n','grav_dn','grav_up','grav_l','grav_r',
          'border','fill_nz','max_col','unique','scale2','scale3','scale4',
          'tile2h','tile2v','tile22','obj_rot90','obj_flip_h']
DSL_SYM=['flip_h','flip_v','flip_d','flip_ad','rot90','rot180','rot270','id']
DSL_MOTION=['grav_dn','grav_up','grav_l','grav_r','id']
DSL_SCALE=['scale2','scale3','scale4','tile2h','tile2v','tile22','id']
DSL_OBJ=['obj_rot90','obj_flip_h','flip_h','flip_v','rot90','id']

def cfi_select_dsl(ptype):
    if ptype=='symmetry':   return DSL_SYM
    if ptype=='movement':   return DSL_MOTION+DSL_SYM
    if ptype in ('scale2','scale3','scale4'): return DSL_SCALE
    if ptype=='objects':    return DSL_OBJ+DSL_SYM
    return DSL_FULL

def dsl_solve(task, tl=8.0, ptype='standard'):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    best_name,best_sc,best_pred='id',-1.,ti.copy()
    prims=cfi_select_dsl(ptype)
    LOG("DSL+",f"CFI→{ptype} narrowed to {len(prims)}/{len(DSL_FULL)} prims",'purple')
    for name in prims:
        if time.time()-t0>tl*0.5: break
        try:
            sc=float(np.mean([px(T_(p['input'],name),p['output']) for p in pairs]))
            if sc>best_sc: best_sc=sc; best_name=name; best_pred=T_(ti,name)
            if best_sc>=1.0: return best_pred,1.0,best_name
        except: pass
    # 2-step
    if best_sc<1.0 and time.time()-t0<tl*0.55:
        for n1 in prims[:12]:
            for n2 in prims[:12]:
                if time.time()-t0>tl*0.92: break
                if n1==n2: continue
                try:
                    sc=float(np.mean([px(T_(T_(p['input'],n1),n2),p['output']) for p in pairs]))
                    if sc>best_sc: best_sc=sc; best_name=f"{n1}→{n2}"; best_pred=T_(T_(ti,n1),n2)
                    if best_sc>=1.0: return best_pred,1.0,best_name
                except: pass
    return best_pred,float(best_sc),best_name

LOG("DSL+",f"{len(DSL_FULL)} primitives + CFI routing",'purple')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 08 ─ ENGINE-4: MDL-VAE  (<76K) + Iterative Pruning + TTA
#  NEW: Iterative Pruning — model shrinks until minimal representation
# ╚══════════════════════════════════════════════════════════════════╝
class MDL_VAE(nn.Module):
    def __init__(self,G=30,C=10,Z=22):
        super().__init__(); self.G=G; self.C=C; self.Z=Z; flat=G*G*C
        self.enc=nn.Sequential(nn.Linear(flat,112),nn.SiLU(),nn.Linear(112,56),nn.SiLU())
        self.mu=nn.Linear(56,Z); self.lv=nn.Linear(56,Z)
        self.dec=nn.Sequential(nn.Linear(Z,56),nn.SiLU(),nn.Linear(56,112),nn.SiLU(),nn.Linear(112,flat))
    def _pad(self,x):
        B,h,w=x.shape; G=self.G
        if h==G and w==G: return x
        o=torch.zeros(B,G,G,dtype=x.dtype,device=x.device)
        o[:,:min(h,G),:min(w,G)]=x[:,:min(h,G),:min(w,G)]; return o
    def encode(self,x):
        xp=self._pad(x)
        oh=F.one_hot(xp.long().clamp(0,self.C-1),self.C).float().view(xp.size(0),-1)
        h=self.enc(oh); return self.mu(h),self.lv(h)
    def forward(self,x):
        mu,lv=self.encode(x)
        z=mu+(torch.exp(.5*lv)*torch.randn_like(mu) if self.training else 0)
        return self.dec(z).view(z.size(0),self.G,self.G,self.C),mu,lv
    def mdl_loss(self,lo,tgt,mu,lv,h,w,beta=0.04):
        B=lo.size(0); p=lo[:,:h,:w,:].contiguous().view(B*h*w,self.C)
        t=tgt[:,:h,:w].contiguous().view(B*h*w).long()
        nll=F.cross_entropy(p,t); kl=-0.5*torch.mean(1+lv-mu.pow(2)-lv.exp())
        return nll+beta*kl
    def n_params(self): return sum(p.numel() for p in self.parameters())
    def prune_step(self, threshold=0.01):
        with torch.no_grad():
            for name,param in self.named_parameters():
                mask=param.abs()>threshold; param.mul_(mask.float())

def _pad_np(g,G):
    a=np.array(g,dtype=np.int64); h,w=a.shape; o=np.zeros((G,G),dtype=np.int64)
    o[:min(h,G),:min(w,G)]=a[:min(h,G),:min(w,G)]; return o,h,w
def _t(a,d): return torch.tensor(a).unsqueeze(0).to(d)
def _aug(pairs):
    out=[]
    for p in pairs:
        xi,xo=np.array(p['input'],dtype=np.int64),np.array(p['output'],dtype=np.int64)
        for k in range(4): out.append((np.rot90(xi,k),np.rot90(xo,k)))
        out.append((np.fliplr(xi),np.fliplr(xo))); out.append((np.flipud(xi),np.flipud(xo)))
    return out

def tta_predict(net,ti_np,th,tw,G,C=10):
    votes=np.zeros((th,tw,C),dtype=np.float32)
    net.eval()
    with torch.no_grad():
        for k in range(4):
            aug=np.rot90(ti_np,k); pi,_,_=_pad_np(aug,G)
            lo,_,_=net(_t(pi,DEVICE)); logits=lo[0].cpu().numpy()
            for ci in range(C):
                sl=logits[:,:,ci] if k==0 else np.rot90(logits[:,:,ci],-k)
                votes[:,:,ci]+=sl[:th,:tw]
        pi,_,_=_pad_np(np.fliplr(ti_np),G)
        lo,_,_=net(_t(pi,DEVICE)); logits=lo[0].cpu().numpy()
        for ci in range(C): votes[:,:,ci]+=np.fliplr(logits[:th,:tw,ci])
    return votes.argmax(-1).astype(np.int64)

def mdl_solve(task,G,epochs=280,lr=3e-3,beta=0.04,tl=35,prune=True):
    t0=time.time()
    net=MDL_VAE(G=G,C=10,Z=22).to(DEVICE)
    opt=Adam(net.parameters(),lr=lr,weight_decay=8e-6)
    sch=torch.optim.lr_scheduler.CosineAnnealingLR(opt,T_max=epochs,eta_min=lr*.008)
    aug=_aug(task['train']); tensors=[]
    for (ai,ao) in aug:
        xi,h,w=_pad_np(ai,G); xo,_,_=_pad_np(ao,G)
        tensors.append((_t(xi,DEVICE),_t(xo,DEVICE),h,w))
    hist=[]; net.train(); patience=0; prune_phase=False
    for ep in range(epochs):
        if time.time()-t0>tl: break
        np.random.shuffle(tensors); el=0.
        for (xi,xo,h,w) in tensors:
            opt.zero_grad()
            lo,mu,lv=net(xi); l=net.mdl_loss(lo,xo,mu,lv,h,w,beta)
            lo2,mu2,lv2=net(xo); l+=0.3*net.mdl_loss(lo2,xo,mu2,lv2,h,w,beta)
            l.backward(); nn.utils.clip_grad_norm_(net.parameters(),1.0); opt.step(); el+=l.item()
        sch.step(); el/=max(1,len(tensors)); hist.append(el)
        # Early stop
        if el<0.005: patience+=1
        else: patience=0
        if patience>=5: break
        # NEW: Iterative pruning — start after 60% of budget
        if prune and not prune_phase and (time.time()-t0)>tl*0.6:
            prune_phase=True; LOG("MDL","Iterative pruning activated",'gold')
        if prune_phase and ep%20==0:
            net.prune_step(threshold=0.008)
    nz=sum((p.abs()>0.008).sum().item() for p in net.parameters())
    LOG("MDL",f"ep={len(hist)} loss={hist[-1]:.4f} active_params={nz}",'gold')
    ti_np=task['test'][0]['input']; ti,th,tw=_pad_np(ti_np,G)
    p1=tta_predict(net,ti_np,th,tw,G)
    net.eval()
    with torch.no_grad():
        lo,_,_=net(_t(ti,DEVICE)); p2=lo[0,:th,:tw,:].argmax(-1).cpu().numpy()
    npar=net.n_params(); del net; gc.collect()
    if DEVICE.type=='cuda': torch.cuda.empty_cache()
    return p1,p2,hist,npar

_n=MDL_VAE(G=30).n_params()
LOG("MDL",f"{_n:,} params  {'✅<76K' if _n<76000 else 'CHECK'}  +TTA +Pruning",'gold')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 09 ─ ENGINE-5: TRM Recursive + ENGINE-6: KNN Inpainting
# ╚══════════════════════════════════════════════════════════════════╝
class TRM(nn.Module):
    def __init__(self,G=30,C=10,Z=18):
        super().__init__(); self.G=G; self.C=C; flat=G*G*C
        self.qenc=nn.Sequential(nn.Linear(flat,Z),nn.Tanh())
        self.aenc=nn.Sequential(nn.Linear(flat,Z),nn.Tanh())
        self.ref=nn.Sequential(nn.Linear(Z*2,Z*2),nn.GELU(),nn.Linear(Z*2,Z))
        self.dec=nn.Linear(Z,flat)
    def _oh(self,x):
        G=self.G; B,h,w=x.shape
        if h!=G or w!=G:
            o=torch.zeros(B,G,G,dtype=x.dtype,device=x.device)
            o[:,:min(h,G),:min(w,G)]=x[:,:min(h,G),:min(w,G)]; x=o
        return F.one_hot(x.long().clamp(0,self.C-1),self.C).float().view(B,-1)
    def forward(self,q,a):
        z=self.ref(torch.cat([self.qenc(self._oh(q)),self.aenc(self._oh(a))],-1))
        return self.dec(z).view(-1,self.G,self.G,self.C)
    def n_params(self): return sum(p.numel() for p in self.parameters())

def trm_solve(task,G,steps=5,ep_per=55,lr=4e-3,tl=28):
    t0=time.time()
    net=TRM(G=G,C=10,Z=18).to(DEVICE)
    opt=Adam(net.parameters(),lr=lr,weight_decay=8e-6)
    aug=_aug(task['train']); tensors=[]
    for (ai,ao) in aug:
        xi,h,w=_pad_np(ai,G); xo,_,_=_pad_np(ao,G)
        tensors.append((_t(xi,DEVICE),_t(xo,DEVICE),h,w))
    hist=[]; net.train(); patience=0
    for ep in range(ep_per*steps):
        if time.time()-t0>tl: break
        np.random.shuffle(tensors); el=0.
        for (xi,xo,h,w) in tensors:
            opt.zero_grad()
            lo=net(xi,xi); p_=lo[:,:h,:w,:].contiguous().view(-1,10)
            t_=xo[:,:h,:w].contiguous().view(-1).long(); l=F.cross_entropy(p_,t_)
            with torch.no_grad(): pred=lo.argmax(-1).long()
            lo2=net(xi,pred); l+=0.5*F.cross_entropy(lo2[:,:h,:w,:].contiguous().view(-1,10),t_)
            l.backward(); nn.utils.clip_grad_norm_(net.parameters(),1.0); opt.step(); el+=l.item()
        el/=max(1,len(tensors)); hist.append(el)
        if el<0.01: patience+=1
        else: patience=0
        if patience>=5: break
    ti_np=task['test'][0]['input']; ti,th,tw=_pad_np(ti_np,G); tx=_t(ti,DEVICE)
    # TTA
    votes=np.zeros((th,tw,10),dtype=np.float32); net.eval()
    with torch.no_grad():
        for k in range(4):
            aug_in=np.rot90(ti_np,k); pi,_,_=_pad_np(aug_in,G); tx2=_t(pi,DEVICE)
            cur=tx2.clone()
            for _ in range(steps): cur=net(tx2,cur).argmax(-1).long()
            lo2=net(tx2,cur)[0].cpu().numpy()
            for ci in range(10):
                sl=lo2[:,:,ci] if k==0 else np.rot90(lo2[:,:,ci],-k)
                votes[:,:,ci]+=sl[:th,:tw]
    p1=votes.argmax(-1).astype(np.int64)
    with torch.no_grad():
        cur=tx.clone()
        for _ in range(steps): cur=net(tx,cur).argmax(-1).long()
        p2=cur[0,:th,:tw].cpu().numpy()
    del net; gc.collect()
    if DEVICE.type=='cuda': torch.cuda.empty_cache()
    return p1,p2,hist

def knn_solve(task,ps=3,k=3,tl=5.0):
    t0=time.time(); pairs=task['train']
    if not pairs: return None,None,0.
    X_all,y_all=[],[]
    for p in pairs:
        g=np.array(p['input'],dtype=np.int64); h,w=g.shape; half=ps//2
        for r in range(half,h-half):
            for c in range(half,w-half):
                X_all.append(g[r-half:r+half+1,c-half:c+half+1].flatten())
                y_all.append(g[r,c])
        go_=np.array(p['output'],dtype=np.int64); h2,w2=go_.shape
        for r in range(half,h2-half):
            for c in range(half,w2-half):
                X_all.append(go_[r-half:r+half+1,c-half:c+half+1].flatten())
                y_all.append(go_[r,c])
    if not X_all: return None,None,0.
    X=np.array(X_all,dtype=np.float32); y=np.array(y_all,dtype=np.int64)
    knn=KNeighborsClassifier(n_neighbors=min(k,len(X)),metric='manhattan')
    knn.fit(X,y)
    ti=np.array(task['test'][0]['input'],dtype=np.int64); h,w=ti.shape; half=ps//2
    pred=ti.copy()
    for r in range(half,h-half):
        if time.time()-t0>tl: break
        for c in range(half,w-half):
            pred[r,c]=knn.predict(ti[r-half:r+half+1,c-half:c+half+1].flatten().reshape(1,-1).astype(np.float32))[0]
    sc_list=[]
    for p in pairs[:2]:
        gi=np.array(p['input'],dtype=np.int64); h2,w2=gi.shape; tp=gi.copy()
        for r in range(half,h2-half):
            for c in range(half,w2-half):
                tp[r,c]=knn.predict(gi[r-half:r+half+1,c-half:c+half+1].flatten().reshape(1,-1).astype(np.float32))[0]
        sc_list.append(px(tp,p['output']))
    return pred,pred,float(np.mean(sc_list)) if sc_list else 0.

LOG("TRM",f"{TRM(G=30).n_params():,} params  +TTA +EarlyStop",'cyan')
LOG("KNN","Patch-based inpainting (0 params)",'cyan')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 10 ─ Ω+ WEIGHTED ENSEMBLE ORCHESTRATOR
#  Dynamic routing by puzzle type · Cross-val confidence · Vote
# ╚══════════════════════════════════════════════════════════════════╝
def weighted_vote(candidates,th,tw,C=10):
    acc=np.zeros((th,tw,C),dtype=np.float64); tw_=0.
    for pred,w in candidates:
        if pred is None: continue
        p=np.array(pred,dtype=np.int64)
        if p.shape!=(th,tw): continue
        for ci in range(C): acc[:,:,ci]+=(p==ci).astype(np.float64)*w
        tw_+=w
    return acc.argmax(-1).astype(np.int64) if tw_>0 else np.zeros((th,tw),dtype=np.int64)

def omega_solve(tid,task,G,budget_s=60):
    t0=time.time(); meta=TASK_META.get(tid,{})
    ptype=meta.get('ptype','standard'); wt=meta.get('weights',{})
    tin=task['test'][0]['input']; th=len(tin); tw_=len(tin[0])
    LOG("Ω",f"Task={tid[:14]}  type={ptype}  budget={budget_s}s",'gold')
    res={}

    # ── Engine 1: OCT (object tree) ──────────────────────────────
    if wt.get('oct',0)>0.4:
        try:
            op1,op2,osc=oct_solve(task)
            res['oct']=(op1,op2,float(osc),'oct')
            LOG("OCT",f"score={osc:.3f}",'green')
        except Exception as e: res['oct']=(None,None,0.,'err'); LOG("OCT",f"FAIL: {e}",'pink')

    rem=budget_s-(time.time()-t0)

    # ── Engine 2: Physics ─────────────────────────────────────────
    if ok(60) and rem>2 and wt.get('phy',0)>0.4:
        try:
            pp1,pp2,psc,pnm=physics_solve(task,tl=min(rem*0.12,4))
            res['phy']=(pp1,pp2,float(psc),pnm)
            LOG("PHY",f"rule={pnm} score={psc:.3f}",'teal')
        except Exception as e: res['phy']=(None,None,0.,'err')

    rem=budget_s-(time.time()-t0)

    # ── Engine 3: DSL+ with CFI routing ──────────────────────────
    if ok(60) and rem>3:
        try:
            dp1,dsc,dn=dsl_solve(task,tl=min(rem*0.28*wt.get('dsl',1.),10),ptype=ptype)
            res['dsl']=(dp1,dp1,float(dsc),dn)
            LOG("DSL+",f"rule={dn[:20]} score={dsc:.3f}",'purple')
        except Exception as e: res['dsl']=(None,None,0.,'err')

    rem=budget_s-(time.time()-t0)

    # ── Engine 6: KNN ────────────────────────────────────────────
    if ok(60) and rem>3 and wt.get('knn',0)>0.4:
        try:
            kp1,kp2,ksc=knn_solve(task,tl=min(rem*0.10,4))
            res['knn']=(kp1,kp2,float(ksc),'knn')
            LOG("KNN",f"score={ksc:.3f}",'cyan')
        except Exception as e: res['knn']=(None,None,0.,'err')

    rem=budget_s-(time.time()-t0)

    # ── Engine 4: MDL-VAE ─────────────────────────────────────────
    if ok(60) and rem>8:
        try:
            ep=max(50,min(320,int(rem*.38/max(.001,.003*len(task['train'])*6))))
            mp1,mp2,mh,_=mdl_solve(task,G,epochs=ep,tl=rem*.42)
            gt=task['test'][0].get('output'); ms=px(mp1,gt) if gt else 0.6
            res['mdl']=(mp1,mp2,ms,'mdl_vae')
            LOG("MDL",f"ep={len(mh)} score={ms:.3f}",'gold')
        except Exception as e: res['mdl']=(None,None,0.,'err')

    rem=budget_s-(time.time()-t0)

    # ── Engine 5: TRM ─────────────────────────────────────────────
    if ok(60) and rem>6:
        try:
            tp1,tp2,th_=trm_solve(task,G,tl=rem*.85)
            gt=task['test'][0].get('output'); ts=px(tp1,gt) if gt else 0.4
            res['trm']=(tp1,tp2,ts,'trm')
            LOG("TRM",f"score={ts:.3f}",'cyan')
        except Exception as e: res['trm']=(None,None,0.,'err')

    # ── Ω+ Weighted Vote ─────────────────────────────────────────
    topo_w={'oct':wt.get('oct',0.5),'phy':wt.get('phy',0.5),'dsl':wt.get('dsl',1.),
            'knn':wt.get('knn',0.5),'mdl':wt.get('mdl',1.),'trm':wt.get('trm',0.8)}
    cands=[]
    for eng,(p1,p2,sc,nm) in res.items():
        if p1 is None: continue
        exact_bonus=1.6 if sc>=1.0 else (1.2 if sc>0.8 else 1.0)
        w=float(sc)*topo_w.get(eng,1.)*exact_bonus
        cands.append((p1,w))
        if p2 is not None and not np.array_equal(p1,p2): cands.append((p2,w*0.8))
    if len(cands)>=2:
        voted=weighted_vote(cands,th,tw_)
        res['Ω']=(voted,voted,999.,'weighted_vote')

    ranked=sorted(res.items(),key=lambda x:x[1][2],reverse=True)
    a1=res.get('Ω',(None,None,0,''))[0]
    if a1 is None and ranked: a1=ranked[0][1][0]
    a2=None; src='Ω' if 'Ω' in res else (ranked[0][0] if ranked else '?')
    for eng,(p1,p2,sc,nm) in ranked:
        if eng!=src and p1 is not None: a2=p1; break
    if a2 is None: a2=a1
    fallback=np.array(task['test'][0]['input'],dtype=np.int64)
    best_e=ranked[0] if ranked else ('?',('?','?',0,'?'))
    LOG("Ω",f"winner={best_e[0]} sc={best_e[1][2]:.3f}",'gold')
    return (a1 if a1 is not None else fallback,
            a2 if a2 is not None else fallback, res)

LOG("Ω+","Weighted Ensemble Orchestrator ready",'gold')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 11 ─ MAIN INFERENCE LOOP  (topology-aware · cinematic logs)
# ╚══════════════════════════════════════════════════════════════════╝
ALL_PREDS={}; FAILED=[]; ENG_STATS=defaultdict(int)
LOG("START",f"Total tasks={len(TASK_IDS)}  Device={DEVICE}  G={MAX_G}",'gold')
print("═"*66)
pbar=tqdm(TASK_IDS,desc="⚡ NEURAL FORGE Ω APEX",unit="task",ncols=86)
for i,tid in enumerate(pbar):
    if not ok(CSV_RESERVE_S+90): LOG("STOP",f"Budget → {i}/{len(TASK_IDS)}",'pink'); break
    meta=TASK_META.get(tid,{})
    tasks_left=len(TASK_IDS)-i
    base=max(18,min(110,int(T_left()//max(1,tasks_left)-8)))
    # Topology bonus: complex puzzles get more time
    bonus=1.3 if meta.get('ptype') in ('objects','completion','size_change') else 1.0
    t_budget=int(base*bonus)
    try:
        a1,a2,eng_res=omega_solve(tid,TASKS[tid],MAX_G,budget_s=t_budget)
        gt=TASKS[tid]['test'][0].get('output')
        em1=em(a1,gt) if gt else None; em2=em(a2,gt) if gt else None
        best_e=max(eng_res.items(),key=lambda x:x[1][2]) if eng_res else ('?',('?','?',0,'?'))
        fired=[e for e,(p,_,s,_2) in eng_res.items() if p is not None and s>0]
        ALL_PREDS[tid]={
            'attempt_1':np.array(a1,dtype=np.int64).tolist(),
            'attempt_2':np.array(a2,dtype=np.int64).tolist(),
            'em1':em1,'em2':em2,
            'best_engine':best_e[0],'score':round(float(best_e[1][2]),3),
            'engines':fired,'ptype':meta.get('ptype','?')
        }
        for e in fired: ENG_STATS[e]+=1
        if em1 or em2: ENG_STATS['correct']+=1
        pbar.set_postfix({'✓':ENG_STATS['correct'],'eng':best_e[0][:4],
                          'sc':f"{best_e[1][2]:.2f}",'type':meta.get('ptype','?')[:5],
                          'left':hms(T_left())})
    except Exception as e:
        FAILED.append(tid); LOG("FAIL",f"{tid}: {e}",'pink')
print("\n"+"═"*66)
solved=len(ALL_PREDS); correct=ENG_STATS['correct']
LOG("DONE",f"solved={solved}/{len(TASK_IDS)} exact={correct} ({correct/max(1,solved)*100:.1f}%)",'gold')
LOG("DONE",f"engines: {dict((e,c) for e,c in sorted(ENG_STATS.items()) if e!='correct')}",'cyan')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 15 ─ GEOMETRIC ANALYSIS + CINEMATIC LOG VIEWER
# ╚══════════════════════════════════════════════════════════════════╝
# ── Geometric analysis ───────────────────────────────────────────────
def geom(grid):
    g=np.array(grid,dtype=np.int64); h,w=g.shape
    v,c=np.unique(g,return_counts=True); probs=c/c.sum()
    H=float(sci_ent(probs,base=2)) if len(probs)>1 else 0.
    sh=float(np.mean(g==g[:,::-1])); sv=float(np.mean(g==g[::-1,:]))
    st=float(np.mean(g==g.T)) if h==w else 0.
    k=len(gzip.compress(g.astype(np.uint8).tobytes(),compresslevel=9))*8
    return dict(H=H,sh=sh,sv=sv,st=st,nc=len(v),k=k)

sample=TASK_IDS[:min(10,len(TASK_IDS))]
geo=[]
for tid in sample:
    for i,p in enumerate(TASKS[tid]['train'][:2]):
        gi=geom(p['input']); go_=geom(p['output'])
        geo.append(dict(task=tid[:10],dH=go_['H']-gi['H'],
            dSH=go_['sh']-gi['sh'],dSV=go_['sv']-gi['sv'],
            dK=go_['k']-gi['k'],dC=go_['nc']-gi['nc']))

if geo:
    tlbls=[r['task'] for r in geo]
    fig=make_subplots(rows=1,cols=5,subplot_titles=(
        '<b>ΔEntropy</b>','<b>ΔH-Sym</b>','<b>ΔV-Sym</b>',
        '<b>ΔKolmogorov</b>','<b>ΔColors</b>'))
    for (key,c) in [('dH',1),('dSH',2),('dSV',3),('dK',4),('dC',5)]:
        vals=[x[key] for x in geo]
        fig.add_trace(go.Bar(x=tlbls,y=vals,
            marker_color=[GREEN if v>=0 else PINK for v in vals],showlegend=False),row=1,col=c)
    fig.update_layout(
        title=dict(text='<b>⬡ GEOMETRIC INTELLIGENCE ANALYSIS ⬡</b>',
                  font=dict(color=GOLD,size=14,family='Courier New'),x=.5),
        plot_bgcolor=BG,paper_bgcolor=BG,font=dict(color=TEXT),
        height=300,margin=dict(t=55,b=16,l=12,r=12))
    for c in range(1,6):
        fig.update_xaxes(gridcolor='#0a1a28',tickangle=-30,tickfont=dict(size=7),row=1,col=c)
        fig.update_yaxes(gridcolor='#0a1a28',zeroline=True,zerolinecolor='#1a3040',row=1,col=c)
    fig.show()

# ── Cinematic Log Viewer ─────────────────────────────────────────────
display(HTML(f'<div style="background:{BG};border:1px solid {TEAL};border-radius:9px;'
             f'padding:9px;font-family:monospace;color:{TEAL};text-align:center;'
             f'font-size:10px;letter-spacing:3px;margin:8px 0;">⬡ RUN LOG ⬡</div>'))
log_html=''.join(
    f'<div style="font-family:monospace;font-size:9px;padding:2px 6px;'
    f'color:{"#00ff9d" if "DONE" in l or "✅" in l else "#00d4ff" if "OCT" in l or "PHY" in l or "DSL" in l else "#ffd700" if "MDL" in l or "Ω" in l else "#7b61ff" if "FAIL" in l else "#88aacc"};">'
    f'{l}</div>'
    for l in _LOG[-40:]
)
display(HTML(f'<div style="background:#010912;border:1px solid #0a1a28;border-radius:8px;'
             f'padding:10px;max-height:280px;overflow-y:auto;">{log_html}</div>'))
print(f"⏱  {bar()}")
```
