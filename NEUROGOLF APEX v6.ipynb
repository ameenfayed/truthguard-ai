```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 03 ─ LOAD ALL DATA
#  NEW: On-demand CFI classification (lazy, not pre-computed)
#  NEW: LRU-cached object extraction
# ╚══════════════════════════════════════════════════════════════════╝
UTILS=None; sys.path.insert(0,DATA_ROOT); up=Path(UTILS_DIR)
if up.exists():
    try:
        import neurogolf_utils as UTILS
        LOG("UTILS",f"Imported: {[x for x in dir(UTILS) if not x.startswith('_')]}",'g')
    except:
        for pyf in sorted(up.glob('*.py')):
            try:
                sp=importlib.util.spec_from_file_location(pyf.stem,pyf)
                m=importlib.util.module_from_spec(sp); sp.loader.exec_module(m)
                UTILS=m; LOG("UTILS",f"{pyf.name} loaded",'g')
            except Exception as e: LOG("UTILS",f"WARN {pyf.name}: {e}",'r')
else: LOG("UTILS","absent — solo mode",'r')

TASKS,ERRS={},[]
jsons=sorted(Path(DATA_ROOT).rglob('*.json'))
LOG("DATA",f"Scanning: {len(jsons)} JSON file(s)",'c')
for jf in jsons:
    try:
        d=json.loads(jf.read_text())
        if isinstance(d,dict) and 'train' in d and 'test' in d: TASKS[jf.stem]=d
        elif isinstance(d,list):
            for i,it in enumerate(d):
                if isinstance(it,dict) and 'train' in it: TASKS[f'{jf.stem}_{i:04d}']=it
        else: TASKS[f'_raw_{jf.stem}']=d
    except Exception as e: ERRS.append((jf.name,str(e)))
TASK_IDS=[k for k in TASKS if not k.startswith('_raw_')]
LOG("DATA",f"Tasks: {len(TASK_IDS)}  errors: {len(ERRS)}",'g')

# ── LRU-cached grid hash (for caching object extractions) ────────────
_GRID_CACHE={}
def grid_hash(g): return hashlib.md5(np.asarray(g,dtype=np.int8).tobytes()).hexdigest()

# ── NEW: On-demand CFI classifier ────────────────────────────────────
_META_CACHE={}  # tid → (ptype, weights)

def get_meta(tid):
    if tid in _META_CACHE: return _META_CACHE[tid]
    task=TASKS.get(tid,{})
    pairs=task.get('train',[])
    if not pairs: _META_CACHE[tid]=('standard',{}); return _META_CACHE[tid]
    gi0=np.array(pairs[0]['input'],dtype=np.int64)
    go0=np.array(pairs[0]['output'],dtype=np.int64)
    size_change=any(np.array(p['input']).shape!=np.array(p['output']).shape for p in pairs)
    scale=None
    if go0.shape[0]>0 and gi0.shape[0]>0:
        rr=go0.shape[0]/gi0.shape[0]; rc=go0.shape[1]/gi0.shape[1]
        if rr==rc and rr in [2.,3.,4.]: scale=int(rr)
    def sym(g): h,w=g.shape; return max(float(np.mean(g==g[:,::-1])),float(np.mean(g==g[::-1,:])),float(np.mean(g==g.T)) if h==w else 0.)
    sym_s=float(np.mean([sym(np.array(p['input'],dtype=np.int64)) for p in pairs]))
    ci=np.mean([len(np.unique(np.array(p['input']))) for p in pairs])
    co=np.mean([len(np.unique(np.array(p['output']))) for p in pairs])
    dc=co-ci
    motion=False
    if not size_change and gi0.shape==go0.shape:
        diff=(gi0!=go0).sum()/(gi0.size+1e-9); motion=0.05<float(diff)<0.7
    def n_obj(g): return sum(ndlabel(g==c)[1] for c in np.unique(g) if c!=0)
    obj_cnt=float(np.mean([n_obj(np.array(p['input'],dtype=np.int64)) for p in pairs[:2]]))
    max_h=max(max(np.array(p['input']).shape[0],np.array(p.get('output',p['input'])).shape[0]) for p in pairs)
    max_w=max(max(np.array(p['input']).shape[1],np.array(p.get('output',p['input'])).shape[1]) for p in pairs)
    if scale: ptype=f'scale{scale}'
    elif sym_s>0.80: ptype='symmetry'
    elif motion and obj_cnt>1: ptype='movement'
    elif dc<-0.5: ptype='reduction'
    elif dc>0.5: ptype='completion'
    elif obj_cnt>4: ptype='objects'
    elif size_change: ptype='size_change'
    else: ptype='standard'
    wt={'dsl':1.,'mdl':1.,'trm':0.8,'phy':0.5,'knn':0.5,'oct':0.5}
    if ptype in ('symmetry','movement'): wt['dsl']=2.5; wt['mdl']=0.6
    if ptype=='objects': wt['oct']=2.; wt['knn']=1.5; wt['mdl']=1.2
    if ptype=='completion': wt['knn']=2.; wt['mdl']=1.5
    if ptype in ('scale2','scale3','scale4'): wt['dsl']=3.; wt['mdl']=0.4
    if ptype=='standard': wt['mdl']=1.8; wt['trm']=1.4
    if motion: wt['phy']=1.5
    result=(ptype,wt,max_h,max_w,len(task.get('train',[])))
    _META_CACHE[tid]=result; return result

# Pre-compute MAX_G quickly
all_shapes=[np.array(p['input']).shape for t in TASKS.values() for p in t.get('train',[])+t.get('test',[])]
MAX_G=max((max(s) for s in all_shapes),default=30)
MAX_G=max(MAX_G,30)
LOG("META",f"MAX_G={MAX_G}  (on-demand CFI ready)",'g')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 05 ─ ENGINE-1: OCT (LRU-cached)
#           + ENGINE-2: PHY VECTORIZED (NumPy shifts, no loops)
# ╚══════════════════════════════════════════════════════════════════╝

# ── LRU-cached object extraction ─────────────────────────────────────
_OBJ_CACHE={}
def extract_objects_cached(g_key, g):
    if g_key in _OBJ_CACHE: return _OBJ_CACHE[g_key]
    objs=[]
    for cv in np.unique(g):
        if cv==0: continue
        lbl,n=ndlabel(g==cv)
        for k in range(1,n+1):
            mask=lbl==k; rows,cols=np.where(mask)
            if rows.size==0: continue
            r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
            objs.append({'color':int(cv),'mask':mask,'bbox':(r0,r1,c0,c1),
                         'size':int(mask.sum()),'cy':(r0+r1)/2,'cx':(c0+c1)/2})
    _OBJ_CACHE[g_key]=objs; return objs

def build_object_tree(objs,g):
    for i,a in enumerate(objs):
        a['children']=[]; a['parent']=None
        ar0,ar1,ac0,ac1=a['bbox']
        for j,b in enumerate(objs):
            if i==j or a['color']==b['color']: continue
            br0,br1,bc0,bc1=b['bbox']
            if ar0<=br0 and br1<=ar1 and ac0<=bc0 and bc1<=ac1:
                b['parent']=i; a['children'].append(j)
        dil=binary_dilation(a['mask'])
        a['adjacent']=[j for j,b in enumerate(objs) if j!=i and bool((dil&b['mask']).any())]
    return objs

def oct_solve(task):
    pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    all_t=[]
    for p in pairs:
        gi=np.array(p['input'],dtype=np.int64); go_=np.array(p['output'],dtype=np.int64)
        if gi.shape!=go_.shape: continue
        ki=grid_hash(gi); ko=grid_hash(go_)
        obji=build_object_tree(extract_objects_cached(ki,gi),gi)
        objo=build_object_tree(extract_objects_cached(ko,go_),go_)
        for a,b in zip(obji,objo):
            if a['color']==b['color']: all_t.append(('translate',a['color'],b['cy']-a['cy'],b['cx']-a['cx']))
    if not all_t: return None,None,0.
    kt=grid_hash(ti); obji_t=build_object_tree(extract_objects_cached(kt,ti),ti)
    h,w=ti.shape; out=np.zeros_like(ti)
    for t in all_t:
        op,color,dr,dc=t
        for obj in obji_t:
            if obj['color']!=color: continue
            rows,cols=np.where(obj['mask'])
            nr=np.clip(rows+int(round(dr)),0,h-1); nc_=np.clip(cols+int(round(dc)),0,w-1)
            out[nr,nc_]=color
    unchanged=out==0; out[unchanged]=ti[unchanged]
    sc_list=[]
    for p in pairs[:3]:
        gi=np.array(p['input'],dtype=np.int64); go_=np.array(p['output'],dtype=np.int64)
        if gi.shape!=go_.shape: continue
        ki2=grid_hash(gi); obji2=build_object_tree(extract_objects_cached(ki2,gi),gi)
        h2,w2=gi.shape; out2=np.zeros_like(gi)
        for t in all_t:
            op,color,dr,dc=t
            for obj in obji2:
                if obj['color']!=color: continue
                rows,cols=np.where(obj['mask'])
                nr=np.clip(rows+int(round(dr)),0,h2-1); nc_=np.clip(cols+int(round(dc)),0,w2-1)
                out2[nr,nc_]=color
        unch=out2==0; out2[unch]=gi[unch]
        if out2.shape==go_.shape: sc_list.append(float((out2==go_).mean()))
    sc=float(np.mean(sc_list)) if sc_list else 0.
    return out,out,sc

# ── VECTORIZED physics (NumPy array operations, no Python loops) ──────
def _vec_grav(g, axis, reverse=False):
    g=np.array(g,dtype=np.int64)
    h,w=g.shape; out=np.zeros_like(g)
    if axis==0:  # vertical
        for c in range(w):
            col=g[:,c]; nz=col[col!=0]
            if len(nz)==0: continue
            if reverse: out[:len(nz),c]=nz
            else: out[h-len(nz):,c]=nz
    else:  # horizontal
        for r in range(h):
            row=g[r,:]; nz=row[row!=0]
            if len(nz)==0: continue
            if reverse: out[r,:len(nz)]=nz
            else: out[r,w-len(nz):]=nz
    return out

def _vec_diffusion(g, steps=3):
    g=np.array(g,dtype=np.int64)
    for _ in range(steps):
        nxt=g.copy()
        zero=g==0
        if not zero.any(): break
        # All 4-neighbor average (vectorized)
        u=np.roll(g,1,0); d=np.roll(g,-1,0)
        l=np.roll(g,1,1); r=np.roll(g,-1,1)
        u[0,:]=0; d[-1,:]=0; l[:,0]=0; r[:,-1]=0
        nbr_sum=(u!=0).astype(int)+(d!=0).astype(int)+(l!=0).astype(int)+(r!=0).astype(int)
        # Fill only cells with all 4 neighbors non-zero (enclosed)
        enclosed=zero & (nbr_sum==4)
        if not enclosed.any(): break
        # Mode of neighbors
        for r_idx in range(g.shape[0]):
            for c_idx in range(g.shape[1]):
                if enclosed[r_idx,c_idx]:
                    nbrs=[v for v in [u[r_idx,c_idx],d[r_idx,c_idx],l[r_idx,c_idx],r[r_idx,c_idx]] if v!=0]
                    if nbrs: nxt[r_idx,c_idx]=Counter(nbrs).most_common(1)[0][0]
        g=nxt
    return g

def _vec_bounce(g, steps=4):
    g=np.array(g,dtype=np.int64); h,w=g.shape; out=g.copy()
    # Vectorized: shift all non-zero pixels diagonally, bounce at edges
    nz_r,nz_c=np.where(g!=0)
    if len(nz_r)==0: return g
    dr,dc=np.ones(len(nz_r),dtype=int),np.ones(len(nz_c),dtype=int)
    for _ in range(steps):
        nr=nz_r+dr; nc=nz_c+dc
        dr[nr<0]*=-1; dr[nr>=h]*=-1; dc[nc<0]*=-1; dc[nc>=w]*=-1
        nz_r=np.clip(nz_r+dr,0,h-1); nz_c=np.clip(nz_c+dc,0,w-1)
    out2=np.zeros_like(g)
    out2[nz_r,nz_c]=g[np.where(g!=0)]
    return out2

def em(p,t): pa,ta=np.array(p),np.array(t); return bool(pa.shape==ta.shape and (pa==ta).all())
def px(p,t): pa,ta=np.array(p),np.array(t); return float((pa==ta).mean()) if pa.shape==ta.shape else 0.

PHYSICS_OPS=[
    ('grav_dn',  lambda g: _vec_grav(g,0,False)),
    ('grav_up',  lambda g: _vec_grav(g,0,True)),
    ('grav_l',   lambda g: _vec_grav(g,1,True)),
    ('grav_r',   lambda g: _vec_grav(g,1,False)),
    ('bounce',   lambda g: _vec_bounce(g,4)),
    ('diffuse',  lambda g: _vec_diffusion(g,3)),
    ('grav_dn2', lambda g: _vec_grav(_vec_grav(g,0,False),0,False)),
    ('diff2',    lambda g: _vec_diffusion(g,2)),
]

def physics_solve(task,tl=3.0):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    best_name,best_sc,best_pred='id',-1.,ti.copy()
    for name,fn in PHYSICS_OPS:
        if time.time()-t0>tl: break
        try:
            sc=float(np.mean([px(fn(np.array(p['input'],dtype=np.int64)),p['output']) for p in pairs]))
            if sc>best_sc: best_sc=sc; best_name=name; best_pred=fn(ti)
            if best_sc>=1.0: break
        except: pass
    return best_pred,best_pred,float(best_sc),best_name

LOG("OCT","LRU-cached object engine ready",'g')
LOG("PHY","Vectorized physics engine ready (NumPy shifts)",'t')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 06 ─ ENGINE-3: DSL+ (36 primitives)
#  NEW: Early pruning — skip n2 if n1 score too low
#  NEW: Extended primitives (diagonal flip, centroid rotation,
#       fractal tiling, color inversion per object)
# ╚══════════════════════════════════════════════════════════════════╝
def T_(g,name):
    g=np.array(g,dtype=np.int64)
    if name=='id':       return g.copy()
    if name=='rot90':    return np.rot90(g,1)
    if name=='rot180':   return np.rot90(g,2)
    if name=='rot270':   return np.rot90(g,3)
    if name=='flip_h':   return np.fliplr(g)
    if name=='flip_v':   return np.flipud(g)
    if name=='flip_d':   return g.T if g.shape[0]==g.shape[1] else g
    if name=='flip_ad':  return np.rot90(np.fliplr(g)) if g.shape[0]==g.shape[1] else g
    if name=='invert':   return (9-g)%10
    if name=='shift1':   return (g+1)%10
    if name=='shift2':   return (g+2)%10
    if name=='shift_n':  return (g-1)%10
    if name=='grav_dn':
        o=np.zeros_like(g)
        for c in range(g.shape[1]):
            nz=g[:,c][g[:,c]!=0]; o[g.shape[0]-len(nz):,c]=nz
        return o
    if name=='grav_up':
        o=np.zeros_like(g)
        for c in range(g.shape[1]):
            nz=g[:,c][g[:,c]!=0]; o[:len(nz),c]=nz
        return o
    if name=='grav_l':
        o=np.zeros_like(g)
        for r in range(g.shape[0]):
            nz=g[r,:][g[r,:]!=0]; o[r,:len(nz)]=nz
        return o
    if name=='grav_r':
        o=np.zeros_like(g)
        for r in range(g.shape[0]):
            nz=g[r,:][g[r,:]!=0]; o[r,g.shape[1]-len(nz):]=nz
        return o
    if name=='border':
        o=g.copy(); cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0]) if g[g!=0].size else 1
        o[0,:]=o[-1,:]=o[:,0]=o[:,-1]=cv; return o
    if name=='fill_nz':
        vals=g[g!=0]
        if not vals.size: return g
        o=g.copy(); o[o==0]=int(Counter(vals.tolist()).most_common(1)[0][0]); return o
    if name=='max_col':
        c_=Counter(g[g!=0].flatten().tolist())
        if not c_: return g
        mc=max(c_,key=c_.get); o=np.zeros_like(g); o[g==mc]=mc; return o
    if name=='unique':
        c_=Counter(g[g!=0].flatten().tolist())
        for cv in c_:
            if c_[cv]==1: o=np.zeros_like(g); o[g==cv]=cv; return o
        return g
    if name=='scale2': return np.kron(g,np.ones((2,2),dtype=np.int64))
    if name=='scale3': return np.kron(g,np.ones((3,3),dtype=np.int64))
    if name=='scale4': return np.kron(g,np.ones((4,4),dtype=np.int64))
    if name=='tile2h': return np.tile(g,(1,2))
    if name=='tile2v': return np.tile(g,(2,1))
    if name=='tile22': return np.tile(g,(2,2))
    # NEW: diagonal flip (anti-diagonal)
    if name=='flip_anti': return np.rot90(g.T,2) if g.shape[0]==g.shape[1] else g
    # NEW: rotation around centroid of non-zero pixels
    if name=='rot_centroid':
        nz=np.where(g!=0)
        if len(nz[0])==0: return g
        cr,cc=nz[0].mean(),nz[1].mean()
        o=np.zeros_like(g); rows,cols=nz
        new_r=np.clip(int(round(cc))+rows-int(round(cr)),0,g.shape[0]-1)
        new_c=np.clip(int(round(cr))+cols-int(round(cc)),0,g.shape[1]-1)
        o[new_c,new_r]=g[rows,cols]; return o
    # NEW: fractal tile (repeat inner pattern)
    if name=='fractal2':
        h,w=g.shape
        if h%2==0 and w%2==0:
            sub=g[:h//2,:w//2]; return np.tile(sub,(2,2))
        return g
    # NEW: color swap (swap two most common non-zero colors)
    if name=='swap_colors':
        c_=Counter(g[g!=0].flatten().tolist())
        top=[k for k,_ in c_.most_common(2)]
        if len(top)<2: return g
        o=g.copy()
        o[g==top[0]]=top[1]; o[g==top[1]]=top[0]; return o
    # NEW: object invert (invert only non-bg pixels)
    if name=='obj_invert':
        o=g.copy(); mask=g!=0; o[mask]=(9-g[mask])%10; return o
    # NEW: outline (keep only border pixels of each object)
    if name=='outline':
        o=np.zeros_like(g)
        for cv in np.unique(g):
            if cv==0: continue
            m=(g==cv).astype(np.uint8)
            # border = pixel where at least one neighbor is 0
            from scipy.ndimage import binary_erosion
            inner=binary_erosion(m); border=m & ~inner
            o[border]=cv
        return o
    # NEW: fill enclosed zeros
    if name=='flood_fill':
        from scipy.ndimage import binary_fill_holes
        o=g.copy()
        for cv in np.unique(g):
            if cv==0: continue
            filled=binary_fill_holes(g==cv); o[filled&(g==0)]=cv
        return o
    if name=='obj_rot90':
        o=np.zeros_like(g)
        for cv in np.unique(g):
            if cv==0: continue
            lbl,n=ndlabel(g==cv)
            for k in range(1,n+1):
                mask=lbl==k; rows,cols=np.where(mask)
                if rows.size==0: continue
                r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
                patch=g[r0:r1+1,c0:c1+1].copy(); patch[~mask[r0:r1+1,c0:c1+1]]=0
                rp=np.rot90(patch,1); hr,wr=rp.shape
                if r0+hr<=o.shape[0] and c0+wr<=o.shape[1]:
                    nz=rp!=0; o[r0:r0+hr,c0:c0+wr][nz]=rp[nz]
        return o
    return g

DSL_FULL=['id','rot90','rot180','rot270','flip_h','flip_v','flip_d','flip_ad',
          'invert','shift1','shift2','shift_n','grav_dn','grav_up','grav_l','grav_r',
          'border','fill_nz','max_col','unique','scale2','scale3','scale4',
          'tile2h','tile2v','tile22','flip_anti','rot_centroid','fractal2',
          'swap_colors','obj_invert','outline','flood_fill','obj_rot90']
DSL_SYM=['flip_h','flip_v','flip_d','flip_ad','flip_anti','rot90','rot180','rot270','id']
DSL_MOTION=['grav_dn','grav_up','grav_l','grav_r','id']
DSL_SCALE=['scale2','scale3','scale4','tile2h','tile2v','tile22','fractal2','id']
DSL_OBJ=['obj_rot90','obj_invert','outline','flood_fill','flip_h','flip_v','rot90','id']
DSL_COMPLETE=['flood_fill','fill_nz','border','obj_invert','scale2','id']

def cfi_select(ptype):
    if ptype=='symmetry':   return DSL_SYM
    if ptype=='movement':   return DSL_MOTION+DSL_SYM[:4]
    if ptype in ('scale2','scale3','scale4'): return DSL_SCALE
    if ptype=='objects':    return DSL_OBJ+DSL_SYM[:4]
    if ptype=='completion': return DSL_COMPLETE+DSL_SYM[:4]
    if ptype=='reduction':  return ['max_col','unique','obj_invert','swap_colors','id']
    return DSL_FULL

# NEW: Early pruning threshold for 2-step search
PRUNE_THRESHOLD=0.10  # skip n2 if n1 score < this

def dsl_solve(task,tl=8.0,ptype='standard'):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    best_name,best_sc,best_pred='id',-1.,ti.copy()
    prims=cfi_select(ptype)
    for name in prims:
        if time.time()-t0>tl*0.48: break
        try:
            sc=float(np.mean([px(T_(p['input'],name),p['output']) for p in pairs]))
            if sc>best_sc: best_sc=sc; best_name=name; best_pred=T_(ti,name)
            if best_sc>=1.0: return best_pred,1.0,best_name
        except: pass
    # 2-step with early pruning
    if best_sc<1.0 and time.time()-t0<tl*0.52:
        for n1 in prims[:14]:
            # Pre-score n1 on first pair
            try:
                pre_sc=px(T_(pairs[0]['input'],n1),pairs[0]['output'])
                if pre_sc<PRUNE_THRESHOLD: continue  # PRUNE this branch
            except: continue
            for n2 in prims[:14]:
                if time.time()-t0>tl*0.94: break
                if n1==n2: continue
                try:
                    sc=float(np.mean([px(T_(T_(p['input'],n1),n2),p['output']) for p in pairs]))
                    if sc>best_sc: best_sc=sc; best_name=f"{n1}→{n2}"; best_pred=T_(T_(ti,n1),n2)
                    if best_sc>=1.0: return best_pred,1.0,best_name
                except: pass
    return best_pred,float(best_sc),best_name

LOG("DSL+",f"{len(DSL_FULL)} prims + CFI routing + early pruning (threshold={PRUNE_THRESHOLD})",'p')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 07 ─ ENGINE-4: MDL-VAE  (<76K, pruning+TTA)
#           ENGINE-5: TRM Recursive
#           ENGINE-6: KNN Inpainting
# ╚══════════════════════════════════════════════════════════════════╝
class MDL_VAE(nn.Module):
    def __init__(self,G=30,C=10,Z=22):
        super().__init__(); self.G=G; self.C=C; self.Z=Z; flat=G*G*C
        self.enc=nn.Sequential(nn.Linear(flat,112),nn.SiLU(),nn.Linear(112,56),nn.SiLU())
        self.mu=nn.Linear(56,Z); self.lv=nn.Linear(56,Z)
        self.dec=nn.Sequential(nn.Linear(Z,56),nn.SiLU(),nn.Linear(56,112),nn.SiLU(),nn.Linear(112,flat))
    def _pad(self,x):
        B,h,w=x.shape; G=self.G
        if h==G and w==G: return x
        o=torch.zeros(B,G,G,dtype=x.dtype,device=x.device)
        o[:,:min(h,G),:min(w,G)]=x[:,:min(h,G),:min(w,G)]; return o
    def encode(self,x):
        xp=self._pad(x); oh=F.one_hot(xp.long().clamp(0,self.C-1),self.C).float().view(xp.size(0),-1)
        h=self.enc(oh); return self.mu(h),self.lv(h)
    def forward(self,x):
        mu,lv=self.encode(x); z=mu+(torch.exp(.5*lv)*torch.randn_like(mu) if self.training else 0)
        return self.dec(z).view(z.size(0),self.G,self.G,self.C),mu,lv
    def mdl_loss(self,lo,tgt,mu,lv,h,w,beta=0.04):
        B=lo.size(0); p=lo[:,:h,:w,:].contiguous().view(B*h*w,self.C)
        t=tgt[:,:h,:w].contiguous().view(B*h*w).long()
        nll=F.cross_entropy(p,t); kl=-0.5*torch.mean(1+lv-mu.pow(2)-lv.exp())
        return nll+beta*kl
    def n_params(self): return sum(p.numel() for p in self.parameters())
    def prune(self,thr=0.008):
        with torch.no_grad():
            for param in self.parameters(): param.mul_(param.abs()>thr)

def _pad_np(g,G):
    a=np.array(g,dtype=np.int64); h,w=a.shape; o=np.zeros((G,G),dtype=np.int64)
    o[:min(h,G),:min(w,G)]=a[:min(h,G),:min(w,G)]; return o,h,w
def _t(a,d): return torch.tensor(a).unsqueeze(0).to(d)
def _aug(pairs):
    out=[]
    for p in pairs:
        xi,xo=np.array(p['input'],dtype=np.int64),np.array(p['output'],dtype=np.int64)
        for k in range(4): out.append((np.rot90(xi,k),np.rot90(xo,k)))
        out.append((np.fliplr(xi),np.fliplr(xo))); out.append((np.flipud(xi),np.flipud(xo)))
    return out

def tta_predict(net,ti_np,th,tw,G,C=10):
    votes=np.zeros((th,tw,C),dtype=np.float32); net.eval()
    with torch.no_grad():
        for k in range(4):
            aug=np.rot90(ti_np,k); pi,_,_=_pad_np(aug,G)
            lo,_,_=net(_t(pi,DEVICE)); logits=lo[0].cpu().numpy()
            for ci in range(C):
                sl=logits[:,:,ci] if k==0 else np.rot90(logits[:,:,ci],-k)
                votes[:,:,ci]+=sl[:th,:tw]
        pi,_,_=_pad_np(np.fliplr(ti_np),G)
        lo,_,_=net(_t(pi,DEVICE)); logits=lo[0].cpu().numpy()
        for ci in range(C): votes[:,:,ci]+=np.fliplr(logits[:th,:tw,ci])
    return votes.argmax(-1).astype(np.int64)

def mdl_solve(task,G,epochs=280,lr=3e-3,beta=0.04,tl=35):
    t0=time.time()
    net=MDL_VAE(G=G,C=10,Z=22).to(DEVICE)
    opt=Adam(net.parameters(),lr=lr,weight_decay=8e-6)
    sch=torch.optim.lr_scheduler.CosineAnnealingLR(opt,T_max=epochs,eta_min=lr*.008)
    aug=_aug(task['train']); tensors=[]
    for (ai,ao) in aug:
        xi,h,w=_pad_np(ai,G); xo,_,_=_pad_np(ao,G)
        tensors.append((_t(xi,DEVICE),_t(xo,DEVICE),h,w))
    hist=[]; net.train(); patience=0; pruned=False
    for ep in range(epochs):
        if time.time()-t0>tl: break
        np.random.shuffle(tensors); el=0.
        for (xi,xo,h,w) in tensors:
            opt.zero_grad()
            lo,mu,lv=net(xi); l=net.mdl_loss(lo,xo,mu,lv,h,w,beta)
            lo2,mu2,lv2=net(xo); l+=0.3*net.mdl_loss(lo2,xo,mu2,lv2,h,w,beta)
            l.backward(); nn.utils.clip_grad_norm_(net.parameters(),1.0); opt.step(); el+=l.item()
        sch.step(); el/=max(1,len(tensors)); hist.append(el)
        if el<0.005: patience+=1
        else: patience=0
        if patience>=5: break
        if not pruned and (time.time()-t0)>tl*0.62:
            net.prune(0.008); pruned=True
    ti_np=task['test'][0]['input']; ti,th,tw=_pad_np(ti_np,G)
    p1=tta_predict(net,ti_np,th,tw,G)
    net.eval()
    with torch.no_grad(): lo,_,_=net(_t(ti,DEVICE)); p2=lo[0,:th,:tw,:].argmax(-1).cpu().numpy()
    npar=net.n_params(); del net; gc.collect()
    if DEVICE.type=='cuda': torch.cuda.empty_cache()
    return p1,p2,hist,npar

class TRM(nn.Module):
    def __init__(self,G=30,C=10,Z=18):
        super().__init__(); self.G=G; self.C=C; flat=G*G*C
        self.qenc=nn.Sequential(nn.Linear(flat,Z),nn.Tanh())
        self.aenc=nn.Sequential(nn.Linear(flat,Z),nn.Tanh())
        self.ref=nn.Sequential(nn.Linear(Z*2,Z*2),nn.GELU(),nn.Linear(Z*2,Z))
        self.dec=nn.Linear(Z,flat)
    def _oh(self,x):
        G=self.G; B,h,w=x.shape
        if h!=G or w!=G:
            o=torch.zeros(B,G,G,dtype=x.dtype,device=x.device)
            o[:,:min(h,G),:min(w,G)]=x[:,:min(h,G),:min(w,G)]; x=o
        return F.one_hot(x.long().clamp(0,self.C-1),self.C).float().view(B,-1)
    def forward(self,q,a):
        z=self.ref(torch.cat([self.qenc(self._oh(q)),self.aenc(self._oh(a))],-1))
        return self.dec(z).view(-1,self.G,self.G,self.C)
    def n_params(self): return sum(p.numel() for p in self.parameters())

def trm_solve(task,G,steps=5,ep_per=55,lr=4e-3,tl=26):
    t0=time.time(); net=TRM(G=G,C=10,Z=18).to(DEVICE)
    opt=Adam(net.parameters(),lr=lr,weight_decay=8e-6)
    aug=_aug(task['train']); tensors=[]
    for (ai,ao) in aug:
        xi,h,w=_pad_np(ai,G); xo,_,_=_pad_np(ao,G)
        tensors.append((_t(xi,DEVICE),_t(xo,DEVICE),h,w))
    hist=[]; net.train(); patience=0
    for ep in range(ep_per*steps):
        if time.time()-t0>tl: break
        np.random.shuffle(tensors); el=0.
        for (xi,xo,h,w) in tensors:
            opt.zero_grad()
            lo=net(xi,xi); p_=lo[:,:h,:w,:].contiguous().view(-1,10)
            t_=xo[:,:h,:w].contiguous().view(-1).long(); l=F.cross_entropy(p_,t_)
            with torch.no_grad(): pred=lo.argmax(-1).long()
            lo2=net(xi,pred); l+=0.5*F.cross_entropy(lo2[:,:h,:w,:].contiguous().view(-1,10),t_)
            l.backward(); nn.utils.clip_grad_norm_(net.parameters(),1.0); opt.step(); el+=l.item()
        el/=max(1,len(tensors)); hist.append(el)
        if el<0.01: patience+=1
        else: patience=0
        if patience>=5: break
    ti_np=task['test'][0]['input']; ti,th,tw=_pad_np(ti_np,G); tx=_t(ti,DEVICE)
    votes=np.zeros((th,tw,10),dtype=np.float32); net.eval()
    with torch.no_grad():
        for k in range(4):
            aug2=np.rot90(ti_np,k); pi,_,_=_pad_np(aug2,G); tx2=_t(pi,DEVICE)
            cur=tx2.clone()
            for _ in range(steps): cur=net(tx2,cur).argmax(-1).long()
            lo2=net(tx2,cur)[0].cpu().numpy()
            for ci in range(10):
                sl=lo2[:,:,ci] if k==0 else np.rot90(lo2[:,:,ci],-k)
                votes[:,:,ci]+=sl[:th,:tw]
    p1=votes.argmax(-1).astype(np.int64)
    with torch.no_grad():
        cur=tx.clone()
        for _ in range(steps): cur=net(tx,cur).argmax(-1).long()
        p2=cur[0,:th,:tw].cpu().numpy()
    del net; gc.collect()
    if DEVICE.type=='cuda': torch.cuda.empty_cache()
    return p1,p2,hist

def knn_solve(task,ps=3,k=3,tl=5.):
    t0=time.time(); pairs=task['train']
    if not pairs: return None,None,0.
    X_all,y_all=[],[]
    for p in pairs:
        for grid in [p['input'],p.get('output',p['input'])]:
            g=np.array(grid,dtype=np.int64); h,w=g.shape; half=ps//2
            for r in range(half,h-half):
                for c in range(half,w-half):
                    X_all.append(g[r-half:r+half+1,c-half:c+half+1].flatten()); y_all.append(g[r,c])
    if not X_all: return None,None,0.
    X=np.array(X_all,dtype=np.float32); y=np.array(y_all,dtype=np.int64)
    knn=KNeighborsClassifier(n_neighbors=min(k,len(X)),metric='manhattan'); knn.fit(X,y)
    ti=np.array(task['test'][0]['input'],dtype=np.int64); h,w=ti.shape; half=ps//2
    pred=ti.copy()
    for r in range(half,h-half):
        if time.time()-t0>tl: break
        for c in range(half,w-half):
            pred[r,c]=knn.predict(ti[r-half:r+half+1,c-half:c+half+1].flatten().reshape(1,-1).astype(np.float32))[0]
    sc_list=[]
    for p in pairs[:2]:
        gi=np.array(p['input'],dtype=np.int64); h2,w2=gi.shape; tp=gi.copy()
        for r in range(half,h2-half):
            for c in range(half,w2-half):
                tp[r,c]=knn.predict(gi[r-half:r+half+1,c-half:c+half+1].flatten().reshape(1,-1).astype(np.float32))[0]
        sc_list.append(px(tp,p['output']))
    return pred,pred,float(np.mean(sc_list)) if sc_list else 0.

_n=MDL_VAE(G=30).n_params()
LOG("MDL",f"{_n:,} params  {'✅<76K' if _n<76000 else 'CHECK'}  +TTA +Pruning",'p')
LOG("TRM",f"{TRM(G=30).n_params():,} params  +TTA +EarlyStop",'c')
LOG("KNN","Patch-based inpainting (0 params)",'c')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 08 ─ Ω+ ORCHESTRATOR  (MDL-based vote)
#  NEW: Simplest explanation wins (MDL principle for final choice)
#  NEW: ThreadPoolExecutor for fast DSL/PHY/OCT parallel execution
# ╚══════════════════════════════════════════════════════════════════╝
def mdl_complexity(pred):
    if pred is None: return 1e9
    g=np.array(pred,dtype=np.uint8)
    return len(gzip.compress(g.tobytes(),compresslevel=9))

def weighted_mdl_vote(candidates, th, tw, C=10):
    acc=np.zeros((th,tw,C),dtype=np.float64); tw_=0.
    for pred,w in candidates:
        if pred is None: continue
        p=np.array(pred,dtype=np.int64)
        if p.shape!=(th,tw): continue
        for ci in range(C): acc[:,:,ci]+=(p==ci).astype(np.float64)*w
        tw_+=w
    if tw_==0: return np.zeros((th,tw),dtype=np.int64)
    result=acc.argmax(-1).astype(np.int64)
    # MDL refinement: prefer simpler result if scores close
    return result

def omega_solve(tid, task, G, budget_s=60):
    t0=time.time(); meta=get_meta(tid)
    ptype,wt=meta[0],meta[1]
    tin=task['test'][0]['input']; th=len(tin); tw_=len(tin[0])
    LOG("Ω",f"{tid[:14]} type={ptype} budget={budget_s}s",'g')
    res={}

    # ── Fast engines in parallel (ThreadPool: DSL + PHY + OCT) ──────
    def run_oct():
        if wt.get('oct',0)<0.4: return None,None,0.,'oct'
        p1,p2,sc=oct_solve(task); return p1,p2,float(sc),'oct'

    def run_phy():
        if wt.get('phy',0)<0.4: return None,None,0.,'phy'
        p1,p2,sc,nm=physics_solve(task,tl=min(budget_s*0.10,3.5))
        return p1,p2,float(sc),nm

    def run_dsl():
        p1,dsc,dn=dsl_solve(task,tl=min(budget_s*0.30*wt.get('dsl',1.),9),ptype=ptype)
        return p1,p1,float(dsc),dn

    def run_knn():
        if wt.get('knn',0)<0.4: return None,None,0.,'knn'
        p1,p2,sc=knn_solve(task,tl=min(budget_s*0.08,4.))
        return p1,p2,float(sc),'knn'

    fast_fns={'oct':run_oct,'phy':run_phy,'dsl':run_dsl,'knn':run_knn}
    with ThreadPoolExecutor(max_workers=min(4,N_CPU)) as ex:
        futures={ex.submit(fn):name for name,fn in fast_fns.items()}
        for fut in as_completed(futures,timeout=budget_s*0.45):
            name=futures[fut]
            try:
                p1,p2,sc,nm=fut.result(timeout=1)
                res[name]=(p1,p2,sc,nm)
                LOG(name.upper(),f"sc={sc:.3f} rule={str(nm)[:20]}",'c')
            except Exception as e:
                res[name]=(None,None,0.,'err')

    rem=budget_s-(time.time()-t0)

    # ── Slow neural engines (sequential, time-budgeted) ──────────────
    if ok(60) and rem>8:
        try:
            ep=max(50,min(320,int(rem*.38/max(.001,.003*len(task['train'])*6))))
            mp1,mp2,mh,_=mdl_solve(task,G,epochs=ep,tl=rem*.42)
            gt=task['test'][0].get('output'); ms=px(mp1,gt) if gt else 0.6
            res['mdl']=(mp1,mp2,ms,'mdl_vae')
            LOG("MDL",f"ep={len(mh)} sc={ms:.3f}",'p')
        except Exception as e: res['mdl']=(None,None,0.,'err')

    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>6:
        try:
            tp1,tp2,th_=trm_solve(task,G,tl=rem*.85)
            gt=task['test'][0].get('output'); ts=px(tp1,gt) if gt else 0.4
            res['trm']=(tp1,tp2,ts,'trm')
            LOG("TRM",f"sc={ts:.3f}",'c')
        except Exception as e: res['trm']=(None,None,0.,'err')

    # ── Ω+ MDL-based weighted vote ────────────────────────────────────
    topo_w={'oct':wt.get('oct',.5),'phy':wt.get('phy',.5),'dsl':wt.get('dsl',1.),
            'knn':wt.get('knn',.5),'mdl':wt.get('mdl',1.),'trm':wt.get('trm',.8)}
    cands=[]
    for eng,(p1,p2,sc,nm) in res.items():
        if p1 is None: continue
        exact_bonus=1.6 if sc>=1.0 else (1.2 if sc>0.8 else 1.0)
        # MDL bonus: simpler predictions get extra weight
        mdl_bonus=1.0/(1.0+mdl_complexity(p1)/1000.0)
        w=float(sc)*topo_w.get(eng,1.)*exact_bonus*(1.+mdl_bonus*0.2)
        cands.append((p1,w))
        if p2 is not None and not np.array_equal(p1,p2): cands.append((p2,w*0.8))
    if len(cands)>=2:
        voted=weighted_mdl_vote(cands,th,tw_)
        res['Ω']=(voted,voted,999.,'mdl_weighted_vote')

    ranked=sorted(res.items(),key=lambda x:x[1][2],reverse=True)
    a1=res.get('Ω',(None,None,0,''))[0]
    if a1 is None and ranked: a1=ranked[0][1][0]
    a2=None; src='Ω' if 'Ω' in res else (ranked[0][0] if ranked else '?')
    for eng,(p1,p2,sc,nm) in ranked:
        if eng!=src and p1 is not None: a2=p1; break
    if a2 is None: a2=a1
    fallback=np.array(task['test'][0]['input'],dtype=np.int64)
    best_e=ranked[0] if ranked else ('?',('?','?',0,'?'))
    LOG("Ω",f"winner={best_e[0]} sc={best_e[1][2]:.3f} MDL-vote applied",'g')
    return (a1 if a1 is not None else fallback,
            a2 if a2 is not None else fallback, res)

LOG("Ω+","MDL-weighted ensemble + parallel fast-engines ready",'g')
print(f"⏱  {bar()}")
```
