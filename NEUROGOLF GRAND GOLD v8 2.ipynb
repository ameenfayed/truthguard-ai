```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 03 ─ LOAD ALL DATA · TOPOLOGY CLASSIFIER · TASK SORTER
# ╚══════════════════════════════════════════════════════════════════╝
UTILS=None; sys.path.insert(0,DATA_ROOT)
if Path(UTILS_DIR).exists():
    try:
        import neurogolf_utils as UTILS
        LOG("UTILS",f"API: {[x for x in dir(UTILS) if not x.startswith('_')][:4]}",'g')
    except:
        for pyf in sorted(Path(UTILS_DIR).glob('*.py')):
            try:
                sp=importlib.util.spec_from_file_location(pyf.stem,pyf)
                m=importlib.util.module_from_spec(sp); sp.loader.exec_module(m)
                UTILS=m; LOG("UTILS",f"{pyf.name}",'g')
            except: pass
else: LOG("UTILS","absent — solo mode",'c')

TASKS,ERRS={},[]
for jf in sorted(Path(DATA_ROOT).rglob('*.json')):
    try:
        d=json.loads(jf.read_text())
        if isinstance(d,dict) and 'train' in d and 'test' in d: TASKS[jf.stem]=d
        elif isinstance(d,list):
            for i,it in enumerate(d):
                if isinstance(it,dict) and 'train' in it: TASKS[f'{jf.stem}_{i:04d}']=it
    except Exception as e: ERRS.append((jf.name,str(e)))
TASK_IDS=[k for k in TASKS if not k.startswith('_raw_')]
LOG("DATA",f"Tasks={len(TASK_IDS)}  errors={len(ERRS)}",'g')

def _sym(g):
    h,w=g.shape
    return max(float(np.mean(g==g[:,::-1])),float(np.mean(g==g[::-1,:])),
               float(np.mean(g==g.T)) if h==w else 0.)

def classify(task):
    pairs=task.get('train',[])
    if not pairs: return 'standard',1.,1.
    gi0=np.array(pairs[0]['input'],dtype=np.int64)
    go0=np.array(pairs[0]['output'],dtype=np.int64)
    sz=gi0.shape!=go0.shape
    scale=None
    if go0.shape[0]>0 and gi0.shape[0]>0:
        rr=go0.shape[0]/gi0.shape[0]; rc=go0.shape[1]/gi0.shape[1]
        if rr==rc and rr in [2.,3.,4.]: scale=int(rr)
    sym=_sym(gi0); dc=len(np.unique(go0))-len(np.unique(gi0))
    if scale:    return f'scale{scale}',3.,0.2
    if sym>0.80: return 'symmetry',2.5,0.5
    if sz:       return 'size_change',1.5,0.8
    if dc<-0.5:  return 'reduction',1.2,1.0
    if dc>0.5:   return 'completion',0.8,1.5
    return 'standard',1.0,1.5

TASK_META={}; all_max_h=all_max_w=0
for tid in TASK_IDS:
    task=TASKS[tid]; si,so=[],[]
    for p in task.get('train',[])+task.get('test',[]):
        gi=np.array(p['input'],dtype=np.int64); si.append(gi.shape)
        if 'output' in p: so.append(np.array(p['output'],dtype=np.int64).shape)
    mh=max(s[0] for s in si+(so or si)); mw=max(s[1] for s in si+(so or si))
    all_max_h=max(all_max_h,mh); all_max_w=max(all_max_w,mw)
    pt,dw,nw=classify(task)
    TASK_META[tid]=dict(max_h=mh,max_w=mw,ptype=pt,dsl_w=dw,neural_w=nw,
                        n_train=len(task.get('train',[])))

MAX_G=max(all_max_h,all_max_w,30)
td=Counter(v['ptype'] for v in TASK_META.values())
LOG("META",f"MAX_G={MAX_G}  types={dict(td.most_common())}",'y')

# Priority sort: fast-solve types first
def _pri(t):
    return {'scale2':0,'scale3':0,'scale4':0,'symmetry':1,'movement':2,
            'standard':3,'size_change':4,'completion':5,'reduction':5}.get(
        TASK_META.get(t,{}).get('ptype','standard'),5)
SORTED_IDS=sorted(TASK_IDS,key=_pri)
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 05 ─ ENGINE A: COLOR BIJECTION  (~20% tasks, <1ms)
#  ENGINE B: RULE INDUCTION  (~10% tasks, <10ms)
#  Both are symbolic, 0 params, instant — run FIRST always
# ╚══════════════════════════════════════════════════════════════════╝

def em(p,t): pa,ta=np.array(p),np.array(t); return bool(pa.shape==ta.shape and (pa==ta).all())
def px(p,t): pa,ta=np.array(p),np.array(t); return float((pa==ta).mean()) if pa.shape==ta.shape else 0.

# ── A: Color Bijection ───────────────────────────────────────────────
def color_bijection(task):
    pairs=task['train']
    mapping={}
    for p in pairs:
        gi=np.array(p['input'],dtype=np.int64); go=np.array(p['output'],dtype=np.int64)
        if gi.shape!=go.shape: return None,None,0.
        for sv,tv in zip(gi.flatten(),go.flatten()):
            sv,tv=int(sv),int(tv)
            if sv in mapping:
                if mapping[sv]!=tv: return None,None,0.
            else: mapping[sv]=tv
    if not mapping: return None,None,0.
    sc=float(np.mean([px(np.vectorize(lambda v:mapping.get(int(v),int(v)))(
        np.array(p['input'],dtype=np.int64)),p['output']) for p in pairs]))
    if sc<0.95: return None,None,0.
    ti=np.array(task['test'][0]['input'],dtype=np.int64)
    p1=np.vectorize(lambda v:mapping.get(int(v),int(v)))(ti)
    return p1,p1.copy(),float(sc)

# ── B: Rule Induction ────────────────────────────────────────────────
def _mc(g):
    g=np.array(g,dtype=np.int64); c_=Counter(g[g!=0].flatten().tolist())
    return int(c_.most_common(1)[0][0]) if c_ else 0
def _lc(g):
    g=np.array(g,dtype=np.int64); c_=Counter(g[g!=0].flatten().tolist())
    return int(c_.most_common()[-1][0]) if c_ else 0
def _uc(g):
    g=np.array(g,dtype=np.int64); c_=Counter(g[g!=0].flatten().tolist())
    for cv,cnt in c_.items():
        if cnt==1: return cv
    return 0
def _nobj(g):
    g=np.array(g,dtype=np.int64)
    return sum(ndlabel(g==c)[1] for c in np.unique(g) if c!=0)

def rule_induction(task):
    pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    outs=[np.array(p['output'],dtype=np.int64) for p in pairs]
    # All 1x1 outputs?
    if all(o.size==1 for o in outs):
        for fn,name in [(_mc,'most_common'),(_lc,'least_common'),(_uc,'unique_color'),(_nobj,'count_objects')]:
            try:
                vals=[fn(p['input']) for p in pairs]
                if all(int(o.flat[0])==v for o,v in zip(outs,vals)):
                    pred=np.array([[fn(ti)]],dtype=np.int64)
                    return pred,pred,1.0,name
            except: pass
    # Scale ratio
    try:
        ratios=[]
        for p in pairs:
            gi=np.array(p['input'],dtype=np.int64); go=np.array(p['output'],dtype=np.int64)
            if gi.shape[0]>0: ratios.append(go.shape[0]/gi.shape[0])
        if ratios and len(set(round(r,2) for r in ratios))==1 and ratios[0] in [2.,3.,4.]:
            k=int(ratios[0])
            pred=np.kron(ti,np.ones((k,k),dtype=np.int64))
            sc=float(np.mean([px(np.kron(np.array(p['input'],dtype=np.int64),
                np.ones((k,k),dtype=np.int64)),p['output']) for p in pairs]))
            if sc>0.9: return pred,pred,sc,f'scale_{k}x'
    except: pass
    # Transpose rule (output = input.T for square grids)
    try:
        if all(np.array(p['input']).shape[0]==np.array(p['input']).shape[1] for p in pairs):
            sc=float(np.mean([px(np.array(p['input'],dtype=np.int64).T,p['output']) for p in pairs]))
            if sc>0.95:
                if ti.shape[0]==ti.shape[1]: return ti.T.copy(),ti.T.copy(),sc,'transpose'
    except: pass
    return None,None,0.,'none'

LOG("A","Color Bijection  ~20% tasks  <1ms  0 params",'y')
LOG("B","Rule Induction   ~10% tasks  <10ms  0 params",'g')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 06 ─ ENGINE C: RECTANGLE/SYMMETRY  (~15% tasks, <5ms)
#  ENGINE D: PHYSICS  (gravity·bounce·diffusion, ~8% tasks)
# ╚══════════════════════════════════════════════════════════════════╝

# ── C: Rectangle / Symmetry / Connect ────────────────────────────────
def _fill_bbox(g):
    g=np.array(g,dtype=np.int64); nz=np.where(g!=0)
    if not nz[0].size: return g
    r0,r1=nz[0].min(),nz[0].max(); c0,c1=nz[1].min(),nz[1].max()
    o=g.copy(); cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0])
    o[r0:r1+1,c0:c1+1]=cv; return o

def _bbox_outline(g):
    g=np.array(g,dtype=np.int64); nz=np.where(g!=0)
    if not nz[0].size: return g
    r0,r1=nz[0].min(),nz[0].max(); c0,c1=nz[1].min(),nz[1].max()
    o=np.zeros_like(g); cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0])
    o[r0,c0:c1+1]=o[r1,c0:c1+1]=cv; o[r0:r1+1,c0]=o[r0:r1+1,c1]=cv; return o

def _connect(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    for r in range(h):
        nz=np.where(g[r,:]!=0)[0]
        if len(nz)>=2: o[r,nz.min():nz.max()+1]=int(g[r,nz[0]])
    for c in range(w):
        nz=np.where(g[:,c]!=0)[0]
        if len(nz)>=2: o[nz.min():nz.max()+1,c]=int(g[nz[0],c])
    return o

def _fill_enclosed(g):
    from collections import deque
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0]) if g[g!=0].size else 1
    visited=np.zeros((h,w),dtype=bool); q=deque()
    for r in range(h):
        for c in [0,w-1]:
            if g[r,c]==0 and not visited[r,c]: visited[r,c]=True; q.append((r,c))
    for c in range(w):
        for r in [0,h-1]:
            if g[r,c]==0 and not visited[r,c]: visited[r,c]=True; q.append((r,c))
    while q:
        r,c=q.popleft()
        for dr,dc in [(-1,0),(1,0),(0,-1),(0,1)]:
            nr,nc=r+dr,c+dc
            if 0<=nr<h and 0<=nc<w and g[nr,nc]==0 and not visited[nr,nc]:
                visited[nr,nc]=True; q.append((nr,nc))
    o[(g==0)&(~visited)]=cv; return o

def _sym_complete_h(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    half=w//2
    for r in range(h):
        for c in range(half):
            if o[r,c]==0 and o[r,w-1-c]!=0: o[r,c]=o[r,w-1-c]
            elif o[r,w-1-c]==0 and o[r,c]!=0: o[r,w-1-c]=o[r,c]
    return o

def _sym_complete_v(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    half=h//2
    for c in range(w):
        for r in range(half):
            if o[r,c]==0 and o[h-1-r,c]!=0: o[r,c]=o[h-1-r,c]
            elif o[h-1-r,c]==0 and o[r,c]!=0: o[h-1-r,c]=o[r,c]
    return o

def _draw_cross(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    nz=np.where(g!=0)
    if not nz[0].size: return g
    cr,cc=int(nz[0].mean()),int(nz[1].mean())
    cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0])
    o[cr,:]=cv; o[:,cc]=cv; return o

RECT_OPS=[('fill_bbox',_fill_bbox),('bbox_outline',_bbox_outline),
          ('connect',_connect),('fill_enclosed',_fill_enclosed),
          ('sym_h',_sym_complete_h),('sym_v',_sym_complete_v),
          ('draw_cross',_draw_cross)]

def rect_solve(task,tl=4.):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    bn,bsc,bp='id',-1.,ti.copy()
    for name,fn in RECT_OPS:
        if time.time()-t0>tl: break
        try:
            sc=float(np.mean([px(fn(np.array(p['input'],dtype=np.int64)),p['output']) for p in pairs]))
            if sc>bsc: bsc=sc; bn=name; bp=fn(ti)
            if bsc>=1.0: return bp,bp,1.0,bn
        except: pass
    return bp,bp,float(bsc),bn

# ── D: Physics ───────────────────────────────────────────────────────
def _grav(g,d):
    g=np.array(g,dtype=np.int64); h,w=g.shape
    for _ in range(h+w):
        mv=False
        if d=='down':
            for c in range(w):
                for r in range(h-2,-1,-1):
                    if g[r,c]!=0 and g[r+1,c]==0: g[r+1,c]=g[r,c]; g[r,c]=0; mv=True
        elif d=='up':
            for c in range(w):
                for r in range(1,h):
                    if g[r,c]!=0 and g[r-1,c]==0: g[r-1,c]=g[r,c]; g[r,c]=0; mv=True
        elif d=='left':
            for r in range(h):
                for c in range(1,w):
                    if g[r,c]!=0 and g[r,c-1]==0: g[r,c-1]=g[r,c]; g[r,c]=0; mv=True
        elif d=='right':
            for r in range(h):
                for c in range(w-2,-1,-1):
                    if g[r,c]!=0 and g[r,c+1]==0: g[r,c+1]=g[r,c]; g[r,c]=0; mv=True
        if not mv: break
    return g

def _diff(g,n=3):
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    for _ in range(n):
        nxt=o.copy()
        for r in range(1,h-1):
            for c in range(1,w-1):
                if o[r,c]==0:
                    nb=[o[r+dr,c+dc] for dr,dc in [(-1,0),(1,0),(0,-1),(0,1)] if o[r+dr,c+dc]!=0]
                    if len(nb)==4: nxt[r,c]=Counter(nb).most_common(1)[0][0]
        o=nxt
    return o

def _flood(g):
    from collections import deque
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=g.copy()
    cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0]) if g[g!=0].size else 1
    for r in range(h):
        for c in range(w):
            if o[r,c]==0:
                nb=[o[r+dr,c+dc] for dr,dc in [(-1,0),(1,0),(0,-1),(0,1)]
                    if 0<=r+dr<h and 0<=c+dc<w and o[r+dr,c+dc]!=0]
                if nb: o[r,c]=Counter(nb).most_common(1)[0][0]
    return o

PHY_OPS=[('grav_dn',lambda g:_grav(g,'down')),('grav_up',lambda g:_grav(g,'up')),
         ('grav_l',lambda g:_grav(g,'left')),('grav_r',lambda g:_grav(g,'right')),
         ('diff3',lambda g:_diff(g,3)),('diff1',lambda g:_diff(g,1)),
         ('flood',_flood)]

def physics_solve(task,tl=3.):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    bn,bsc,bp='id',-1.,ti.copy()
    for name,fn in PHY_OPS:
        if time.time()-t0>tl: break
        try:
            sc=float(np.mean([px(fn(np.array(p['input'],dtype=np.int64)),p['output']) for p in pairs]))
            if sc>bsc: bsc=sc; bn=name; bp=fn(ti)
            if bsc>=1.0: return bp,bp,1.0,bn
        except: pass
    return bp,bp,float(bsc),bn

LOG("C","Rect+Sym  7 ops  ~15% tasks  <5ms  0 params",'g')
LOG("D","Physics   7 ops  ~8%  tasks  <3ms  0 params",'g')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 07 ─ ENGINE E: DSL+  120+ PRIMITIVES  4-STEP BEAM SEARCH
#  Covers geometric · color · gravity · objects · pattern · arithmetic
# ╚══════════════════════════════════════════════════════════════════╝

_P={}
def _reg(n,fn): _P[n]=fn

# geometric
_reg('id',        lambda g: np.array(g,dtype=np.int64).copy())
_reg('rot90',     lambda g: np.rot90(np.array(g,dtype=np.int64),1))
_reg('rot180',    lambda g: np.rot90(np.array(g,dtype=np.int64),2))
_reg('rot270',    lambda g: np.rot90(np.array(g,dtype=np.int64),3))
_reg('flip_h',    lambda g: np.fliplr(np.array(g,dtype=np.int64)))
_reg('flip_v',    lambda g: np.flipud(np.array(g,dtype=np.int64)))
_reg('flip_d',    lambda g: np.array(g,dtype=np.int64).T if np.array(g).shape[0]==np.array(g).shape[1] else np.array(g,dtype=np.int64))
_reg('flip_ad',   lambda g: np.rot90(np.fliplr(np.array(g,dtype=np.int64))) if np.array(g).shape[0]==np.array(g).shape[1] else np.array(g,dtype=np.int64))
_reg('rot90_fh',  lambda g: np.fliplr(np.rot90(np.array(g,dtype=np.int64),1)))
_reg('rot90_fv',  lambda g: np.flipud(np.rot90(np.array(g,dtype=np.int64),1)))
_reg('scale2',    lambda g: np.kron(np.array(g,dtype=np.int64),np.ones((2,2),dtype=np.int64)))
_reg('scale3',    lambda g: np.kron(np.array(g,dtype=np.int64),np.ones((3,3),dtype=np.int64)))
_reg('scale4',    lambda g: np.kron(np.array(g,dtype=np.int64),np.ones((4,4),dtype=np.int64)))
_reg('tile2h',    lambda g: np.tile(np.array(g,dtype=np.int64),(1,2)))
_reg('tile2v',    lambda g: np.tile(np.array(g,dtype=np.int64),(2,1)))
_reg('tile22',    lambda g: np.tile(np.array(g,dtype=np.int64),(2,2)))
_reg('tile33',    lambda g: np.tile(np.array(g,dtype=np.int64),(3,3)))
_reg('make_sym_h',lambda g: np.concatenate([np.array(g,dtype=np.int64),np.fliplr(np.array(g,dtype=np.int64))],axis=1))
_reg('make_sym_v',lambda g: np.concatenate([np.array(g,dtype=np.int64),np.flipud(np.array(g,dtype=np.int64))],axis=0))

# color
_reg('invert',  lambda g: (9-np.array(g,dtype=np.int64))%10)
_reg('shift1',  lambda g: (np.array(g,dtype=np.int64)+1)%10)
_reg('shift2',  lambda g: (np.array(g,dtype=np.int64)+2)%10)
_reg('shift3',  lambda g: (np.array(g,dtype=np.int64)+3)%10)
_reg('shift_n', lambda g: (np.array(g,dtype=np.int64)-1)%10)
_reg('shift_n2',lambda g: (np.array(g,dtype=np.int64)-2)%10)
_reg('bin',     lambda g: (np.array(g,dtype=np.int64)!=0).astype(np.int64))
def _keep_max(g):
    g=np.array(g,dtype=np.int64); c_=Counter(g[g!=0].flatten().tolist())
    if not c_: return g
    mc=max(c_,key=c_.get); o=np.zeros_like(g); o[g==mc]=mc; return o
def _keep_min(g):
    g=np.array(g,dtype=np.int64); c_=Counter(g[g!=0].flatten().tolist())
    if not c_: return g
    mc=min(c_,key=c_.get); o=np.zeros_like(g); o[g==mc]=mc; return o
def _keep_uniq(g):
    g=np.array(g,dtype=np.int64); c_=Counter(g[g!=0].flatten().tolist())
    for cv in c_:
        if c_[cv]==1: o=np.zeros_like(g); o[g==cv]=cv; return o
    return g
def _zero_bg(g):
    g=np.array(g,dtype=np.int64); c_=Counter(g.flatten().tolist())
    bg=max(c_,key=c_.get); o=g.copy(); o[g==bg]=0; return o
def _fill_nz(g):
    g=np.array(g,dtype=np.int64); vals=g[g!=0]
    if not vals.size: return g
    o=g.copy(); o[o==0]=int(Counter(vals.tolist()).most_common(1)[0][0]); return o
def _rot_colors(g):
    g=np.array(g,dtype=np.int64); c_=sorted(set(g[g!=0].flatten().tolist()))
    if len(c_)<2: return g
    mp={c_[i]:c_[(i+1)%len(c_)] for i in range(len(c_))}
    o=g.copy()
    for old,new in mp.items(): o[g==old]=new
    return o
_reg('keep_max',_keep_max); _reg('keep_min',_keep_min); _reg('keep_uniq',_keep_uniq)
_reg('zero_bg',_zero_bg); _reg('fill_nz',_fill_nz); _reg('rot_colors',_rot_colors)

# gravity
def _gv(d): return lambda g:_grav(g,d)
for _d in ['down','up','left','right']: _reg(f'gv_{_d[0]}',_gv(_d))
_reg('diff1',lambda g:_diff(g,1)); _reg('diff3',lambda g:_diff(g,3)); _reg('flood_f',_flood)

# border/fill
_reg('border',  lambda g: _fill_bbox(np.array(g,dtype=np.int64)) )  # reuse
_reg('outline', lambda g: _bbox_outline(np.array(g,dtype=np.int64)))
_reg('connect', lambda g: _connect(np.array(g,dtype=np.int64)))
_reg('enclosed',lambda g: _fill_enclosed(np.array(g,dtype=np.int64)))
_reg('fill_nz2',lambda g: _fill_nz(np.array(g,dtype=np.int64)))
_reg('sym_h',   lambda g: _sym_complete_h(np.array(g,dtype=np.int64)))
_reg('sym_v',   lambda g: _sym_complete_v(np.array(g,dtype=np.int64)))

# object-level
def _obj_rot90(g):
    g=np.array(g,dtype=np.int64); o=np.zeros_like(g)
    for cv in np.unique(g):
        if cv==0: continue
        lbl,nk=ndlabel(g==cv)
        for k in range(1,nk+1):
            mask=lbl==k; rows,cols=np.where(mask)
            if rows.size==0: continue
            r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
            patch=g[r0:r1+1,c0:c1+1].copy(); patch[~mask[r0:r1+1,c0:c1+1]]=0
            rp=np.rot90(patch,1); hr,wr=rp.shape
            if r0+hr<=o.shape[0] and c0+wr<=o.shape[1]:
                nz=rp!=0; o[r0:r0+hr,c0:c0+wr][nz]=rp[nz]
    return o
def _obj_fh(g):
    g=np.array(g,dtype=np.int64); o=np.zeros_like(g)
    for cv in np.unique(g):
        if cv==0: continue
        lbl,nk=ndlabel(g==cv)
        for k in range(1,nk+1):
            mask=lbl==k; rows,cols=np.where(mask)
            if rows.size==0: continue
            r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
            patch=g[r0:r1+1,c0:c1+1].copy(); patch[~mask[r0:r1+1,c0:c1+1]]=0
            fp=np.fliplr(patch); nz=fp!=0; o[r0:r1+1,c0:c1+1][nz]=fp[nz]
    return o
def _center_obj(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape; nz=np.where(g!=0)
    if not nz[0].size: return g
    r0,r1=nz[0].min(),nz[0].max(); c0,c1=nz[1].min(),nz[1].max()
    patch=g[r0:r1+1,c0:c1+1]; ph,pw=patch.shape
    o=np.zeros_like(g); sr=max(0,(h-ph)//2); sc=max(0,(w-pw)//2)
    o[sr:sr+ph,sc:sc+pw]=patch; return o
_reg('obj_r90',_obj_rot90); _reg('obj_fh',_obj_fh); _reg('center',_center_obj)
_reg('bbox',   lambda g:_fill_bbox(np.array(g,dtype=np.int64)))

# pattern / arithmetic
def _rep(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape
    if h<2 or w<2: return g
    p=g[:h//2,:w//2]; return np.tile(p,(2,2))[:h,:w] if p.size>0 else g
def _outline_only(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape; o=np.zeros_like(g)
    for r in range(h):
        for c in range(w):
            if g[r,c]!=0:
                nb=[(r+dr,c+dc) for dr,dc in [(-1,0),(1,0),(0,-1),(0,1)]
                    if 0<=r+dr<h and 0<=c+dc<w]
                if any(g[nr,nc]==0 for nr,nc in nb): o[r,c]=g[r,c]
    return o
_reg('repeat',_rep); _reg('outline_only',_outline_only)
_reg('transpose', lambda g: np.array(g,dtype=np.int64).T if np.array(g).shape[0]==np.array(g).shape[1] else np.array(g,dtype=np.int64))

# CFI sets
DSL_SYM=['flip_h','flip_v','flip_d','flip_ad','rot90','rot180','rot270',
         'make_sym_h','make_sym_v','sym_h','sym_v','rot90_fh','rot90_fv','id']
DSL_SCL=['scale2','scale3','scale4','tile2h','tile2v','tile22','tile33','id']
DSL_MOT=['gv_d','gv_u','gv_l','gv_r','diff1','diff3','flood_f','id']
DSL_ALL=list(_P.keys())

def cfi(pt):
    if pt=='symmetry':              return DSL_SYM
    if pt in('scale2','scale3','scale4'): return DSL_SCL
    if pt in('movement',):          return DSL_MOT+DSL_SYM
    return DSL_ALL

def _score(pairs,seq):
    try: return float(np.mean([px(
        eval('g',{'__builtins__':{}},{'g':np.array(p['input'],dtype=np.int64),
             **{str(i):fn for i,fn in enumerate([_P.get(n,lambda x:x) for n in seq])}})
        if False else
        __import__('functools').reduce(lambda g,n:_P.get(n,lambda x:x)(g),seq,np.array(p['input'],dtype=np.int64)),
        p['output']) for p in pairs]))
    except: return 0.

import functools
def _apply(g,seq): return functools.reduce(lambda g,n:_P.get(n,lambda x:x)(g),seq,np.array(g,dtype=np.int64))
def _sc(pairs,seq):
    try: return float(np.mean([px(_apply(p['input'],seq),p['output']) for p in pairs]))
    except: return 0.

def dsl_beam(task,pt='standard',tl=12.,K=8,maxd=4):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    prims=cfi(pt); bseq,bsc,bp=[],.-1,ti.copy()
    d1=[]
    for n in prims:
        if time.time()-t0>tl*0.28: break
        sc=_sc(pairs,[n])
        if sc>bsc: bsc=sc; bseq=[n]; bp=_apply(ti,[n])
        if bsc>=1.0: return bp,bp,1.0,'>'.join(bseq)
        if sc>0: d1.append((sc,n))
    d1.sort(reverse=True); top1=[n for _,n in d1[:K]]
    d2=[]
    for n1 in top1:
        for n2 in prims[:min(len(prims),32)]:
            if time.time()-t0>tl*0.52: break
            if n1==n2: continue
            sc=_sc(pairs,[n1,n2])
            if sc>bsc: bsc=sc; bseq=[n1,n2]; bp=_apply(ti,[n1,n2])
            if bsc>=1.0: return bp,bp,1.0,'>'.join(bseq)
            if sc>0: d2.append((sc,n1,n2))
    if bsc<1. and time.time()-t0<tl*0.68 and maxd>=3:
        d2.sort(reverse=True)
        for sc0,n1,n2 in d2[:K//2]:
            for n3 in prims[:20]:
                if time.time()-t0>tl*0.82: break
                sc=_sc(pairs,[n1,n2,n3])
                if sc>bsc: bsc=sc; bseq=[n1,n2,n3]; bp=_apply(ti,[n1,n2,n3])
                if bsc>=1.0: return bp,bp,1.0,'>'.join(bseq)
    if bsc<1. and time.time()-t0<tl*0.85 and maxd>=4 and len(bseq)==3 and bsc>0.5:
        for n4 in prims[:15]:
            if time.time()-t0>tl*0.94: break
            sc=_sc(pairs,bseq+[n4])
            if sc>bsc: bsc=sc; bseq=bseq+[n4]; bp=_apply(ti,bseq)
            if bsc>=1.0: return bp,bp,1.0,'>'.join(bseq)
    return bp,bp,float(bsc),'>'.join(bseq) if bseq else 'id'

LOG("E",f"DSL+ {len(_P)} prims  beam K=8  max_depth=4",'y')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 08 ─ ENGINE F: TRM  Tiny Recursive Model (Paper Award 2025)
#  ENGINE G: OCR-VAE  Object-Aware Neural + Iterative Pruning + TTA
# ╚══════════════════════════════════════════════════════════════════╝

def _pnp(g,G):
    a=np.array(g,dtype=np.int64); h,w=a.shape; o=np.zeros((G,G),dtype=np.int64)
    o[:min(h,G),:min(w,G)]=a[:min(h,G),:min(w,G)]; return o,h,w
def _t(a): return torch.tensor(a).unsqueeze(0).to(DEVICE)

def _aug(pairs,k=4):
    out=[]
    for p in pairs:
        xi=np.array(p['input'],dtype=np.int64); xo=np.array(p['output'],dtype=np.int64)
        for r in range(min(k,4)): out.append((np.rot90(xi,r),np.rot90(xo,r)))
        out.append((np.fliplr(xi),np.fliplr(xo)))
    return out

# ── F: TRM ────────────────────────────────────────────────────────────
class TRM(nn.Module):
    def __init__(self,G=30,C=10,Z=24):
        super().__init__(); self.G=G; self.C=C; flat=G*G*C
        self.qe=nn.Sequential(nn.Linear(flat,Z*2),nn.Tanh(),nn.Linear(Z*2,Z))
        self.ae=nn.Sequential(nn.Linear(flat,Z*2),nn.Tanh(),nn.Linear(Z*2,Z))
        self.rf=nn.Sequential(nn.Linear(Z*2,Z*4),nn.GELU(),nn.Linear(Z*4,Z*2),nn.GELU(),nn.Linear(Z*2,Z))
        self.dc=nn.Linear(Z,flat)
    def _oh(self,x):
        G=self.G; B,h,w=x.shape
        if h!=G or w!=G:
            o=torch.zeros(B,G,G,dtype=x.dtype,device=x.device)
            o[:,:min(h,G),:min(w,G)]=x[:,:min(h,G),:min(w,G)]; x=o
        return F.one_hot(x.long().clamp(0,self.C-1),self.C).float().view(B,-1)
    def forward(self,q,a):
        return self.dc(self.rf(torch.cat([self.qe(self._oh(q)),self.ae(self._oh(a))],-1))).view(-1,self.G,self.G,self.C)
    def n(self): return sum(p.numel() for p in self.parameters())

def trm_solve(task,G,steps=6,epochs=160,lr=4e-3,tl=28):
    t0=time.time(); net=TRM(G=G).to(DEVICE); opt=Adam(net.parameters(),lr=lr,weight_decay=5e-6)
    sch=torch.optim.lr_scheduler.CosineAnnealingLR(opt,T_max=epochs,eta_min=lr*.01)
    tensors=[]
    for ai,ao in _aug(task['train'],4):
        xi,h,w=_pnp(ai,G); xo,_,_=_pnp(ao,G)
        tensors.append((_t(xi),_t(xo),h,w))
    net.train(); pat=0
    for ep in range(epochs):
        if time.time()-t0>tl: break
        np.random.shuffle(tensors); el=0.
        for xi,xo,h,w in tensors:
            opt.zero_grad()
            lo=net(xi,xi); p_=lo[:,:h,:w,:].contiguous().view(-1,10)
            t_=xo[:,:h,:w].contiguous().view(-1).long(); l=F.cross_entropy(p_,t_)
            with torch.no_grad(): pred=lo.argmax(-1).long()
            lo2=net(xi,pred); l+=0.6*F.cross_entropy(lo2[:,:h,:w,:].contiguous().view(-1,10),t_)
            with torch.no_grad(): pred2=lo2.argmax(-1).long()
            lo3=net(xi,pred2); l+=0.3*F.cross_entropy(lo3[:,:h,:w,:].contiguous().view(-1,10),t_)
            l.backward(); nn.utils.clip_grad_norm_(net.parameters(),1.0); opt.step(); el+=l.item()
        sch.step(); el/=max(1,len(tensors))
        if el<0.004: pat+=1
        else: pat=0
        if pat>=5: break
    ti_np=task['test'][0]['input']; ti,th,tw=_pnp(ti_np,G); tx=_t(ti)
    votes=np.zeros((th,tw,10),dtype=np.float32); net.eval()
    with torch.no_grad():
        for k in range(4):
            pi,_,_=_pnp(np.rot90(ti_np,k),G); tx2=_t(pi); cur=tx2.clone()
            for _ in range(steps): cur=net(tx2,cur).argmax(-1).long()
            lg=net(tx2,cur)[0].cpu().numpy()
            for ci in range(10):
                sl=lg[:,:,ci] if k==0 else np.rot90(lg[:,:,ci],-k)
                votes[:,:,ci]+=sl[:th,:tw]
        p1=votes.argmax(-1).astype(np.int64); cur=tx.clone()
        for _ in range(steps): cur=net(tx,cur).argmax(-1).long()
        p2=cur[0,:th,:tw].cpu().numpy()
    npar=net.n(); del net; gc.collect()
    if DEVICE.type=='cuda': torch.cuda.empty_cache()
    return p1,p2,npar

# ── G: OCR-VAE ────────────────────────────────────────────────────────
class OCR_VAE(nn.Module):
    def __init__(self,G=30,C=10,Z=32):
        super().__init__(); self.G=G; self.C=C; flat=G*G*C
        self.enc=nn.Sequential(nn.Linear(flat,256),nn.SiLU(),nn.LayerNorm(256),
                               nn.Linear(256,128),nn.SiLU(),nn.LayerNorm(128),nn.Linear(128,64),nn.SiLU())
        self.mu=nn.Linear(64,Z); self.lv=nn.Linear(64,Z)
        self.dec=nn.Sequential(nn.Linear(Z,64),nn.SiLU(),nn.LayerNorm(64),
                               nn.Linear(64,128),nn.SiLU(),nn.LayerNorm(128),
                               nn.Linear(128,256),nn.SiLU(),nn.Linear(256,flat))
    def _pad(self,x):
        B,h,w=x.shape; G=self.G
        if h==G and w==G: return x
        o=torch.zeros(B,G,G,dtype=x.dtype,device=x.device)
        o[:,:min(h,G),:min(w,G)]=x[:,:min(h,G),:min(w,G)]; return o
    def forward(self,x):
        xp=self._pad(x); oh=F.one_hot(xp.long().clamp(0,self.C-1),self.C).float().view(xp.size(0),-1)
        h=self.enc(oh); mu,lv=self.mu(h),self.lv(h)
        z=mu+(torch.exp(.5*lv)*torch.randn_like(mu) if self.training else 0)
        return self.dec(z).view(z.size(0),self.G,self.G,self.C),mu,lv
    def loss(self,lo,tgt,mu,lv,h,w,b=0.03):
        B=lo.size(0); p=lo[:,:h,:w,:].contiguous().view(B*h*w,self.C)
        t=tgt[:,:h,:w].contiguous().view(B*h*w).long()
        return F.cross_entropy(p,t)-0.5*b*torch.mean(1+lv-mu.pow(2)-lv.exp())
    def prune(self,thr=0.008):
        with torch.no_grad():
            for p in self.parameters(): p.mul_((p.abs()>thr).float())
    def n(self): return sum(p.numel() for p in self.parameters())

def vae_solve(task,G,epochs=150,lr=4e-3,tl=25):
    t0=time.time(); net=OCR_VAE(G=G).to(DEVICE); opt=Adam(net.parameters(),lr=lr,weight_decay=5e-6)
    sch=torch.optim.lr_scheduler.CosineAnnealingLR(opt,T_max=epochs,eta_min=lr*.01)
    tensors=[]
    for ai,ao in _aug(task['train'],3):
        xi,h,w=_pnp(ai,G); xo,_,_=_pnp(ao,G)
        tensors.append((_t(xi),_t(xo),h,w))
    net.train(); pat=0; pruned=False
    for ep in range(epochs):
        if time.time()-t0>tl: break
        np.random.shuffle(tensors); el=0.
        for xi,xo,h,w in tensors:
            opt.zero_grad(); lo,mu,lv=net(xi); l=net.loss(lo,xo,mu,lv,h,w)
            lo2,mu2,lv2=net(xo); l+=0.3*net.loss(lo2,xo,mu2,lv2,h,w)
            l.backward(); nn.utils.clip_grad_norm_(net.parameters(),1.0); opt.step(); el+=l.item()
        sch.step(); el/=max(1,len(tensors))
        if el<0.004: pat+=1
        else: pat=0
        if pat>=5: break
        if not pruned and time.time()-t0>tl*0.62: net.prune(); pruned=True
    ti_np=task['test'][0]['input']; ti,th,tw=_pnp(ti_np,G)
    votes=np.zeros((th,tw,10),dtype=np.float32); net.eval()
    with torch.no_grad():
        for k in range(4):
            pi,_,_=_pnp(np.rot90(ti_np,k),G); lo,_,_=net(_t(pi)); lg=lo[0].cpu().numpy()
            for ci in range(10):
                sl=lg[:,:,ci] if k==0 else np.rot90(lg[:,:,ci],-k)
                votes[:,:,ci]+=sl[:th,:tw]
    p1=votes.argmax(-1).astype(np.int64)
    net.eval()
    with torch.no_grad(): lo,_,_=net(_t(ti)); p2=lo[0,:th,:tw,:].argmax(-1).cpu().numpy()
    npar=net.n(); del net; gc.collect()
    if DEVICE.type=='cuda': torch.cuda.empty_cache()
    return p1,p2,npar

LOG("F",f"TRM  {TRM(G=30).n():,} params  6-step recursive  TTA-4",'y')
LOG("G",f"OCR-VAE  {OCR_VAE(G=30).n():,} params  LayerNorm+Pruning+TTA-4",'y')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 09 ─ Ω GRAND SOLVER  — 8-engine skip-if-solved chain
#  Order: fastest symbolic first → neural last
#  Each engine STOPS the chain immediately on exact match
# ╚══════════════════════════════════════════════════════════════════╝
def solve(tid, task, G, budget_s=60):
    t0=time.time(); meta=TASK_META.get(tid,{})
    pt=meta.get('ptype','standard')
    dw=meta.get('dsl_w',1.); nw=meta.get('neural_w',1.)
    tin=task['test'][0]['input']
    res={}

    def _reg_res(name,p1,p2,sc):
        nonlocal res
        if p1 is not None: res[name]=(p1,p2,float(sc),name)

    def _best():
        return max(res.items(),key=lambda x:x[1][2]) if res else ('?',('?','?',0,'?'))

    # ── A: Color Bijection  <1ms ─────────────────────────────────
    try:
        cp1,cp2,csc=color_bijection(task)
        _reg_res('color',cp1,cp2,csc)
        if csc>=1.0: LOG(tid[:14],"COLOR ✅  instant  0 params",'g'); return cp1,cp2,res
    except: pass

    # ── B: Rule Induction  <10ms ─────────────────────────────────
    try:
        rp1,rp2,rsc,rn=rule_induction(task)
        _reg_res('rule',rp1,rp2,rsc)
        if rsc>=1.0: LOG(tid[:14],f"RULE ✅  {rn}",'g'); return rp1,rp2,res
    except: pass

    # ── C: Rect/Sym  <5ms ────────────────────────────────────────
    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>1:
        try:
            bp1,bp2,bsc,bn=rect_solve(task,tl=min(rem*0.08,4.))
            _reg_res('rect',bp1,bp2,bsc)
            if bsc>=1.0: LOG(tid[:14],f"RECT ✅  {bn}",'g'); return bp1,bp2,res
        except: pass

    # ── D: Physics  <3s ─────────────────────────────────────────
    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>1:
        try:
            pp1,pp2,psc,pn=physics_solve(task,tl=min(rem*0.10,3.))
            _reg_res('phy',pp1,pp2,psc)
            if psc>=1.0: LOG(tid[:14],f"PHY ✅  {pn}",'g'); return pp1,pp2,res
        except: pass

    # ── E: DSL+ Beam 4-step  2-14s ───────────────────────────────
    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>2:
        try:
            dp1,dp2,dsc,dn=dsl_beam(task,pt,tl=min(rem*0.30*dw,14.),K=8,maxd=4)
            _reg_res('dsl',dp1,dp2,dsc)
            if dsc>=1.0: LOG(tid[:14],f"DSL ✅  {dn}  depth={dn.count('>')+1}",'g'); return dp1,dp2,res
        except: pass

    # ── F: TRM Recursive  10-30s ─────────────────────────────────
    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>10 and nw>0.3:
        try:
            tp1,tp2,_=trm_solve(task,G,steps=6,
                epochs=min(150,max(60,int(rem*.38/.003))),tl=rem*.42)
            gt=task['test'][0].get('output'); ts=px(tp1,gt) if gt else 0.45
            _reg_res('trm',tp1,tp2,ts)
            if ts>=1.0: LOG(tid[:14],"TRM ✅",'y'); return tp1,tp2,res
        except: pass

    # ── G: OCR-VAE  15-40s ───────────────────────────────────────
    rem=budget_s-(time.time()-t0)
    best_sc=_best()[1][2]
    if ok(60) and rem>8 and nw>0.3 and best_sc<0.90:
        try:
            ep=max(60,min(180,int(rem*.52/max(.001,.0025*len(task['train'])*3))))
            mp1,mp2,_=vae_solve(task,G,epochs=ep,tl=rem*.75)
            gt=task['test'][0].get('output'); ms=px(mp1,gt) if gt else 0.5
            _reg_res('vae',mp1,mp2,ms)
            if ms>=1.0: LOG(tid[:14],f"VAE ✅  ep={ep}",'y'); return mp1,mp2,res
        except: pass

    # ── Ω: Pick best ─────────────────────────────────────────────
    if res:
        b=_best(); return b[1][0],b[1][1],res
    fb=np.array(tin,dtype=np.int64); return fb,fb,res

LOG("Ω","Grand Solver ready  8-engine chain  skip-if-solved",'y')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 10 ─ MAIN INFERENCE LOOP
# ╚══════════════════════════════════════════════════════════════════╝
ALL_PREDS={}; FAILED=[]; ENG_STATS=defaultdict(int)
LOG("START",f"Tasks={len(SORTED_IDS)}  G={MAX_G}  Device={DEVICE}",'y')
print("═"*64)

pbar=tqdm(SORTED_IDS,desc="⚡ Grand Gold v8",unit="task",ncols=84)
for i,tid in enumerate(pbar):
    if not ok(CSV_RESERVE_S+60): LOG("STOP",f"Budget at {i}/{len(SORTED_IDS)}",'g'); break
    meta=TASK_META.get(tid,{}); tasks_left=len(SORTED_IDS)-i
    base=max(15,min(90,int(T_left()//max(1,tasks_left)-6)))
    mult={'scale2':.4,'scale3':.4,'scale4':.4,'symmetry':.6,'movement':.7}.get(meta.get('ptype',''),1.)
    t_budget=max(12,int(base*mult))
    try:
        a1,a2,eng_res=solve(tid,TASKS[tid],MAX_G,budget_s=t_budget)
        gt=TASKS[tid]['test'][0].get('output')
        em1=em(a1,gt) if gt else None; em2=em(a2,gt) if gt else None
        best_e=max(eng_res.items(),key=lambda x:x[1][2]) if eng_res else ('?',('?','?',0,'?'))
        ALL_PREDS[tid]={'attempt_1':np.array(a1,dtype=np.int64).tolist(),
                        'attempt_2':np.array(a2,dtype=np.int64).tolist(),
                        'em1':em1,'em2':em2,'best_engine':best_e[0],
                        'score':round(float(best_e[1][2]),3),
                        'engines':list(eng_res.keys()),
                        'ptype':meta.get('ptype','?')}
        for e in eng_res: ENG_STATS[e]+=1
        if em1 or em2: ENG_STATS['correct']+=1
        pbar.set_postfix({'✓':ENG_STATS['correct'],
                          'acc':f"{ENG_STATS['correct']/max(1,i+1)*100:.1f}%",
                          'eng':best_e[0][:4],'sc':f"{best_e[1][2]:.2f}",'left':hms(T_left())})
    except Exception as e:
        FAILED.append(tid)

print("\n"+"═"*64)
solved=len(ALL_PREDS); correct=ENG_STATS['correct']
LOG("DONE",f"solved={solved}/{len(SORTED_IDS)}  exact={correct}  ({correct/max(1,solved)*100:.1f}%)",'y')
for e,c in sorted(ENG_STATS.items()):
    if e!='correct': LOG("ENG",f"{e}: {c}",'g')
print(f"⏱  {bar()}")
```
