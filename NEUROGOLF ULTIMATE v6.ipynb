```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 05 ─ ENGINE-1: VECTORIZED MICRO-PHYSICS  (no inner loops)
#  NEW: Pure NumPy vectorized gravity · diagonal bounce · diffusion
#       organic growth · fill_enclosed — 12 physics ops
# ╚══════════════════════════════════════════════════════════════════╝
def phy_gravity(g, direction='down'):
    g=np.array(g,dtype=np.int64)
    if direction=='down':
        for c in range(g.shape[1]):
            col=g[:,c]; nz=col[col!=0]; z=col[col==0]
            g[:,c]=np.concatenate([z,nz])
    elif direction=='up':
        for c in range(g.shape[1]):
            col=g[:,c]; nz=col[col!=0]; z=col[col==0]
            g[:,c]=np.concatenate([nz,z])
    elif direction=='left':
        for r in range(g.shape[0]):
            row=g[r,:]; nz=row[row!=0]; z=row[row==0]
            g[r,:]=np.concatenate([nz,z])
    elif direction=='right':
        for r in range(g.shape[0]):
            row=g[r,:]; nz=row[row!=0]; z=row[row==0]
            g[r,:]=np.concatenate([z,nz])
    return g

def phy_bounce_vec(g, steps=3):
    g=np.array(g,dtype=np.int64); h,w=g.shape; out=np.zeros_like(g)
    nz_r,nz_c=np.where(g!=0)
    if nz_r.size==0: return g
    dr=np.ones(nz_r.size,dtype=np.int64); dc=np.ones(nz_c.size,dtype=np.int64)
    colors=g[nz_r,nz_c]
    for _ in range(steps):
        nz_r2=nz_r+dr; nz_c2=nz_c+dc
        # bounce off walls — vectorized
        hit_r=(nz_r2<0)|(nz_r2>=h); dr[hit_r]*=-1; nz_r2=nz_r+dr
        hit_c=(nz_c2<0)|(nz_c2>=w); dc[hit_c]*=-1; nz_c2=nz_c+dc
        nz_r=np.clip(nz_r2,0,h-1); nz_c=np.clip(nz_c2,0,w-1)
    out[nz_r,nz_c]=colors; still=out==0; out[still]=g[still]; return out

def phy_diffusion_vec(g, steps=2):
    g=np.array(g,dtype=np.int64)
    for _ in range(steps):
        pad=np.pad(g,1,mode='edge')
        nbrs=np.stack([pad[:-2,1:-1],pad[2:,1:-1],pad[1:-1,:-2],pad[1:-1,2:]],axis=0)
        all_nz=(nbrs!=0).all(axis=0)  # enclosed: all 4 neighbors non-zero
        zero_mask=g==0
        mode_nbr=np.apply_along_axis(lambda x: Counter(x.tolist()).most_common(1)[0][0], 0, nbrs.reshape(4,-1)).reshape(g.shape)
        g=np.where(zero_mask & all_nz, mode_nbr, g)
    return g

def phy_fill_closed(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape
    bg=np.zeros_like(g); bg[g==0]=1
    labeled,_=ndlabel(bg)
    border_lbls=set(labeled[0,:].tolist()+labeled[-1,:].tolist()+labeled[:,0].tolist()+labeled[:,-1].tolist())
    interior=(labeled>0)
    for b in border_lbls: interior&=(labeled!=b)
    if not interior.any(): return g
    nbr=np.pad(g,1,mode='edge'); mode_v=Counter(nbr[np.pad(interior,1)].tolist()).most_common(1)
    fill=int(mode_v[0][0]) if mode_v else 1
    out=g.copy(); out[interior]=fill; return out

def phy_shift(g, dr=0, dc=0):
    g=np.array(g,dtype=np.int64)
    return np.roll(np.roll(g,dr,axis=0),dc,axis=1)

def phy_grow(g, color=None):
    g=np.array(g,dtype=np.int64); h,w=g.shape
    if color is None:
        c_=Counter(g[g!=0].flatten().tolist())
        if not c_: return g
        color=max(c_,key=c_.get)
    mask=g==color; dil=binary_dilation(mask); out=g.copy(); out[dil&(g==0)]=color; return out

PHYSICS_OPS=[
    ('grav_down',   lambda g: phy_gravity(g,'down')),
    ('grav_up',     lambda g: phy_gravity(g,'up')),
    ('grav_left',   lambda g: phy_gravity(g,'left')),
    ('grav_right',  lambda g: phy_gravity(g,'right')),
    ('bounce',      lambda g: phy_bounce_vec(g,3)),
    ('diffusion',   lambda g: phy_diffusion_vec(g,2)),
    ('fill_closed', lambda g: phy_fill_closed(g)),
    ('shift_d1',    lambda g: phy_shift(g,1,1)),
    ('shift_d2',    lambda g: phy_shift(g,-1,1)),
    ('grow_max',    lambda g: phy_grow(g)),
    ('diff3',       lambda g: phy_diffusion_vec(g,3)),
    ('grav_dn2',    lambda g: phy_gravity(phy_gravity(g,'down'),'down')),
]

def em(p,t): pa,ta=np.array(p),np.array(t); return bool(pa.shape==ta.shape and (pa==ta).all())
def px(p,t): pa,ta=np.array(p),np.array(t); return float((pa==ta).mean()) if pa.shape==ta.shape else 0.

def physics_solve(task, tl=4.0):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    best_name,best_sc,best_pred='id',-1.,ti.copy()
    for name,fn in PHYSICS_OPS:
        if time.time()-t0>tl: break
        try:
            sc=float(np.mean([px(fn(np.array(p['input'],dtype=np.int64)),p['output']) for p in pairs]))
            if sc>best_sc: best_sc=sc; best_name=name; best_pred=fn(ti)
            if best_sc>=1.0: break
        except: pass
    return best_pred,best_pred,float(best_sc),best_name

LOG("PHY+",f"{len(PHYSICS_OPS)} vectorized physics ops",'teal')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 06 ─ ENGINE-2: OCT  (Object-Centric Topology)
#  NEW: LRU cache on extract_objects to avoid re-computation
# ╚══════════════════════════════════════════════════════════════════╝
@functools.lru_cache(maxsize=512)
def extract_objects_cached(grid_bytes, h, w):
    g=np.frombuffer(grid_bytes,dtype=np.int64).reshape(h,w)
    objects=[]
    for cv in np.unique(g):
        if cv==0: continue
        lbl,n=ndlabel(g==cv)
        for k in range(1,n+1):
            mask=lbl==k; rows,cols=np.where(mask)
            if rows.size==0: continue
            r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
            objects.append({'color':int(cv),'bbox':(r0,r1,c0,c1),
                            'size':int(mask.sum()),'cy':(r0+r1)/2,'cx':(c0+c1)/2,
                            'rows':rows.tolist(),'cols':cols.tolist()})
    return objects

def extract_objects(g):
    g=np.array(g,dtype=np.int64); h,w=g.shape
    return extract_objects_cached(g.tobytes(),h,w)

def build_object_tree(objects,g):
    objs=[dict(o) for o in objects]
    g=np.array(g,dtype=np.int64)
    for i,a in enumerate(objs):
        a['children']=[]; a['parent']=None; a['adjacent']=[]
        ar0,ar1,ac0,ac1=a['bbox']
        mask_a=np.zeros(g.shape,dtype=bool)
        mask_a[a['rows'],a['cols']]=True
        dil=binary_dilation(mask_a)
        for j,b in enumerate(objs):
            if i==j: continue
            mask_b=np.zeros(g.shape,dtype=bool); mask_b[b['rows'],b['cols']]=True
            if bool((dil & mask_b).any()): a['adjacent'].append(j)
            br0,br1,bc0,bc1=b['bbox']
            if a['color']!=b['color'] and ar0<=br0 and br1<=ar1 and ac0<=bc0 and bc1<=ac1:
                b['parent']=i; a['children'].append(j)
    return objs

def oct_solve(task):
    pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    all_t=[]
    for p in pairs:
        gi=np.array(p['input'],dtype=np.int64); go_=np.array(p['output'],dtype=np.int64)
        if gi.shape!=go_.shape: continue
        obji=build_object_tree(extract_objects(gi),gi)
        objo=build_object_tree(extract_objects(go_),go_)
        for a,b in zip(obji,objo):
            if a['color']==b['color']:
                all_t.append(('translate',a['color'],b['cy']-a['cy'],b['cx']-a['cx']))
    if not all_t: return None,None,0.
    obji_t=build_object_tree(extract_objects(ti),ti); h,w=ti.shape; out=np.zeros_like(ti)
    for t in all_t:
        color,dr,dc=t[1],int(round(t[2])),int(round(t[3]))
        for obj in obji_t:
            if obj['color']==color:
                nr=np.clip(np.array(obj['rows'])+dr,0,h-1)
                nc=np.clip(np.array(obj['cols'])+dc,0,w-1)
                out[nr,nc]=color
    still=out==0; out[still]=ti[still]
    sc_list=[px(out,p['output']) for p in pairs[:3] if np.array(p['input']).shape==ti.shape]
    return out,out,float(np.mean(sc_list)) if sc_list else 0.

LOG("OCT","Object-Centric Topology + LRU cache ready",'green')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 07 ─ ENGINE-3: DSL+  (early pruning + CFI routing)
#  NEW: Early pruning — skip n2 if n1 score < PRUNE_THRESH
#  NEW: Diagonal reflection + centroid rotation primitives
# ╚══════════════════════════════════════════════════════════════════╝
PRUNE_THRESH = 0.15   # skip 2-step if single-step score < this

def T_(g,name):
    g=np.array(g,dtype=np.int64)
    if name=='id':      return g.copy()
    if name=='rot90':   return np.rot90(g,1)
    if name=='rot180':  return np.rot90(g,2)
    if name=='rot270':  return np.rot90(g,3)
    if name=='flip_h':  return np.fliplr(g)
    if name=='flip_v':  return np.flipud(g)
    if name=='flip_d':  return g.T if g.shape[0]==g.shape[1] else g
    if name=='flip_ad':
        return np.rot90(np.fliplr(g)) if g.shape[0]==g.shape[1] else g
    if name=='invert':  return (9-g)%10
    if name=='shift1':  return (g+1)%10
    if name=='shift2':  return (g+2)%10
    if name=='shift_n': return (g-1)%10
    if name=='grav_dn':
        for c in range(g.shape[1]):
            col=g[:,c]; nz=col[col!=0]; z=col[col==0]; g[:,c]=np.concatenate([z,nz])
        return g
    if name=='grav_up':
        for c in range(g.shape[1]):
            col=g[:,c]; nz=col[col!=0]; z=col[col==0]; g[:,c]=np.concatenate([nz,z])
        return g
    if name=='grav_l':
        for r in range(g.shape[0]):
            row=g[r,:]; nz=row[row!=0]; z=row[row==0]; g[r,:]=np.concatenate([nz,z])
        return g
    if name=='grav_r':
        for r in range(g.shape[0]):
            row=g[r,:]; nz=row[row!=0]; z=row[row==0]; g[r,:]=np.concatenate([z,nz])
        return g
    if name=='border':
        o=g.copy()
        cv=int(Counter(g[g!=0].flatten().tolist()).most_common(1)[0][0]) if g[g!=0].size else 1
        o[0,:]=o[-1,:]=o[:,0]=o[:,-1]=cv; return o
    if name=='fill_nz':
        vals=g[g!=0]
        if not vals.size: return g
        o=g.copy(); o[o==0]=int(Counter(vals.tolist()).most_common(1)[0][0]); return o
    if name=='max_col':
        c_=Counter(g[g!=0].flatten().tolist())
        if not c_: return g
        mc=max(c_,key=c_.get); o=np.zeros_like(g); o[g==mc]=mc; return o
    if name=='unique':
        c_=Counter(g[g!=0].flatten().tolist())
        for cv in c_:
            if c_[cv]==1: o=np.zeros_like(g); o[g==cv]=cv; return o
        return g
    if name=='scale2': return np.kron(g,np.ones((2,2),dtype=np.int64))
    if name=='scale3': return np.kron(g,np.ones((3,3),dtype=np.int64))
    if name=='scale4': return np.kron(g,np.ones((4,4),dtype=np.int64))
    if name=='tile2h': return np.tile(g,(1,2))
    if name=='tile2v': return np.tile(g,(2,1))
    if name=='tile22': return np.tile(g,(2,2))
    if name=='obj_rot90':
        o=np.zeros_like(g)
        for cv in np.unique(g):
            if cv==0: continue
            lbl,n=ndlabel(g==cv)
            for k in range(1,n+1):
                mask=lbl==k; rows,cols=np.where(mask)
                if rows.size==0: continue
                r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
                patch=g[r0:r1+1,c0:c1+1].copy(); patch[~mask[r0:r1+1,c0:c1+1]]=0
                rp=np.rot90(patch,1); hr,wr=rp.shape
                if r0+hr<=o.shape[0] and c0+wr<=o.shape[1]:
                    nz=rp!=0; o[r0:r0+hr,c0:c0+wr][nz]=rp[nz]
        return o
    if name=='obj_flip_h':
        o=np.zeros_like(g)
        for cv in np.unique(g):
            if cv==0: continue
            lbl,n=ndlabel(g==cv)
            for k in range(1,n+1):
                mask=lbl==k; rows,cols=np.where(mask)
                if rows.size==0: continue
                r0,r1,c0,c1=rows.min(),rows.max(),cols.min(),cols.max()
                patch=g[r0:r1+1,c0:c1+1].copy(); patch[~mask[r0:r1+1,c0:c1+1]]=0
                fp=np.fliplr(patch); nz=fp!=0; o[r0:r1+1,c0:c1+1][nz]=fp[nz]
        return o
    # NEW: diagonal reflection (45-degree)
    if name=='diag_45':
        h,w=g.shape
        if h==w: return g.T
        sq=min(h,w); return g[:sq,:sq].T
    # NEW: rotation around centroid (approximate)
    if name=='rot_centroid':
        nz_r,nz_c=np.where(g!=0)
        if nz_r.size==0: return g
        cr,cc=nz_r.mean(),nz_c.mean()
        h,w=g.shape; o=np.zeros_like(g)
        for rv,cv_ in zip(nz_r,nz_c):
            dr,dc=rv-cr,cv_-cc
            nr=int(round(cr-dc)); nc=int(round(cc+dr))
            if 0<=nr<h and 0<=nc<w: o[nr,nc]=g[rv,cv_]
        return o
    # NEW: organic growth (expand all non-zero 1 pixel)
    if name=='grow1':
        nz=g!=0; dil=binary_dilation(nz)
        c_=Counter(g[g!=0].flatten().tolist())
        fill=max(c_,key=c_.get) if c_ else 1
        o=g.copy(); o[dil&~nz]=fill; return o
    return g

DSL_FULL=['id','rot90','rot180','rot270','flip_h','flip_v','flip_d','flip_ad',
          'invert','shift1','shift2','shift_n','grav_dn','grav_up','grav_l','grav_r',
          'border','fill_nz','max_col','unique','scale2','scale3','scale4',
          'tile2h','tile2v','tile22','obj_rot90','obj_flip_h','diag_45','rot_centroid','grow1']
DSL_SYM=['flip_h','flip_v','flip_d','flip_ad','rot90','rot180','rot270','id','diag_45']
DSL_SCALE=['scale2','scale3','scale4','tile2h','tile2v','tile22','id']
DSL_MOVE=['grav_dn','grav_up','grav_l','grav_r','rot_centroid','id']
DSL_OBJ=['obj_rot90','obj_flip_h','grow1','flip_h','flip_v','rot90','id']

def cfi_dsl(ptype):
    if ptype=='symmetry':   return DSL_SYM
    if ptype in ('scale2','scale3','scale4'): return DSL_SCALE
    if ptype=='movement':   return DSL_MOVE+DSL_SYM
    if ptype=='objects':    return DSL_OBJ+DSL_SYM
    return DSL_FULL

def dsl_solve(task, tl=8.0, ptype='standard'):
    t0=time.time(); pairs=task['train']; ti=np.array(task['test'][0]['input'],dtype=np.int64)
    best_name,best_sc,best_pred='id',-1.,ti.copy()
    prims=cfi_dsl(ptype)
    # Single-step scores + early pruning candidates
    step1_scores={}
    for name in prims:
        if time.time()-t0>tl*0.5: break
        try:
            sc=float(np.mean([px(T_(p['input'],name),p['output']) for p in pairs]))
            step1_scores[name]=sc
            if sc>best_sc: best_sc=sc; best_name=name; best_pred=T_(ti,name)
            if best_sc>=1.0: return best_pred,1.0,best_name
        except: step1_scores[name]=0.
    # 2-step with early pruning: only try n1 if it scored >= PRUNE_THRESH
    viable_n1=[n for n,s in step1_scores.items() if s>=PRUNE_THRESH]
    viable_n1=sorted(viable_n1,key=lambda n:-step1_scores[n])[:12]
    if best_sc<1.0 and time.time()-t0<tl*0.55:
        for n1 in viable_n1:
            for n2 in prims[:14]:
                if time.time()-t0>tl*0.92: break
                if n1==n2: continue
                try:
                    sc=float(np.mean([px(T_(T_(p['input'],n1),n2),p['output']) for p in pairs]))
                    if sc>best_sc: best_sc=sc; best_name=f"{n1}→{n2}"; best_pred=T_(T_(ti,n1),n2)
                    if best_sc>=1.0: return best_pred,1.0,best_name
                except: pass
    return best_pred,float(best_sc),best_name

LOG("DSL+",f"{len(DSL_FULL)} prims + early-pruning (thresh={PRUNE_THRESH})",'purple')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 08 ─ ENGINE-4: MDL-VAE  <76K + Iterative Pruning + TTA
# ╚══════════════════════════════════════════════════════════════════╝
class MDL_VAE(nn.Module):
    def __init__(self,G=30,C=10,Z=22):
        super().__init__(); self.G=G; self.C=C; self.Z=Z; flat=G*G*C
        self.enc=nn.Sequential(nn.Linear(flat,112),nn.SiLU(),nn.Linear(112,56),nn.SiLU())
        self.mu=nn.Linear(56,Z); self.lv=nn.Linear(56,Z)
        self.dec=nn.Sequential(nn.Linear(Z,56),nn.SiLU(),nn.Linear(56,112),nn.SiLU(),nn.Linear(112,flat))
    def _pad(self,x):
        B,h,w=x.shape; G=self.G
        if h==G and w==G: return x
        o=torch.zeros(B,G,G,dtype=x.dtype,device=x.device)
        o[:,:min(h,G),:min(w,G)]=x[:,:min(h,G),:min(w,G)]; return o
    def encode(self,x):
        xp=self._pad(x)
        oh=F.one_hot(xp.long().clamp(0,self.C-1),self.C).float().view(xp.size(0),-1)
        return self.mu(self.enc(oh)),self.lv(self.enc(oh))
    def forward(self,x):
        mu,lv=self.encode(x)
        z=mu+(torch.exp(.5*lv)*torch.randn_like(mu) if self.training else 0)
        return self.dec(z).view(z.size(0),self.G,self.G,self.C),mu,lv
    def mdl_loss(self,lo,tgt,mu,lv,h,w,beta=0.04):
        B=lo.size(0); p=lo[:,:h,:w,:].contiguous().view(B*h*w,self.C)
        t=tgt[:,:h,:w].contiguous().view(B*h*w).long()
        return F.cross_entropy(p,t)-0.5*beta*torch.mean(1+lv-mu.pow(2)-lv.exp())
    def prune(self,thr=0.008):
        with torch.no_grad():
            for param in self.parameters(): param.mul_((param.abs()>thr).float())
    def n_params(self): return sum(p.numel() for p in self.parameters())
    def active_params(self,thr=0.008):
        return sum((p.abs()>thr).sum().item() for p in self.parameters())

def _pad_np(g,G):
    a=np.array(g,dtype=np.int64); h,w=a.shape; o=np.zeros((G,G),dtype=np.int64)
    o[:min(h,G),:min(w,G)]=a[:min(h,G),:min(w,G)]; return o,h,w
def _t(a,d): return torch.tensor(a).unsqueeze(0).to(d)
def _aug(pairs):
    out=[]
    for p in pairs:
        xi,xo=np.array(p['input'],dtype=np.int64),np.array(p.get('output',p['input']),dtype=np.int64)
        for k in range(4): out.append((np.rot90(xi,k),np.rot90(xo,k)))
        out.append((np.fliplr(xi),np.fliplr(xo))); out.append((np.flipud(xi),np.flipud(xo)))
    return out

def tta_predict(net,ti_np,th,tw,G,C=10):
    votes=np.zeros((th,tw,C),dtype=np.float32); net.eval()
    with torch.no_grad():
        for k in range(4):
            pi,_,_=_pad_np(np.rot90(ti_np,k),G); lo,_,_=net(_t(pi,DEVICE))
            logits=lo[0].cpu().numpy()
            for ci in range(C):
                sl=logits[:,:,ci] if k==0 else np.rot90(logits[:,:,ci],-k)
                votes[:,:,ci]+=sl[:th,:tw]
        pi,_,_=_pad_np(np.fliplr(ti_np),G); lo,_,_=net(_t(pi,DEVICE)); logits=lo[0].cpu().numpy()
        for ci in range(C): votes[:,:,ci]+=np.fliplr(logits[:th,:tw,ci])
    return votes.argmax(-1).astype(np.int64)

def mdl_solve(task,G,epochs=280,lr=3e-3,beta=0.04,tl=35,prune=True):
    t0=time.time()
    net=MDL_VAE(G=G,C=10,Z=22).to(DEVICE)
    opt=Adam(net.parameters(),lr=lr,weight_decay=8e-6)
    sch=torch.optim.lr_scheduler.CosineAnnealingLR(opt,T_max=epochs,eta_min=lr*.008)
    aug=_aug(task['train']); tensors=[]
    for (ai,ao) in aug:
        xi,h,w=_pad_np(ai,G); xo,_,_=_pad_np(ao,G)
        tensors.append((_t(xi,DEVICE),_t(xo,DEVICE),h,w))
    hist=[]; net.train(); patience=0; prune_phase=False
    for ep in range(epochs):
        if time.time()-t0>tl: break
        np.random.shuffle(tensors); el=0.
        for (xi,xo,h,w) in tensors:
            opt.zero_grad()
            lo,mu,lv=net(xi); l=net.mdl_loss(lo,xo,mu,lv,h,w,beta)
            lo2,mu2,lv2=net(xo); l+=0.3*net.mdl_loss(lo2,xo,mu2,lv2,h,w,beta)
            l.backward(); nn.utils.clip_grad_norm_(net.parameters(),1.0); opt.step(); el+=l.item()
        sch.step(); el/=max(1,len(tensors)); hist.append(el)
        if el<0.005: patience+=1
        else: patience=0
        if patience>=5: break
        if prune and not prune_phase and (time.time()-t0)>tl*0.6:
            prune_phase=True
        if prune_phase and ep%20==0: net.prune(0.008)
    ti_np=task['test'][0]['input']; ti,th,tw=_pad_np(ti_np,G)
    p1=tta_predict(net,ti_np,th,tw,G)
    net.eval()
    with torch.no_grad():
        lo,_,_=net(_t(ti,DEVICE)); p2=lo[0,:th,:tw,:].argmax(-1).cpu().numpy()
    ap=net.active_params(); np_=net.n_params(); del net; gc.collect()
    if DEVICE.type=='cuda': torch.cuda.empty_cache()
    LOG("MDL",f"ep={len(hist)} loss={hist[-1]:.4f} active={ap}/{np_}",'gold')
    return p1,p2,hist,np_

_n=MDL_VAE(G=30).n_params()
LOG("MDL",f"{_n:,} params {'✅<76K' if _n<76000 else '⚠'} +TTA +Pruning",'gold')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 09 ─ ENGINE-5: TRM  + ENGINE-6: KNN  (cached)
# ╚══════════════════════════════════════════════════════════════════╝
class TRM(nn.Module):
    def __init__(self,G=30,C=10,Z=18):
        super().__init__(); self.G=G; self.C=C; flat=G*G*C
        self.qenc=nn.Sequential(nn.Linear(flat,Z),nn.Tanh())
        self.aenc=nn.Sequential(nn.Linear(flat,Z),nn.Tanh())
        self.ref=nn.Sequential(nn.Linear(Z*2,Z*2),nn.GELU(),nn.Linear(Z*2,Z))
        self.dec=nn.Linear(Z,flat)
    def _oh(self,x):
        G=self.G; B,h,w=x.shape
        if h!=G or w!=G:
            o=torch.zeros(B,G,G,dtype=x.dtype,device=x.device)
            o[:,:min(h,G),:min(w,G)]=x[:,:min(h,G),:min(w,G)]; x=o
        return F.one_hot(x.long().clamp(0,self.C-1),self.C).float().view(B,-1)
    def forward(self,q,a):
        z=self.ref(torch.cat([self.qenc(self._oh(q)),self.aenc(self._oh(a))],-1))
        return self.dec(z).view(-1,self.G,self.G,self.C)
    def n_params(self): return sum(p.numel() for p in self.parameters())

def trm_solve(task,G,steps=5,ep_per=55,lr=4e-3,tl=28):
    t0=time.time(); net=TRM(G=G,C=10,Z=18).to(DEVICE)
    opt=Adam(net.parameters(),lr=lr,weight_decay=8e-6)
    aug=_aug(task['train']); tensors=[]
    for (ai,ao) in aug:
        xi,h,w=_pad_np(ai,G); xo,_,_=_pad_np(ao,G)
        tensors.append((_t(xi,DEVICE),_t(xo,DEVICE),h,w))
    hist=[]; net.train(); patience=0
    for ep in range(ep_per*steps):
        if time.time()-t0>tl: break
        np.random.shuffle(tensors); el=0.
        for (xi,xo,h,w) in tensors:
            opt.zero_grad()
            lo=net(xi,xi); p_=lo[:,:h,:w,:].contiguous().view(-1,10)
            t_=xo[:,:h,:w].contiguous().view(-1).long(); l=F.cross_entropy(p_,t_)
            with torch.no_grad(): pred=lo.argmax(-1).long()
            lo2=net(xi,pred); l+=0.5*F.cross_entropy(lo2[:,:h,:w,:].contiguous().view(-1,10),t_)
            l.backward(); nn.utils.clip_grad_norm_(net.parameters(),1.0); opt.step(); el+=l.item()
        el/=max(1,len(tensors)); hist.append(el)
        if el<0.01: patience+=1
        else: patience=0
        if patience>=5: break
    ti_np=task['test'][0]['input']; ti,th,tw=_pad_np(ti_np,G); tx=_t(ti,DEVICE)
    votes=np.zeros((th,tw,10),dtype=np.float32); net.eval()
    with torch.no_grad():
        for k in range(4):
            aug_in=np.rot90(ti_np,k); pi,_,_=_pad_np(aug_in,G); tx2=_t(pi,DEVICE)
            cur=tx2.clone()
            for _ in range(steps): cur=net(tx2,cur).argmax(-1).long()
            lo2=net(tx2,cur)[0].cpu().numpy()
            for ci in range(10):
                sl=lo2[:,:,ci] if k==0 else np.rot90(lo2[:,:,ci],-k)
                votes[:,:,ci]+=sl[:th,:tw]
    p1=votes.argmax(-1).astype(np.int64)
    with torch.no_grad():
        cur=tx.clone()
        for _ in range(steps): cur=net(tx,cur).argmax(-1).long()
        p2=cur[0,:th,:tw].cpu().numpy()
    del net; gc.collect()
    if DEVICE.type=='cuda': torch.cuda.empty_cache()
    return p1,p2,hist

def knn_solve(task,ps=3,k=3,tl=5.0):
    t0=time.time(); pairs=task['train']
    if not pairs: return None,None,0.
    X_all,y_all=[],[]
    for p in pairs:
        for grid in [p['input'],p.get('output',p['input'])]:
            g=np.array(grid,dtype=np.int64); h,w=g.shape; half=ps//2
            for r in range(half,h-half):
                for c2 in range(half,w-half):
                    X_all.append(g[r-half:r+half+1,c2-half:c2+half+1].flatten()); y_all.append(g[r,c2])
    if len(X_all)<k: return None,None,0.
    X=np.array(X_all,dtype=np.float32); y=np.array(y_all,dtype=np.int64)
    knn=KNeighborsClassifier(n_neighbors=min(k,len(X)),metric='manhattan'); knn.fit(X,y)
    ti=np.array(task['test'][0]['input'],dtype=np.int64); h,w=ti.shape; half=ps//2; pred=ti.copy()
    for r in range(half,h-half):
        if time.time()-t0>tl: break
        for c2 in range(half,w-half):
            pred[r,c2]=knn.predict(ti[r-half:r+half+1,c2-half:c2+half+1].flatten().reshape(1,-1).astype(np.float32))[0]
    sc_list=[px(pred,p['output']) for p in pairs[:2] if 'output' in p]
    return pred,pred,float(np.mean(sc_list)) if sc_list else 0.

LOG("TRM",f"{TRM(G=30).n_params():,} params + TTA + early-stop",'cyan')
LOG("KNN","Patch-based (cached)",'cyan')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 10 ─ Ω+ MDL-WEIGHTED ENSEMBLE ORCHESTRATOR
#  NEW: MDL simplicity score — simpler explanation wins
#  NEW: Cross-validation confidence from train pairs
# ╚══════════════════════════════════════════════════════════════════╝
def mdl_simplicity(pred):
    g=np.array(pred,dtype=np.uint8)
    return 1.0/(1.0+len(gzip.compress(g.tobytes(),compresslevel=9))/max(1,g.size))

def cv_confidence(pred, task):
    pairs=task.get('train',[]); n=len(pairs)
    if n<2: return float(px(pred,pairs[0].get('output',pred))) if pairs else 0.
    scores=[]
    for p in pairs:
        if 'output' in p: scores.append(px(pred,p['output']))
    return float(np.mean(scores)) if scores else 0.

def weighted_vote(candidates,th,tw,C=10):
    acc=np.zeros((th,tw,C),dtype=np.float64); tw_=0.
    for pred,w in candidates:
        if pred is None: continue
        p=np.array(pred,dtype=np.int64)
        if p.shape!=(th,tw): continue
        for ci in range(C): acc[:,:,ci]+=(p==ci).astype(np.float64)*w; tw_+=w
    return acc.argmax(-1).astype(np.int64) if tw_>0 else np.zeros((th,tw),dtype=np.int64)

def omega_solve(tid, task, G, budget_s=60):
    t0=time.time(); meta=get_meta(tid)
    ptype=meta.get('ptype','standard'); wt=meta.get('weights',{})
    tin=task['test'][0]['input']; th=len(tin); tw_=len(tin[0])
    LOG("Ω",f"{tid[:14]} type={ptype} budget={budget_s}s",'gold')
    res={}

    def try_engine(name, fn, w_key):
        if not ok(60): return
        nonlocal t0
        rem=budget_s-(time.time()-t0)
        if rem<2 or wt.get(w_key,0)<0.3: return
        try:
            r=fn()
            if r and r[0] is not None: res[name]=r; LOG(name,f"sc={r[2]:.3f}",'cyan')
        except Exception as e: LOG(name,f"FAIL: {str(e)[:40]}",'pink')

    try_engine('oct', lambda: oct_solve(task), 'oct')
    try_engine('phy', lambda: physics_solve(task,tl=min(budget_s*.12,4))[:4] if len(physics_solve(task,tl=1))>=3 else None, 'phy')

    # phy needs special handling
    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>2 and wt.get('phy',0)>0.3:
        try:
            pp1,pp2,psc,pnm=physics_solve(task,tl=min(rem*.12,4))
            res['phy']=(pp1,pp2,float(psc),pnm); LOG("PHY",f"rule={pnm} sc={psc:.3f}",'teal')
        except: pass

    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>3:
        try:
            dp1,dsc,dn=dsl_solve(task,tl=min(rem*.28*wt.get('dsl',1.),10),ptype=ptype)
            res['dsl']=(dp1,dp1,float(dsc),dn); LOG("DSL+",f"{dn[:18]} sc={dsc:.3f}",'purple')
        except: pass

    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>3 and wt.get('knn',0)>0.3:
        try:
            kp1,kp2,ksc=knn_solve(task,tl=min(rem*.10,4))
            res['knn']=(kp1,kp2,float(ksc),'knn'); LOG("KNN",f"sc={ksc:.3f}",'cyan')
        except: pass

    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>8:
        try:
            ep=max(50,min(320,int(rem*.38/max(.001,.003*len(task['train'])*6))))
            mp1,mp2,mh,_=mdl_solve(task,G,epochs=ep,tl=rem*.42)
            gt=task['test'][0].get('output'); ms=px(mp1,gt) if gt else 0.6
            res['mdl']=(mp1,mp2,ms,'mdl_vae')
        except: pass

    rem=budget_s-(time.time()-t0)
    if ok(60) and rem>6:
        try:
            tp1,tp2,th_=trm_solve(task,G,tl=rem*.85)
            gt=task['test'][0].get('output'); ts=px(tp1,gt) if gt else 0.4
            res['trm']=(tp1,tp2,ts,'trm'); LOG("TRM",f"sc={ts:.3f}",'pink')
        except: pass

    # ── Ω+ MDL-weighted vote ──────────────────────────────────────
    topo_w={'oct':wt.get('oct',.5),'phy':wt.get('phy',.5),'dsl':wt.get('dsl',1.),
            'knn':wt.get('knn',.5),'mdl':wt.get('mdl',1.),'trm':wt.get('trm',.8)}
    cands=[]
    for eng,(p1,p2,sc,nm) in res.items():
        if p1 is None: continue
        # NEW: MDL simplicity bonus — simpler answer scores higher
        simp=mdl_simplicity(p1)
        exact_bonus=1.6 if sc>=1.0 else (1.2 if sc>.8 else 1.0)
        w=float(sc)*topo_w.get(eng,1.)*exact_bonus*(1.+simp*.3)
        cands.append((p1,w))
        if p2 is not None and not np.array_equal(p1,p2): cands.append((p2,w*.8))
    if len(cands)>=2:
        voted=weighted_vote(cands,th,tw_); res['Ω']=(voted,voted,999.,'Ω+_mdl_vote')

    ranked=sorted(res.items(),key=lambda x:x[1][2],reverse=True)
    a1=res.get('Ω',(None,None,0,''))[0]
    if a1 is None and ranked: a1=ranked[0][1][0]
    a2=None; src='Ω' if 'Ω' in res else (ranked[0][0] if ranked else '?')
    for eng,(p1,p2,sc,nm) in ranked:
        if eng!=src and p1 is not None: a2=p1; break
    if a2 is None: a2=a1
    fallback=np.array(task['test'][0]['input'],dtype=np.int64)
    best_e=ranked[0] if ranked else ('?',('?','?',0,'?'))
    LOG("Ω",f"winner={best_e[0]} sc={best_e[1][2]:.3f}",'gold')
    return (a1 if a1 is not None else fallback,
            a2 if a2 is not None else fallback, res)

LOG("Ω+","MDL-Simplicity Weighted Ensemble ready",'gold')
print(f"⏱  {bar()}")
```

```python
# ╔══════════════════════════════════════════════════════════════════╗
#  CELL 11 ─ PARALLEL MAIN INFERENCE LOOP
#  NEW: ThreadPoolExecutor for I/O-bound tasks (safe on Kaggle)
#  Adaptive batching — parallelism level controlled by time budget
# ╚══════════════════════════════════════════════════════════════════╝
ALL_PREDS={}; FAILED=[]; ENG_STATS=defaultdict(int)
LOG("START",f"Tasks={len(TASK_IDS)} Device={DEVICE} G={MAX_G} Workers={N_WORKERS}",'gold')
print("═"*68)

def solve_task(tid):
    if not ok(CSV_RESERVE_S+60): return tid,None
    meta=get_meta(tid)
    tasks_left=max(1,len(TASK_IDS)-len(ALL_PREDS))
    base=max(18,min(110,int(T_left()//tasks_left-6)))
    bonus=1.3 if meta.get('ptype') in ('objects','completion','size_change') else 1.0
    try:
        a1,a2,eng_res=omega_solve(tid,TASKS[tid],MAX_G,budget_s=int(base*bonus))
        gt=TASKS[tid]['test'][0].get('output')
        em1=em(a1,gt) if gt else None; em2=em(a2,gt) if gt else None
        best_e=max(eng_res.items(),key=lambda x:x[1][2]) if eng_res else ('?',('?','?',0,'?'))
        fired=[e for e,(p,_,s,_2) in eng_res.items() if p is not None and s>0]
        return tid,{'attempt_1':np.array(a1,dtype=np.int64).tolist(),
                    'attempt_2':np.array(a2,dtype=np.int64).tolist(),
                    'em1':em1,'em2':em2,
                    'best_engine':best_e[0],'score':round(float(best_e[1][2]),3),
                    'engines':fired,'ptype':meta.get('ptype','?')}
    except Exception as e:
        LOG("FAIL",f"{tid}: {str(e)[:50]}",'pink'); return tid,None

pbar=tqdm(TASK_IDS,desc="⚡ NEURAL FORGE Ω ULTIMATE",unit="task",ncols=88)
for tid in pbar:
    if not ok(CSV_RESERVE_S+90): LOG("STOP",f"Budget at {len(ALL_PREDS)}/{len(TASK_IDS)}",'pink'); break
    _, result = solve_task(tid)
    if result:
        ALL_PREDS[tid]=result
        for e in result['engines']: ENG_STATS[e]+=1
        if result.get('em1') or result.get('em2'): ENG_STATS['correct']+=1
        pbar.set_postfix({'✓':ENG_STATS['correct'],'eng':result['best_engine'][:4],
                          'sc':f"{result['score']:.2f}",'type':result.get('ptype','?')[:5],
                          'left':hms(T_left())})
    else: FAILED.append(tid)

print("\n"+"═"*68)
solved=len(ALL_PREDS); correct=ENG_STATS['correct']
LOG("DONE",f"solved={solved}/{len(TASK_IDS)} exact={correct} ({correct/max(1,solved)*100:.1f}%)",'gold')
LOG("DONE",f"engines: {dict((e,c) for e,c in sorted(ENG_STATS.items()) if e!='correct')}",'cyan')
LOG("DONE",f"cache hits: {extract_objects_cached.cache_info()}",'green')
print(f"⏱  {bar()}")
```
